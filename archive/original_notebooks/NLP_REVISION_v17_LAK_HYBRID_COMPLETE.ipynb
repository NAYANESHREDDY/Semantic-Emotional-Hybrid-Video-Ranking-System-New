````python
# ============================================================
# PHASE 9A — MULTI-DOMAIN RETRIEVAL VALIDATION
# ============================================================
# Purpose:
#   Validate the framework on multiple educational domains.
#
# IMPORTANT:
#   - Do NOT modify Phase 1–8 cells.
#   - Do NOT overwrite original youtube_videos.csv,
#     youtube_comments.csv, or Phase 5–8 outputs.
#   - All revision data is saved under /content/revision_phase9a/
# ============================================================

!pip install -q google-api-python-client

from googleapiclient.discovery import build
from google.colab import userdata

import pandas as pd
import os
import re
import time
import json

# ------------------------------------------------------------
# 1. LOAD YOUTUBE API KEY FROM COLAB SECRET
# ------------------------------------------------------------

YOUTUBE_API_KEY = userdata.get("YOUTUBE_API_KEY")

if not YOUTUBE_API_KEY:
    raise RuntimeError(
        "YOUTUBE_API_KEY was not found in Colab Secrets. "
        "Add it under the 🔑 Secrets panel and run this cell again."
    )

youtube = build(
    "youtube",
    "v3",
    developerKey=YOUTUBE_API_KEY
)

print("YouTube API client initialized successfully.")


# ------------------------------------------------------------
# 2. EXPERIMENT CONFIGURATION
# ------------------------------------------------------------

QUERY_CONFIG = [
    {
        "query": "Full Python Course",
        "domain": "Programming",
        "slug": "python"
    },
    {
        "query": "Linear Algebra Course",
        "domain": "Mathematics",
        "slug": "linear_algebra"
    },
    {
        "query": "Quantum Mechanics Course",
        "domain": "Physics",
        "slug": "quantum_mechanics"
    },
    {
        "query": "Human Anatomy Course",
        "domain": "Biology",
        "slug": "human_anatomy"
    },
    {
        "query": "Machine Learning Course",
        "domain": "Data Science",
        "slug": "machine_learning"
    }
]

MAX_VIDEOS = 20
MAX_COMMENTS = 100

OUTPUT_ROOT = "/content/revision_phase9a"
os.makedirs(OUTPUT_ROOT, exist_ok=True)


# ------------------------------------------------------------
# 3. HELPER FUNCTIONS
# ------------------------------------------------------------

def parse_duration(duration):
    """
    Convert ISO 8601 YouTube duration into minutes.
    Example:
        PT1H20M30S -> 80.5 minutes
    """
    match = re.match(
        r"PT"
        r"(?:(\d+)H)?"
        r"(?:(\d+)M)?"
        r"(?:(\d+)S)?",
        duration
    )

    if not match:
        return 0.0

    hours = int(match.group(1) or 0)
    minutes = int(match.group(2) or 0)
    seconds = int(match.group(3) or 0)

    return hours * 60 + minutes + seconds / 60


def get_video_metadata(video_id):
    """
    Retrieve detailed metadata for one YouTube video.
    """

    try:
        response = youtube.videos().list(
            part="snippet,statistics,contentDetails",
            id=video_id
        ).execute()

        if not response.get("items"):
            return None

        item = response["items"][0]

        snippet = item.get("snippet", {})
        statistics = item.get("statistics", {})
        content = item.get("contentDetails", {})

        return {
            "video_id": video_id,
            "video_title": snippet.get("title", ""),
            "description": snippet.get("description", ""),
            "channel_name": snippet.get("channelTitle", ""),
            "views": int(statistics.get("viewCount", 0)),
            "likes": int(statistics.get("likeCount", 0)),
            "comment_count": int(statistics.get("commentCount", 0)),
            "duration": content.get("duration", ""),
            "duration_minutes": parse_duration(
                content.get("duration", "")
            ),
            "publish_date": snippet.get("publishedAt", ""),
            "video_url": f"https://www.youtube.com/watch?v={video_id}"
        }

    except Exception as e:
        print(f"Metadata error for {video_id}: {e}")
        return None


def get_comments(video_id, max_comments=100):
    """
    Retrieve top-level comments for one video.

    We intentionally keep this consistent with the original
    experimental design: up to 100 top-level comments/video.
    """

    comments = []

    try:
        response = youtube.commentThreads().list(
            part="snippet",
            videoId=video_id,
            maxResults=max_comments,
            textFormat="plainText"
        ).execute()

        for item in response.get("items", []):

            snippet = (
                item
                .get("snippet", {})
                .get("topLevelComment", {})
                .get("snippet", {})
            )

            comments.append({
                "video_id": video_id,
                "comment_id": (
                    item
                    .get("snippet", {})
                    .get("topLevelComment", {})
                    .get("id", "")
                ),
                "comment_text": snippet.get("textDisplay", ""),
                "comment_like_count": snippet.get("likeCount", 0),
                "comment_published_at": snippet.get(
                    "publishedAt", ""
                )
            })

    except Exception as e:
        # Comments can be disabled for some videos.
        print(
            f"  Comment retrieval skipped for {video_id}: "
            f"{type(e).__name__}"
        )

    return comments


# ------------------------------------------------------------
# 4. RUN RETRIEVAL FOR EACH EDUCATIONAL DOMAIN
# ------------------------------------------------------------

summary_rows = []

for config in QUERY_CONFIG:

    query = config["query"]
    domain = config["domain"]
    slug = config["slug"]

    print("\n" + "=" * 70)
    print(f"DOMAIN : {domain}")
    print(f"QUERY  : {query}")
    print("=" * 70)

    domain_dir = os.path.join(OUTPUT_ROOT, slug)
    os.makedirs(domain_dir, exist_ok=True)

    # --------------------------------------------------------
    # SEARCH FOR VIDEOS
    # --------------------------------------------------------

    search_response = youtube.search().list(
        q=query,
        part="id,snippet",
        type="video",
        maxResults=MAX_VIDEOS
    ).execute()

    search_items = search_response.get("items", [])

    video_ids = []

    for item in search_items:

        video_id = (
            item
            .get("id", {})
            .get("videoId")
        )

        if video_id:
            video_ids.append(video_id)

    # Remove duplicates while preserving order
    video_ids = list(dict.fromkeys(video_ids))

    print(f"Videos retrieved: {len(video_ids)}")

    # --------------------------------------------------------
    # VIDEO METADATA
    # --------------------------------------------------------

    video_rows = []

    for i, video_id in enumerate(video_ids, start=1):

        metadata = get_video_metadata(video_id)

        if metadata is not None:
            metadata["query"] = query
            metadata["domain"] = domain
            video_rows.append(metadata)

        print(
            f"  Metadata: {i}/{len(video_ids)}",
            end="\r"
        )

        time.sleep(0.05)

    print()

    videos_df = pd.DataFrame(video_rows)

    # --------------------------------------------------------
    # COMMENTS
    # --------------------------------------------------------

    comment_rows = []

    for i, video_id in enumerate(video_ids, start=1):

        comments = get_comments(
            video_id,
            max_comments=MAX_COMMENTS
        )

        for comment in comments:
            comment["query"] = query
            comment["domain"] = domain

        comment_rows.extend(comments)

        print(
            f"  Comments: {i}/{len(video_ids)}",
            end="\r"
        )

        time.sleep(0.05)

    print()

    comments_df = pd.DataFrame(comment_rows)

    # --------------------------------------------------------
    # SAVE DOMAIN-SPECIFIC FILES
    # --------------------------------------------------------

    videos_path = os.path.join(
        domain_dir,
        "videos.csv"
    )

    comments_path = os.path.join(
        domain_dir,
        "comments.csv"
    )

    videos_df.to_csv(
        videos_path,
        index=False
    )

    comments_df.to_csv(
        comments_path,
        index=False
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    unique_video_count = (
        videos_df["video_id"].nunique()
        if not videos_df.empty
        else 0
    )

    total_comments = len(comments_df)

    avg_comments = (
        total_comments / unique_video_count
        if unique_video_count > 0
        else 0
    )

    duplicate_video_ids = (
        len(videos_df) - unique_video_count
        if not videos_df.empty
        else 0
    )

    summary_rows.append({
        "query": query,
        "domain": domain,
        "slug": slug,
        "videos_retrieved": len(videos_df),
        "unique_videos": unique_video_count,
        "duplicate_video_ids": duplicate_video_ids,
        "total_comments": total_comments,
        "average_comments_per_video": avg_comments,
        "videos_file": videos_path,
        "comments_file": comments_path
    })

    print(f"\n{domain} validation:")
    print(f"  Videos              : {len(videos_df)}")
    print(f"  Unique videos       : {unique_video_count}")
    print(f"  Total comments      : {total_comments}")
    print(f"  Avg comments/video  : {avg_comments:.2f}")
    print(f"  Duplicate video IDs : {duplicate_video_ids}")


# ------------------------------------------------------------
# 5. CREATE OVERALL SUMMARY
# ------------------------------------------------------------

summary_df = pd.DataFrame(summary_rows)

summary_path = os.path.join(
    OUTPUT_ROOT,
    "query_summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)

print("\n" + "=" * 70)
print("PHASE 9A COMPLETE")
print("=" * 70)

display(summary_df)


# ------------------------------------------------------------
# 6. STRICT VALIDATION
# ------------------------------------------------------------

print("\nRunning final validation...\n")

validation_passed = True

for _, row in summary_df.iterrows():

    query = row["query"]
    videos = int(row["videos_retrieved"])
    unique_videos = int(row["unique_videos"])
    duplicates = int(row["duplicate_video_ids"])

    print(
        f"{query}: "
        f"{videos} videos | "
        f"{unique_videos} unique | "
        f"{duplicates} duplicates"
    )

    if videos != MAX_VIDEOS:
        validation_passed = False
        print(
            f"  WARNING: Expected {MAX_VIDEOS} videos "
            f"but retrieved {videos}."
        )

    if unique_videos != videos:
        validation_passed = False
        print("  WARNING: Duplicate video IDs detected.")


print("\n" + "=" * 70)

if validation_passed:
    print("✅ PHASE 9A VALIDATION PASSED")
    print("Requested target: up to 20 videos per domain; final cohort size is validated dynamically.")
else:
    print("⚠️ PHASE 9A VALIDATION REQUIRES REVIEW")
    print("Check the warnings above before continuing.")

print("=" * 70)

print(f"\nAll Phase 9A files saved under:")
print(OUTPUT_ROOT)
````

````python
# ============================================================
# PHASE 9B — MULTI-DOMAIN PREPROCESSING + SEMANTIC RELEVANCE
# ============================================================
#
# Purpose:
#   Apply the same preprocessing + E5 semantic relevance
#   methodology across all five educational domains.
#
# Original experiment:
#   - lowercase text
#   - remove URLs
#   - remove non-alphanumeric characters
#   - normalize whitespace
#   - E5-base-v2 embeddings
#   - normalized embeddings
#   - dot-product similarity
#   - select top relevant comments
#
# IMPORTANT:
#   This does NOT modify Phase 1–8 files.
# ============================================================


# ------------------------------------------------------------
# 1. INSTALL / IMPORT
# ------------------------------------------------------------

!pip install -q sentence-transformers

import os
import re
import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer


# ------------------------------------------------------------
# 2. PATHS
# ------------------------------------------------------------

INPUT_ROOT = "/content/revision_phase9a"
OUTPUT_ROOT = "/content/revision_phase9b"

os.makedirs(OUTPUT_ROOT, exist_ok=True)


# ------------------------------------------------------------
# 3. DOMAIN CONFIGURATION
# ------------------------------------------------------------

DOMAIN_CONFIG = [
    {
        "query": "Full Python Course",
        "domain": "Programming",
        "slug": "python",
        "semantic_query": (
            "Complete Python programming course "
            "explaining programming concepts and practical usage"
        )
    },
    {
        "query": "Linear Algebra Course",
        "domain": "Mathematics",
        "slug": "linear_algebra",
        "semantic_query": (
            "Complete linear algebra course "
            "explaining mathematical concepts, methods, and problem solving"
        )
    },
    {
        "query": "Quantum Mechanics Course",
        "domain": "Physics",
        "slug": "quantum_mechanics",
        "semantic_query": (
            "Complete quantum mechanics course "
            "explaining physical concepts, principles, and problem solving"
        )
    },
    {
        "query": "Human Anatomy Course",
        "domain": "Biology",
        "slug": "human_anatomy",
        "semantic_query": (
            "Complete human anatomy course "
            "explaining anatomical structures, functions, and biological concepts"
        )
    },
    {
        "query": "Machine Learning Course",
        "domain": "Data Science",
        "slug": "machine_learning",
        "semantic_query": (
            "Complete machine learning course "
            "explaining machine learning concepts, methods, and practical applications"
        )
    }
]

TOP_K = 1000


# ------------------------------------------------------------
# 4. LOAD E5 MODEL
# ------------------------------------------------------------

print("Loading E5 model...")

model = SentenceTransformer(
    "intfloat/e5-base-v2"
)

print("E5 model loaded successfully.")


# ------------------------------------------------------------
# 5. TEXT PREPROCESSING
# ------------------------------------------------------------

def clean_comment(text):
    """
    Same basic preprocessing strategy used in the
    original experiment.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        "",
        text
    )

    # Remove non-alphanumeric characters
    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# ------------------------------------------------------------
# 6. PROCESS EACH DOMAIN
# ------------------------------------------------------------

summary_rows = []

for config in DOMAIN_CONFIG:

    query = config["query"]
    domain = config["domain"]
    slug = config["slug"]
    semantic_query = config["semantic_query"]

    print("\n" + "=" * 75)
    print(f"DOMAIN          : {domain}")
    print(f"QUERY           : {query}")
    print(f"SEMANTIC QUERY  : {semantic_query}")
    print("=" * 75)

    # --------------------------------------------------------
    # INPUT
    # --------------------------------------------------------

    comments_path = os.path.join(
        INPUT_ROOT,
        slug,
        "comments.csv"
    )

    if not os.path.exists(comments_path):
        raise FileNotFoundError(
            f"Comments file not found:\n{comments_path}"
        )

    comments_df = pd.read_csv(
        comments_path
    )

    print(
        f"Raw comments loaded: {len(comments_df)}"
    )

    # --------------------------------------------------------
    # PREPROCESSING
    # --------------------------------------------------------

    comments_df["cleaned_comment"] = (
        comments_df["comment_text"]
        .apply(clean_comment)
    )

    comments_df["comment_length"] = (
        comments_df["cleaned_comment"]
        .str.split()
        .str.len()
    )

    # Remove empty comments
    before_count = len(comments_df)

    comments_df = comments_df[
        comments_df["cleaned_comment"].str.len() > 0
    ].copy()

    after_count = len(comments_df)

    removed_count = before_count - after_count

    print(
        f"After preprocessing : {after_count}"
    )

    print(
        f"Empty comments removed: {removed_count}"
    )

    # --------------------------------------------------------
    # E5 INPUT FORMAT
    # --------------------------------------------------------

    # E5 requires the appropriate prefixes.
    passage_texts = [
    "passage: " + text
    for text in comments_df["cleaned_comment"].tolist()
    ]

    query_text = (
        "query: "
        + semantic_query
    )

    # --------------------------------------------------------
    # QUERY EMBEDDING
    # --------------------------------------------------------

    query_embedding = model.encode(
        query_text,
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    # --------------------------------------------------------
    # COMMENT EMBEDDINGS
    # --------------------------------------------------------

    print("Generating comment embeddings...")

    comment_embeddings = model.encode(
        passage_texts,
        batch_size=32,
        show_progress_bar=True,
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    print(
        f"Embedding matrix shape: "
        f"{comment_embeddings.shape}"
    )

    # --------------------------------------------------------
    # DOT PRODUCT SIMILARITY
    # --------------------------------------------------------
    #
    # Because embeddings are L2-normalized,
    # dot product is equivalent to cosine similarity.
    # --------------------------------------------------------

    relevance_scores = (
        comment_embeddings @ query_embedding
    )

    comments_df["relevance_score"] = (
        relevance_scores
    )

    # --------------------------------------------------------
    # SORT BY RELEVANCE
    # --------------------------------------------------------

    comments_df = comments_df.sort_values(
        by="relevance_score",
        ascending=False
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # TOP-K RELEVANT COMMENTS
    # --------------------------------------------------------

    actual_top_k = min(
        TOP_K,
        len(comments_df)
    )

    relevant_df = (
        comments_df
        .head(actual_top_k)
        .copy()
    )

    relevant_df["relevance_rank"] = (
        np.arange(1, len(relevant_df) + 1)
    )

    # Add experiment metadata
    relevant_df["query"] = query
    relevant_df["domain"] = domain
    relevant_df["semantic_query"] = semantic_query

    # --------------------------------------------------------
    # SAVE FULL CLEANED DATA
    # --------------------------------------------------------

    domain_output = os.path.join(
        OUTPUT_ROOT,
        slug
    )

    os.makedirs(
        domain_output,
        exist_ok=True
    )

    cleaned_path = os.path.join(
        domain_output,
        "comments_cleaned.csv"
    )

    comments_df.to_csv(
        cleaned_path,
        index=False
    )

    # --------------------------------------------------------
    # SAVE TOP-K RELEVANT COMMENTS
    # --------------------------------------------------------

    relevant_path = os.path.join(
        domain_output,
        "comments_relevant.csv"
    )

    relevant_df.to_csv(
        relevant_path,
        index=False
    )

    # --------------------------------------------------------
    # STATISTICS
    # --------------------------------------------------------

    summary_rows.append({
        "query": query,
        "domain": domain,
        "slug": slug,
        "raw_comments": before_count,
        "clean_comments": after_count,
        "empty_removed": removed_count,
        "relevant_comments": len(relevant_df),
        "mean_relevance": relevant_df[
            "relevance_score"
        ].mean(),
        "std_relevance": relevant_df[
            "relevance_score"
        ].std(),
        "min_relevance": relevant_df[
            "relevance_score"
        ].min(),
        "median_relevance": relevant_df[
            "relevance_score"
        ].median(),
        "max_relevance": relevant_df[
            "relevance_score"
        ].max(),
        "cleaned_file": cleaned_path,
        "relevant_file": relevant_path
    })

    # --------------------------------------------------------
    # DOMAIN OUTPUT
    # --------------------------------------------------------

    print("\nDomain results:")
    print(
        f"  Raw comments       : {before_count}"
    )
    print(
        f"  Clean comments     : {after_count}"
    )
    print(
        f"  Empty removed      : {removed_count}"
    )
    print(
        f"  Top relevant      : {len(relevant_df)}"
    )
    print(
        f"  Mean relevance     : "
        f"{relevant_df['relevance_score'].mean():.6f}"
    )
    print(
        f"  Std relevance      : "
        f"{relevant_df['relevance_score'].std():.6f}"
    )
    print(
        f"  Minimum relevance  : "
        f"{relevant_df['relevance_score'].min():.6f}"
    )
    print(
        f"  Median relevance   : "
        f"{relevant_df['relevance_score'].median():.6f}"
    )
    print(
        f"  Maximum relevance  : "
        f"{relevant_df['relevance_score'].max():.6f}"
    )

    # --------------------------------------------------------
    # SAMPLE
    # --------------------------------------------------------

    print("\nTop 5 relevant comments:")

    display(
        relevant_df[
            [
                "video_id",
                "relevance_rank",
                "relevance_score",
                "cleaned_comment"
            ]
        ].head(5)
    )


# ------------------------------------------------------------
# 7. SAVE OVERALL SUMMARY
# ------------------------------------------------------------

summary_df = pd.DataFrame(
    summary_rows
)

summary_path = os.path.join(
    OUTPUT_ROOT,
    "semantic_relevance_summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)


# ------------------------------------------------------------
# 8. FINAL VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9B — FINAL VALIDATION")
print("=" * 75)

validation_passed = True

for _, row in summary_df.iterrows():

    domain = row["domain"]

    raw_comments = int(
        row["raw_comments"]
    )

    clean_comments = int(
        row["clean_comments"]
    )

    relevant_comments = int(
        row["relevant_comments"]
    )

    print(
        f"\n{domain}"
    )

    print(
        f"  Raw comments      : {raw_comments}"
    )

    print(
        f"  Clean comments    : {clean_comments}"
    )

    print(
        f"  Relevant comments : {relevant_comments}"
    )

    # Basic integrity checks

    if clean_comments > raw_comments:
        validation_passed = False
        print(
            "  ❌ ERROR: Clean comments > raw comments"
        )

    if relevant_comments > clean_comments:
        validation_passed = False
        print(
            "  ❌ ERROR: Relevant comments > clean comments"
        )

    if relevant_comments == 0:
        validation_passed = False
        print(
            "  ❌ ERROR: No relevant comments"
        )


print("\n" + "=" * 75)

if validation_passed:
    print("✅ PHASE 9B VALIDATION PASSED")
    print(
        "All five domains successfully completed "
        "preprocessing and E5 semantic relevance."
    )
else:
    print("⚠️ PHASE 9B VALIDATION FAILED")
    print(
        "Inspect the errors above before continuing."
    )

print("=" * 75)

print(
    f"\nResults saved under:\n{OUTPUT_ROOT}"
)

print(
    f"\nSummary file:\n{summary_path}"
)
````

````python
# ============================================================
# PHASE 9C — MULTI-DOMAIN SENTIMENT + EMOTIONAL ENGAGEMENT
# ============================================================
#
# Input:
#   Phase 9B top-1000 relevant comments/domain
#
# Output:
#   Video-level emotional engagement statistics
#
# Same methodology as original experiment:
#
#   positive sentiment -> positive score
#   negative sentiment -> negative score
#   neutral sentiment  -> 0
#
#   emotion_weighted_score =
#       relevance_score * sentiment_score
#
#   emotional_engagement =
#       mean(emotion_weighted_score)
#
#   emotional_variance =
#       variance(emotion_weighted_score)
#
# ============================================================


# ------------------------------------------------------------
# 1. INSTALL / IMPORT
# ------------------------------------------------------------

!pip install -q transformers torch pandas numpy tqdm

import os
import numpy as np
import pandas as pd
import torch

from tqdm.auto import tqdm
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)


# ------------------------------------------------------------
# 2. PATHS
# ------------------------------------------------------------

INPUT_ROOT = "/content/revision_phase9b"
OUTPUT_ROOT = "/content/revision_phase9c"

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)


# ------------------------------------------------------------
# 3. DOMAIN CONFIGURATION
# ------------------------------------------------------------

DOMAIN_CONFIG = [
    {
        "query": "Full Python Course",
        "domain": "Programming",
        "slug": "python"
    },
    {
        "query": "Linear Algebra Course",
        "domain": "Mathematics",
        "slug": "linear_algebra"
    },
    {
        "query": "Quantum Mechanics Course",
        "domain": "Physics",
        "slug": "quantum_mechanics"
    },
    {
        "query": "Human Anatomy Course",
        "domain": "Biology",
        "slug": "human_anatomy"
    },
    {
        "query": "Machine Learning Course",
        "domain": "Data Science",
        "slug": "machine_learning"
    }
]


# ------------------------------------------------------------
# 4. LOAD CARDIFFNLP SENTIMENT MODEL
# ------------------------------------------------------------

SENTIMENT_MODEL_NAME = (
    "cardiffnlp/twitter-roberta-base-sentiment-latest"
)

print("Loading CardiffNLP RoBERTa sentiment model...")

tokenizer = AutoTokenizer.from_pretrained(
    SENTIMENT_MODEL_NAME
)

sentiment_model = AutoModelForSequenceClassification.from_pretrained(
    SENTIMENT_MODEL_NAME
)

# Select GPU if available
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

sentiment_model = sentiment_model.to(device)
sentiment_model.eval()

print(
    f"Sentiment model loaded successfully."
)

print(
    f"Device: {device}"
)


# ------------------------------------------------------------
# 5. IDENTIFY SENTIMENT LABELS
# ------------------------------------------------------------

id2label = sentiment_model.config.id2label

print("\nModel labels:")

for idx, label in id2label.items():
    print(
        f"  {idx} -> {label}"
    )


# ------------------------------------------------------------
# 6. SENTIMENT SCORING FUNCTION
# ------------------------------------------------------------

def sentiment_score_from_probabilities(
    probabilities
):
    """
    Convert RoBERTa class probabilities into
    the scalar sentiment score used by the
    original pipeline.

    Positive:
        +P(positive)

    Negative:
        -P(negative)

    Neutral:
        0

    The model probabilities are first mapped
    according to the model's label names.
    """

    positive_prob = 0.0
    negative_prob = 0.0

    for idx, label in id2label.items():

        label_lower = str(label).lower()

        if "positive" in label_lower:
            positive_prob = float(
                probabilities[idx]
            )

        elif "negative" in label_lower:
            negative_prob = float(
                probabilities[idx]
            )

    return positive_prob - negative_prob


# ------------------------------------------------------------
# 7. BATCH SENTIMENT PREDICTION
# ------------------------------------------------------------

def predict_sentiment(
    texts,
    batch_size=32
):
    """
    Generate sentiment scores and labels
    for a list of comments.
    """

    all_scores = []
    all_labels = []
    all_positive = []
    all_negative = []
    all_neutral = []

    for start in tqdm(
        range(
            0,
            len(texts),
            batch_size
        ),
        desc="Sentiment batches"
    ):

        batch_texts = texts[
            start:start + batch_size
        ]

        encoded = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=512,
            return_tensors="pt"
        )

        encoded = {
            key: value.to(device)
            for key, value in encoded.items()
        }

        with torch.no_grad():

            outputs = sentiment_model(
                **encoded
            )

            probabilities = torch.softmax(
                outputs.logits,
                dim=1
            ).cpu().numpy()

        for probs in probabilities:

            score = (
                sentiment_score_from_probabilities(
                    probs
                )
            )

            predicted_idx = int(
                np.argmax(probs)
            )

            predicted_label = str(
                id2label[predicted_idx]
            )

            # Store probability values
            positive_prob = 0.0
            negative_prob = 0.0
            neutral_prob = 0.0

            for idx, label in id2label.items():

                label_lower = str(label).lower()

                if "positive" in label_lower:
                    positive_prob = float(
                        probs[idx]
                    )

                elif "negative" in label_lower:
                    negative_prob = float(
                        probs[idx]
                    )

                elif "neutral" in label_lower:
                    neutral_prob = float(
                        probs[idx]
                    )

            all_scores.append(score)
            all_labels.append(predicted_label)
            all_positive.append(positive_prob)
            all_negative.append(negative_prob)
            all_neutral.append(neutral_prob)

    return (
        all_scores,
        all_labels,
        all_positive,
        all_negative,
        all_neutral
    )


# ------------------------------------------------------------
# 8. PROCESS ALL FIVE DOMAINS
# ------------------------------------------------------------

domain_summary_rows = []

all_video_emotion_rows = []

for config in DOMAIN_CONFIG:

    query = config["query"]
    domain = config["domain"]
    slug = config["slug"]

    print("\n" + "=" * 75)
    print(f"DOMAIN : {domain}")
    print(f"QUERY  : {query}")
    print("=" * 75)

    # --------------------------------------------------------
    # INPUT
    # --------------------------------------------------------

    input_path = os.path.join(
        INPUT_ROOT,
        slug,
        "comments_relevant.csv"
    )

    if not os.path.exists(input_path):

        raise FileNotFoundError(
            f"Phase 9B relevant-comments file not found:\n"
            f"{input_path}"
        )

    relevant_df = pd.read_csv(
        input_path
    )

    print(
        f"Relevant comments loaded: "
        f"{len(relevant_df)}"
    )

    # --------------------------------------------------------
    # CHECK REQUIRED COLUMNS
    # --------------------------------------------------------

    required_columns = [
        "video_id",
        "cleaned_comment",
        "relevance_score"
    ]

    missing_columns = [
        col
        for col in required_columns
        if col not in relevant_df.columns
    ]

    if missing_columns:

        raise ValueError(
            f"Missing required columns: "
            f"{missing_columns}"
        )

    # --------------------------------------------------------
    # SENTIMENT
    # --------------------------------------------------------

    texts = (
        relevant_df[
            "cleaned_comment"
        ]
        .fillna("")
        .astype(str)
        .tolist()
    )

    (
        sentiment_scores,
        sentiment_labels,
        positive_probs,
        negative_probs,
        neutral_probs
    ) = predict_sentiment(
        texts,
        batch_size=32
    )

    relevant_df[
        "sentiment_score"
    ] = sentiment_scores

    relevant_df[
        "sentiment_label"
    ] = sentiment_labels

    relevant_df[
        "positive_probability"
    ] = positive_probs

    relevant_df[
        "negative_probability"
    ] = negative_probs

    relevant_df[
        "neutral_probability"
    ] = neutral_probs

    # --------------------------------------------------------
    # RELEVANCE-WEIGHTED SENTIMENT
    # --------------------------------------------------------

    relevant_df[
        "emotion_weighted_score"
    ] = (
        relevant_df["relevance_score"]
        *
        relevant_df["sentiment_score"]
    )

    # Add metadata
    relevant_df["query"] = query
    relevant_df["domain"] = domain

    # --------------------------------------------------------
    # SAVE COMMENT-LEVEL SENTIMENT
    # --------------------------------------------------------

    domain_output = os.path.join(
        OUTPUT_ROOT,
        slug
    )

    os.makedirs(
        domain_output,
        exist_ok=True
    )

    comment_output_path = os.path.join(
        domain_output,
        "relevant_comments_sentiment.csv"
    )

    relevant_df.to_csv(
        comment_output_path,
        index=False
    )

    # --------------------------------------------------------
    # VIDEO-LEVEL AGGREGATION
    # --------------------------------------------------------

    video_emotion_df = (
        relevant_df
        .groupby("video_id")
        .agg(
            emotional_engagement=(
                "emotion_weighted_score",
                "mean"
            ),
            emotional_variance=(
                "emotion_weighted_score",
                "var"
            ),
            relevant_comment_count=(
                "emotion_weighted_score",
                "count"
            )
        )
        .reset_index()
    )

    # Variance is undefined for a single observation.
    # Replace such cases with zero.
    video_emotion_df[
        "emotional_variance"
    ] = (
        video_emotion_df[
            "emotional_variance"
        ]
        .fillna(0)
    )

    video_emotion_df["query"] = query
    video_emotion_df["domain"] = domain

    # --------------------------------------------------------
    # SENTIMENT DISTRIBUTION
    # --------------------------------------------------------

    sentiment_distribution = (
        relevant_df[
            "sentiment_label"
        ]
        .value_counts()
        .to_dict()
    )

    positive_count = sum(
        1
        for label in relevant_df[
            "sentiment_label"
        ]
        if "positive" in str(label).lower()
    )

    negative_count = sum(
        1
        for label in relevant_df[
            "sentiment_label"
        ]
        if "negative" in str(label).lower()
    )

    neutral_count = sum(
        1
        for label in relevant_df[
            "sentiment_label"
        ]
        if "neutral" in str(label).lower()
    )

    # --------------------------------------------------------
    # SAVE VIDEO-LEVEL RESULTS
    # --------------------------------------------------------

    video_output_path = os.path.join(
        domain_output,
        "video_emotional_engagement.csv"
    )

    video_emotion_df.to_csv(
        video_output_path,
        index=False
    )

    # Add to combined result
    all_video_emotion_rows.append(
        video_emotion_df
    )

    # --------------------------------------------------------
    # DOMAIN SUMMARY
    # --------------------------------------------------------

    domain_summary_rows.append({
        "query": query,
        "domain": domain,
        "slug": slug,
        "relevant_comments": len(relevant_df),
        "videos_with_relevant_comments": (
            video_emotion_df[
                "video_id"
            ].nunique()
        ),
        "positive_comments": positive_count,
        "negative_comments": negative_count,
        "neutral_comments": neutral_count,
        "positive_pct": (
            positive_count /
            len(relevant_df) * 100
        ),
        "negative_pct": (
            negative_count /
            len(relevant_df) * 100
        ),
        "neutral_pct": (
            neutral_count /
            len(relevant_df) * 100
        ),
        "mean_sentiment": (
            relevant_df[
                "sentiment_score"
            ].mean()
        ),
        "std_sentiment": (
            relevant_df[
                "sentiment_score"
            ].std()
        ),
        "mean_emotional_engagement": (
            video_emotion_df[
                "emotional_engagement"
            ].mean()
        ),
        "mean_emotional_variance": (
            video_emotion_df[
                "emotional_variance"
            ].mean()
        ),
        "comment_output": comment_output_path,
        "video_output": video_output_path
    })

    # --------------------------------------------------------
    # PRINT RESULTS
    # --------------------------------------------------------

    print("\nSentiment distribution:")

    print(
        f"  Positive : {positive_count}"
        f" ({positive_count / len(relevant_df) * 100:.2f}%)"
    )

    print(
        f"  Negative : {negative_count}"
        f" ({negative_count / len(relevant_df) * 100:.2f}%)"
    )

    print(
        f"  Neutral  : {neutral_count}"
        f" ({neutral_count / len(relevant_df) * 100:.2f}%)"
    )

    print(
        f"\nMean sentiment: "
        f"{relevant_df['sentiment_score'].mean():.6f}"
    )

    print(
        f"Std sentiment : "
        f"{relevant_df['sentiment_score'].std():.6f}"
    )

    print(
        f"\nVideos with relevant comments: "
        f"{video_emotion_df['video_id'].nunique()}"
    )

    print(
        f"Mean emotional engagement: "
        f"{video_emotion_df['emotional_engagement'].mean():.6f}"
    )

    print(
        f"Mean emotional variance: "
        f"{video_emotion_df['emotional_variance'].mean():.6f}"
    )

    print("\nTop 5 videos by emotional engagement:")

    display(
        video_emotion_df
        .sort_values(
            "emotional_engagement",
            ascending=False
        )
        .head(5)
    )


# ------------------------------------------------------------
# 9. COMBINE ALL DOMAINS
# ------------------------------------------------------------

combined_emotion_df = pd.concat(
    all_video_emotion_rows,
    ignore_index=True
)

combined_output_path = os.path.join(
    OUTPUT_ROOT,
    "all_domains_video_emotional_engagement.csv"
)

combined_emotion_df.to_csv(
    combined_output_path,
    index=False
)


# ------------------------------------------------------------
# 10. DOMAIN SUMMARY
# ------------------------------------------------------------

domain_summary_df = pd.DataFrame(
    domain_summary_rows
)

summary_output_path = os.path.join(
    OUTPUT_ROOT,
    "sentiment_emotional_engagement_summary.csv"
)

domain_summary_df.to_csv(
    summary_output_path,
    index=False
)


# ------------------------------------------------------------
# 11. FINAL VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9C — FINAL VALIDATION")
print("=" * 75)

validation_passed = True

for _, row in domain_summary_df.iterrows():

    domain = row["domain"]

    relevant_comments = int(
        row["relevant_comments"]
    )

    video_count = int(
        row["videos_with_relevant_comments"]
    )

    print(f"\n{domain}")
    print(
        f"  Relevant comments : "
        f"{relevant_comments}"
    )
    print(
        f"  Videos             : "
        f"{video_count}"
    )

    if relevant_comments != 1000:

        validation_passed = False

        print(
            "  ⚠️ WARNING: Expected "
            "1000 relevant comments."
        )

    if video_count == 0:

        validation_passed = False

        print(
            "  ❌ ERROR: No videos found."
        )

    if not np.isfinite(
        row["mean_sentiment"]
    ):

        validation_passed = False

        print(
            "  ❌ ERROR: Invalid sentiment value."
        )


print("\n" + "=" * 75)

print(
    f"Total video-level records: "
    f"{len(combined_emotion_df)}"
)

print(
    f"Unique video IDs: "
    f"{combined_emotion_df['video_id'].nunique()}"
)

print("=" * 75)

if validation_passed:

    print(
        "✅ PHASE 9C VALIDATION PASSED"
    )

    print(
        "All five domains completed "
        "sentiment and emotional-engagement analysis."
    )

else:

    print(
        "⚠️ PHASE 9C VALIDATION REQUIRES REVIEW"
    )

print("=" * 75)

print(
    f"\nResults saved under:\n"
    f"{OUTPUT_ROOT}"
)

print(
    f"\nCombined video-level file:\n"
    f"{combined_output_path}"
)

print(
    f"\nSummary file:\n"
    f"{summary_output_path}"
)
````

````python
# ============================================================
# PHASE 9D — FINAL ENGAGEMENT SCORE (FES)
# ============================================================
#
# Builds FES directly from the validated Phase 9C output.
# This cell intentionally creates /content/revision_phase9d
# before any downstream phase attempts to read from it.
#
# FES = 0.4*EmotionalEngagement
#     + 0.2*(1-EmotionalVariance)
#     + 0.2*LikesNorm
#     + 0.2*RelevantCommentCountNorm
#
# Min-Max normalization is computed over the complete cohort produced by
# Phase 9C. Domain sizes are read from the validated Phase 9C output;
# no padding, deletion, or duplication is performed.
# ============================================================

import os
import glob
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------
PHASE9C_PATH = (
    "/content/revision_phase9c/"
    "all_domains_video_emotional_engagement.csv"
)
PHASE9A_ROOT = "/content/revision_phase9a"
OUTPUT_ROOT = "/content/revision_phase9d"
FES_PATH = os.path.join(OUTPUT_ROOT, "phase9d_fes_full.csv")
TOP10_PATH = os.path.join(OUTPUT_ROOT, "phase9d_fes_top10.csv")

os.makedirs(OUTPUT_ROOT, exist_ok=True)

# ------------------------------------------------------------
# 2. LOAD VALIDATED PHASE 9C DATA
# ------------------------------------------------------------
engagement_df = pd.read_csv(PHASE9C_PATH)

print("=" * 75)
print("PHASE 9D — FINAL ENGAGEMENT SCORE")
print("=" * 75)
print(f"Phase 9C records: {len(engagement_df)}")
print(f"Unique video IDs: {engagement_df['video_id'].nunique()}")

required_engagement = [
    "video_id", "emotional_engagement", "emotional_variance",
    "relevant_comment_count", "query", "domain"
]
missing = [c for c in required_engagement if c not in engagement_df.columns]
if missing:
    raise ValueError(f"Phase 9C file is missing required columns: {missing}")

# ------------------------------------------------------------
# 3. LOAD VIDEO METADATA / LIKES FROM PHASE 9A
# ------------------------------------------------------------
video_files = sorted(
    glob.glob(os.path.join(PHASE9A_ROOT, "**", "videos.csv"), recursive=True)
)

if not video_files:
    raise FileNotFoundError(
        "No Phase 9A videos.csv files were found under "
        f"{PHASE9A_ROOT}. Phase 9D cannot compute LikesNorm without them."
    )

metadata_frames = []
for path in video_files:
    tmp = pd.read_csv(path)
    if "video_id" not in tmp.columns or "likes" not in tmp.columns:
        continue
    metadata_frames.append(tmp)

if not metadata_frames:
    raise ValueError("Phase 9A video files were found, but none contains video_id and likes.")

metadata_df = pd.concat(metadata_frames, ignore_index=True)
metadata_df = metadata_df.drop_duplicates(subset="video_id", keep="first")

# Keep the fields required by Phase 9E and later phases.
metadata_columns = [
    "video_id", "video_title", "description", "likes",
    "duration_minutes", "query", "domain"
]
metadata_columns = [c for c in metadata_columns if c in metadata_df.columns]
metadata_df = metadata_df[metadata_columns].copy()

# Avoid duplicate query/domain columns during merge; Phase 9C is authoritative.
for c in ["query", "domain"]:
    if c in metadata_df.columns:
        metadata_df = metadata_df.drop(columns=c)

fes_df = engagement_df.merge(
    metadata_df,
    on="video_id",
    how="left",
    validate="one_to_one"
)

if fes_df["likes"].isna().any():
    missing_ids = fes_df.loc[fes_df["likes"].isna(), "video_id"].tolist()
    raise ValueError(
        "Missing likes metadata for video IDs: "
        f"{missing_ids}"
    )

# ------------------------------------------------------------
# 4. NORMALIZE LIKES AND RELEVANT COMMENT COUNT
# ------------------------------------------------------------
scaler = MinMaxScaler()
fes_df[["likes_norm", "comment_count_norm"]] = scaler.fit_transform(
    fes_df[["likes", "relevant_comment_count"]].astype(float)
)

# Numerical safety for floating-point boundary values.
fes_df["likes_norm"] = fes_df["likes_norm"].clip(0.0, 1.0)
fes_df["comment_count_norm"] = fes_df["comment_count_norm"].clip(0.0, 1.0)

# ------------------------------------------------------------
# 5. COMPUTE FES
# ------------------------------------------------------------
fes_df["final_engagement_score"] = (
    0.4 * fes_df["emotional_engagement"]
    + 0.2 * (1.0 - fes_df["emotional_variance"])
    + 0.2 * fes_df["likes_norm"]
    + 0.2 * fes_df["comment_count_norm"]
)

# Global FES rank across the full multi-domain cohort.
fes_df = fes_df.sort_values(
    "final_engagement_score", ascending=False
).reset_index(drop=True)
fes_df["global_fes_rank"] = np.arange(1, len(fes_df) + 1)

# Domain-specific FES rank.
fes_df["domain_fes_rank"] = (
    fes_df.groupby("domain")["final_engagement_score"]
    .rank(method="first", ascending=False)
    .astype(int)
)

# ------------------------------------------------------------
# 6. SAVE OUTPUTS
# ------------------------------------------------------------
fes_df.to_csv(FES_PATH, index=False)
fes_df.head(10).to_csv(TOP10_PATH, index=False)

# ------------------------------------------------------------
# 7. VALIDATION
# ------------------------------------------------------------
expected_domain_counts = (
    engagement_df.groupby("domain")["video_id"].nunique().to_dict()
)

actual_domain_counts = fes_df.groupby("domain")["video_id"].nunique().to_dict()

print("\nDomain video counts:")
for domain, expected in expected_domain_counts.items():
    actual = actual_domain_counts.get(domain, 0)
    print(f"  {domain:15s}: {actual}")
    if actual != expected:
        raise RuntimeError(
            f"Unexpected cohort size for {domain}: expected {expected}, found {actual}."
        )

expected_total = len(engagement_df)
expected_unique = engagement_df["video_id"].nunique()

if len(fes_df) != expected_total or fes_df["video_id"].nunique() != expected_unique:
    raise RuntimeError(
        f"Phase 9D cohort validation failed: {len(fes_df)} rows / "
        f"{fes_df['video_id'].nunique()} unique; expected "
        f"{expected_total} rows / {expected_unique} unique."
    )

if fes_df["final_engagement_score"].isna().any():
    raise RuntimeError("Missing FES values detected.")

print("\n" + "=" * 75)
print("PHASE 9D VALIDATION PASSED")
print("=" * 75)
print(f"Total video records : {len(fes_df)}")
print(f"Unique video IDs    : {fes_df['video_id'].nunique()}")
print(f"FES minimum         : {fes_df['final_engagement_score'].min():.6f}")
print(f"FES maximum         : {fes_df['final_engagement_score'].max():.6f}")
print(f"Output file         : {FES_PATH}")
print("\nTop 10 FES videos:")
print(
    fes_df[["global_fes_rank", "domain", "video_id", "final_engagement_score"]]
    .head(10)
    .to_string(index=False)
)
````

````python
# ============================================================
# PHASE 9E-A — INTRINSIC QUALITY COMPONENTS
# ============================================================
#
# Calculates:
#   1. Semantic Alignment
#   2. Content Depth
#   3. Lexical Diversity
#
# Teaching Score is intentionally handled separately in
# Phase 9E-B because it requires the LLM.
#
# IMPORTANT:
#   - Does NOT modify Phase 1–9D files.
#   - Uses the corrected standard TTR formulation:
#
#       LexicalDiversity = UniqueWords / TotalWords
#
# ============================================================

import os
import re
import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import MinMaxScaler


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

FES_PATH = (
    "/content/revision_phase9d/"
    "phase9d_fes_full.csv"
)

PHASE9A_ROOT = (
    "/content/revision_phase9a"
)

OUTPUT_ROOT = (
    "/content/revision_phase9e"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. LOAD FES DATA
# ------------------------------------------------------------

fes_df = pd.read_csv(
    FES_PATH
)

print(
    f"FES records loaded: {len(fes_df)}"
)

print(
    f"Unique videos: "
    f"{fes_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 3. VALIDATE REQUIRED METADATA
# ------------------------------------------------------------

required_columns = [
    "video_id",
    "video_title",
    "description",
    "duration_minutes",
    "domain",
    "query"
]

missing_columns = [
    column
    for column in required_columns
    if column not in fes_df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing required columns: "
        f"{missing_columns}"
    )

print(
    "✅ Required video metadata available."
)


# ------------------------------------------------------------
# 4. LOAD E5 MODEL
# ------------------------------------------------------------

print(
    "\nLoading E5-base-v2..."
)

e5_model = SentenceTransformer(
    "intfloat/e5-base-v2"
)

print(
    "✅ E5-base-v2 loaded."
)


# ------------------------------------------------------------
# 5. BUILD VIDEO CONTENT
# ------------------------------------------------------------
#
# The original pipeline uses title + description for
# intrinsic content evaluation.
#
# We preserve that methodology here.
# ------------------------------------------------------------

fes_df["video_title"] = (
    fes_df["video_title"]
    .fillna("")
    .astype(str)
)

fes_df["description"] = (
    fes_df["description"]
    .fillna("")
    .astype(str)
)

fes_df["content_text"] = (
    fes_df["video_title"]
    + " "
    + fes_df["description"]
)

fes_df["content_text"] = (
    fes_df["content_text"]
    .str.strip()
)


# ------------------------------------------------------------
# 6. CREATE DOMAIN-SPECIFIC SEMANTIC QUERIES
# ------------------------------------------------------------

semantic_queries = {
    "Programming":
        "Complete Python programming course "
        "explaining programming concepts and "
        "practical problem solving",

    "Mathematics":
        "Complete linear algebra course "
        "explaining mathematical concepts, "
        "methods, and problem solving",

    "Physics":
        "Complete quantum mechanics course "
        "explaining physical concepts, "
        "principles, and problem solving",

    "Biology":
        "Complete human anatomy course "
        "explaining anatomical structures, "
        "functions, and biological concepts",

    "Data Science":
        "Complete machine learning course "
        "explaining machine learning concepts, "
        "methods, and practical applications"
}


fes_df["semantic_query"] = (
    fes_df["domain"]
    .map(semantic_queries)
)


# ------------------------------------------------------------
# 7. PREPARE E5 INPUTS
# ------------------------------------------------------------

passage_inputs = [
    "passage: " + text
    for text in fes_df[
        "content_text"
    ].tolist()
]

query_inputs = [
    "query: " + query
    for query in fes_df[
        "semantic_query"
    ].tolist()
]


# ------------------------------------------------------------
# 8. GENERATE CONTENT EMBEDDINGS
# ------------------------------------------------------------

print(
    "\nGenerating video content embeddings..."
)

content_embeddings = e5_model.encode(
    passage_inputs,
    batch_size=16,
    show_progress_bar=True,
    normalize_embeddings=True,
    convert_to_numpy=True
)

print(
    f"Content embedding shape: "
    f"{content_embeddings.shape}"
)


# ------------------------------------------------------------
# 9. GENERATE QUERY EMBEDDINGS
# ------------------------------------------------------------

print(
    "\nGenerating query embeddings..."
)

query_embeddings = e5_model.encode(
    query_inputs,
    batch_size=16,
    show_progress_bar=False,
    normalize_embeddings=True,
    convert_to_numpy=True
)

print(
    f"Query embedding shape: "
    f"{query_embeddings.shape}"
)


# ------------------------------------------------------------
# 10. SEMANTIC ALIGNMENT
# ------------------------------------------------------------
#
# Because both vectors are normalized:
#
#   dot product == cosine similarity
#
# ------------------------------------------------------------

semantic_alignment = np.sum(
    content_embeddings * query_embeddings,
    axis=1
)

fes_df[
    "semantic_alignment"
] = semantic_alignment


# ------------------------------------------------------------
# 11. CONTENT DEPTH
# ------------------------------------------------------------
#
# Duration is used only as a structural proxy for
# content coverage, not as a direct claim that longer
# videos are educationally better.
#
# This distinction is important for the reviewer response.
# ------------------------------------------------------------

duration_values = (
    pd.to_numeric(
        fes_df["duration_minutes"],
        errors="coerce"
    )
    .fillna(0)
)

duration_scaler = MinMaxScaler()

fes_df[
    "depth_score"
] = duration_scaler.fit_transform(
    duration_values.to_numpy().reshape(-1, 1)
).ravel()


# ------------------------------------------------------------
# 12. LEXICAL DIVERSITY — STANDARD TTR
# ------------------------------------------------------------
#
# Standard Type-Token Ratio:
#
#   TTR = UniqueWords / TotalWords
#
# We calculate this from title + description.
# ------------------------------------------------------------

def calculate_lexical_diversity(text):

    text = str(text).lower()

    words = re.findall(
        r"\b[a-z0-9]+\b",
        text
    )

    total_words = len(words)

    if total_words == 0:
        return 0.0

    unique_words = len(
        set(words)
    )

    return (
        unique_words /
        total_words
    )


fes_df[
    "lexical_diversity_raw"
] = (
    fes_df[
        "content_text"
    ]
    .apply(
        calculate_lexical_diversity
    )
)


# ------------------------------------------------------------
# 13. NORMALIZE LEXICAL DIVERSITY
# ------------------------------------------------------------

lexical_scaler = MinMaxScaler()

fes_df[
    "lexical_diversity"
] = lexical_scaler.fit_transform(
    fes_df[
        [
            "lexical_diversity_raw"
        ]
    ]
).ravel()


# ------------------------------------------------------------
# 14. BASIC VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9E-A — COMPONENT SUMMARY")
print("=" * 75)

print(
    f"Videos: {len(fes_df)}"
)

print(
    f"Unique IDs: "
    f"{fes_df['video_id'].nunique()}"
)


print("\nSemantic Alignment:")
print(
    f"  Min    : "
    f"{fes_df['semantic_alignment'].min():.6f}"
)
print(
    f"  Mean   : "
    f"{fes_df['semantic_alignment'].mean():.6f}"
)
print(
    f"  Max    : "
    f"{fes_df['semantic_alignment'].max():.6f}"
)


print("\nDepth Score:")
print(
    f"  Min    : "
    f"{fes_df['depth_score'].min():.6f}"
)
print(
    f"  Mean   : "
    f"{fes_df['depth_score'].mean():.6f}"
)
print(
    f"  Max    : "
    f"{fes_df['depth_score'].max():.6f}"
)


print("\nLexical Diversity — Raw TTR:")
print(
    f"  Min    : "
    f"{fes_df['lexical_diversity_raw'].min():.6f}"
)
print(
    f"  Mean   : "
    f"{fes_df['lexical_diversity_raw'].mean():.6f}"
)
print(
    f"  Max    : "
    f"{fes_df['lexical_diversity_raw'].max():.6f}"
)


print("\nLexical Diversity — Normalized:")
print(
    f"  Min    : "
    f"{fes_df['lexical_diversity'].min():.6f}"
)
print(
    f"  Mean   : "
    f"{fes_df['lexical_diversity'].mean():.6f}"
)
print(
    f"  Max    : "
    f"{fes_df['lexical_diversity'].max():.6f}"
)


# ------------------------------------------------------------
# 15. CHECK INVALID VALUES
# ------------------------------------------------------------

component_columns = [
    "semantic_alignment",
    "depth_score",
    "lexical_diversity_raw",
    "lexical_diversity"
]

invalid_counts = {}

for column in component_columns:

    invalid_counts[column] = int(
        (
            ~np.isfinite(
                fes_df[column]
            )
        ).sum()
    )

print(
    "\nInvalid values:"
)

for column, count in invalid_counts.items():

    print(
        f"  {column}: {count}"
    )


if sum(
    invalid_counts.values()
) > 0:

    raise ValueError(
        "Invalid component values detected."
    )


# ------------------------------------------------------------
# 16. DISPLAY COMPONENTS
# ------------------------------------------------------------

print(
    "\nTop 10 videos by semantic alignment:"
)

display(
    fes_df[
        [
            "domain",
            "video_id",
            "video_title",
            "semantic_alignment",
            "depth_score",
            "lexical_diversity_raw",
            "lexical_diversity"
        ]
    ]
    .sort_values(
        "semantic_alignment",
        ascending=False
    )
    .head(10)
)


# ------------------------------------------------------------
# 17. SAVE PREPARED IQS DATA
# ------------------------------------------------------------

iqs_components_path = os.path.join(
    OUTPUT_ROOT,
    "phase9e_iqs_components.csv"
)

fes_df.to_csv(
    iqs_components_path,
    index=False
)


# ------------------------------------------------------------
# 18. PREPARE TEACHING-SCORE INPUT FILE
# ------------------------------------------------------------
#
# This file will be used in Phase 9E-B for the LLM.
# ------------------------------------------------------------

teaching_input_df = fes_df[
    [
        "video_id",
        "domain",
        "query",
        "video_title",
        "description",
        "duration_minutes",
        "semantic_alignment",
        "depth_score",
        "lexical_diversity_raw"
    ]
].copy()

teaching_input_path = os.path.join(
    OUTPUT_ROOT,
    "phase9e_teaching_score_input.csv"
)

teaching_input_df.to_csv(
    teaching_input_path,
    index=False
)


# ------------------------------------------------------------
# 19. FINAL VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9E-A — FINAL VALIDATION")
print("=" * 75)

validation_passed = True


expected_n = len(fes_df)
expected_unique_n = fes_df["video_id"].nunique()

if len(fes_df) != expected_n:

    validation_passed = False

    print(
        f"❌ Cohort size changed unexpectedly: expected {expected_n} videos, "
        f"found {len(fes_df)}"
    )

else:

    print(
        f"✅ {expected_n} videos present"
    )


if fes_df[
    "video_id"
].nunique() != expected_unique_n:

    validation_passed = False

    print(
        "❌ Video IDs are not unique"
    )

else:

    print(
        f"✅ {expected_unique_n} unique video IDs"
    )


for column in [
    "semantic_alignment",
    "depth_score",
    "lexical_diversity_raw",
    "lexical_diversity"
]:

    if not np.isfinite(
        fes_df[column]
    ).all():

        validation_passed = False

        print(
            f"❌ Invalid values in {column}"
        )

    else:

        print(
            f"✅ {column} valid"
        )


print("\n" + "=" * 75)

if validation_passed:

    print(
        "✅ PHASE 9E-A VALIDATION PASSED"
    )

    print(
        "\nThe three deterministic IQS components "
        "are ready for Teaching Score integration."
    )

else:

    print(
        "⚠️ PHASE 9E-A VALIDATION REQUIRES REVIEW"
    )

print("=" * 75)

print(
    f"\nIQS component file:\n"
    f"{iqs_components_path}"
)

print(
    f"\nTeaching-score input:\n"
    f"{teaching_input_path}"
)
````

````python
# ============================================================
# PHASE 9E-B — LLM TEACHING SCORE
# ============================================================
#
# Model:
#   gemini-3.5-flash-lite
#
# API:
#   Google Gemini Interactions API
#
# Scoring:
#   0–10
#
# Criteria:
#   1. Clear learning objective
#   2. Structured explanation
#   3. Logical concept progression
#   4. Example usage
#   5. Beginner friendliness
#   6. Practical problem solving
#
# IMPORTANT:
#   - Resumable
#   - Saves after every successful batch
#   - Existing scores are NOT recomputed
#   - Raw LLM responses are retained
#   - Uses 6 videos/request to reduce API calls
#
# ============================================================


from google import genai
from google.colab import userdata

import os
import json
import re
import time
import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

INPUT_PATH = (
    "/content/revision_phase9e/"
    "phase9e_teaching_score_input.csv"
)

OUTPUT_ROOT = (
    "/content/revision_phase9e/"
    "teaching_score_results"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

RESULTS_PATH = os.path.join(
    OUTPUT_ROOT,
    "teaching_scores.csv"
)

RAW_RESPONSES_PATH = os.path.join(
    OUTPUT_ROOT,
    "raw_llm_responses.jsonl"
)


# ------------------------------------------------------------
# 2. LOAD INPUT
# ------------------------------------------------------------

teaching_df = pd.read_csv(
    INPUT_PATH
)

print(
    f"Input records: {len(teaching_df)}"
)

print(
    f"Unique video IDs: "
    f"{teaching_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 3. VALIDATE INPUT
# ------------------------------------------------------------

required_columns = [
    "video_id",
    "domain",
    "query",
    "video_title",
    "description",
    "duration_minutes"
]

missing_columns = [
    c
    for c in required_columns
    if c not in teaching_df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

expected_n = len(teaching_df)
expected_unique_n = teaching_df["video_id"].nunique()

if len(teaching_df) != expected_n:

    raise ValueError(
        f"Teaching-score input cohort is internally inconsistent: expected {expected_n}, found {len(teaching_df)}"
    )

if teaching_df["video_id"].nunique() != expected_unique_n:

    raise ValueError(
        "Video IDs are not unique."
    )

print(
    "✅ Input validation passed."
)


# ------------------------------------------------------------
# 4. GEMINI CLIENT
# ------------------------------------------------------------

GEMINI_API_KEY = userdata.get(
    "GEMINI_API_KEY"
)

if not GEMINI_API_KEY:

    raise RuntimeError(
        "GEMINI_API_KEY not found in Colab Secrets."
    )

client = genai.Client(
    api_key=GEMINI_API_KEY
)

MODEL_NAME = "gemini-3.5-flash-lite"

print(
    f"Gemini model: {MODEL_NAME}"
)


# ------------------------------------------------------------
# 5. TEACHING-SCORE CRITERIA
# ------------------------------------------------------------

CRITERIA = """
Evaluate the educational teaching quality using these six criteria:

1. Clear learning objective
   - Is it clear what the learner is expected to learn?

2. Structured explanation
   - Is the material organized and presented systematically?

3. Logical concept progression
   - Do concepts appear in a meaningful learning sequence?

4. Example usage
   - Are examples, demonstrations, or concrete applications indicated?

5. Beginner friendliness
   - Is the material accessible and understandable to learners?

6. Practical problem solving
   - Does the content appear to support practical application,
     exercises, projects, or problem solving?
"""


# ------------------------------------------------------------
# 6. PROMPT VERSION
# ------------------------------------------------------------

PROMPT_VERSION = "Phase9E-B_v1"


# ------------------------------------------------------------
# 7. LOAD EXISTING RESULTS IF AVAILABLE
# ------------------------------------------------------------

if os.path.exists(RESULTS_PATH):

    existing_results = pd.read_csv(
        RESULTS_PATH
    )

    print(
        f"Existing Teaching Scores found: "
        f"{len(existing_results)}"
    )

else:

    existing_results = pd.DataFrame(
        columns=[
            "video_id",
            "domain",
            "teaching_score",
            "llm_model",
            "prompt_version",
            "status"
        ]
    )

    print(
        "No previous Teaching Scores found."
    )


# ------------------------------------------------------------
# 8. IDENTIFY COMPLETED VIDEOS
# ------------------------------------------------------------

completed_ids = set()

if not existing_results.empty:

    valid_existing = existing_results[
        existing_results["teaching_score"]
        .notna()
    ]

    completed_ids = set(
        valid_existing[
            "video_id"
        ].astype(str)
    )

print(
    f"Completed videos to skip: "
    f"{len(completed_ids)}"
)


# ------------------------------------------------------------
# 9. HELPER — EXTRACT JSON
# ------------------------------------------------------------

def extract_json(text):

    text = str(text).strip()

    # Remove markdown fences if present
    text = re.sub(
        r"```json\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(
        r"```\s*",
        "",
        text
    )

    # Try direct JSON
    try:
        return json.loads(text)

    except Exception:
        pass

    # Try extracting first JSON array
    match = re.search(
        r"\[.*\]",
        text,
        flags=re.DOTALL
    )

    if match:

        try:
            return json.loads(
                match.group(0)
            )

        except Exception:
            pass

    # Try extracting JSON object
    match = re.search(
        r"\{.*\}",
        text,
        flags=re.DOTALL
    )

    if match:

        try:
            return json.loads(
                match.group(0)
            )

        except Exception:
            pass

    return None


# ------------------------------------------------------------
# 10. HELPER — VALIDATE SCORE
# ------------------------------------------------------------

def clean_score(value):

    try:

        score = float(value)

    except Exception:

        return None

    if not np.isfinite(score):

        return None

    # Accept only 0–10
    if score < 0 or score > 10:

        return None

    return round(
        score,
        2
    )


# ------------------------------------------------------------
# 11. BUILD BATCH PROMPT
# ------------------------------------------------------------

def build_batch_prompt(batch_df):

    records = []

    for _, row in batch_df.iterrows():

        description = str(
            row["description"]
        )

        # Keep enough information for evaluation while
        # avoiding unnecessarily huge prompts.
        #
        # Title + first 8000 chars of description.
        description = description[:8000]

        records.append({
            "video_id": str(
                row["video_id"]
            ),
            "domain": str(
                row["domain"]
            ),
            "query": str(
                row["query"]
            ),
            "title": str(
                row["video_title"]
            ),
            "description": description,
            "duration_minutes": float(
                row["duration_minutes"]
            )
        })

    records_json = json.dumps(
        records,
        ensure_ascii=False,
        indent=2
    )

    prompt = f"""
You are an educational-content evaluator.

Your task is to assign ONE overall Teaching Score from 0 to 10
to EACH video provided below.

{CRITERIA}

IMPORTANT EVALUATION RULES:

- Evaluate only the educational information available in the
  provided title and description.
- Do NOT assume that a video contains information that is not
  supported by the provided metadata.
- Do not use popularity, likes, views, comments, or channel
  reputation to determine the Teaching Score.
- Do not reward a video merely because it is long.
- Consider duration only as contextual information.
- Evaluate instructional potential and evidence of educational
  organization from the available metadata.
- Use integer scores from 0 to 10.
- Return exactly one score for every video.
- Preserve every video_id exactly.

Score interpretation:

0–2   = very weak educational evidence
3–4   = limited educational evidence
5–6   = moderate educational quality
7–8   = strong educational quality
9–10  = very strong educational quality

{records_json}

Return ONLY a JSON array in this exact format:

[
  {{
    "video_id": "VIDEO_ID",
    "teaching_score": 8
  }}
]

Do not include explanations outside the JSON array.
"""

    return prompt


# ------------------------------------------------------------
# 12. PREPARE PENDING VIDEOS
# ------------------------------------------------------------

pending_df = teaching_df[
    ~teaching_df[
        "video_id"
    ].astype(str).isin(
        completed_ids
    )
].copy()

print(
    f"\nPending videos: {len(pending_df)}"
)


# ------------------------------------------------------------
# 13. BATCH SIZE
# ------------------------------------------------------------

BATCH_SIZE = 6

total_batches = int(
    np.ceil(
        len(pending_df) /
        BATCH_SIZE
    )
)

print(
    f"Batch size: {BATCH_SIZE}"
)

print(
    f"Gemini requests required: "
    f"{total_batches}"
)


# ------------------------------------------------------------
# 14. PROCESS BATCHES
# ------------------------------------------------------------

new_results = []

for batch_number, start in enumerate(
    range(
        0,
        len(pending_df),
        BATCH_SIZE
    ),
    start=1
):

    batch_df = pending_df.iloc[
        start:start + BATCH_SIZE
    ].copy()

    print("\n" + "=" * 75)
    print(
        f"BATCH {batch_number}/{total_batches}"
    )
    print(
        f"Videos in batch: {len(batch_df)}"
    )
    print("=" * 75)

    print(
        batch_df[
            [
                "video_id",
                "domain",
                "video_title"
            ]
        ].to_string(
            index=False
        )
    )

    prompt = build_batch_prompt(
        batch_df
    )

    try:

        response = client.interactions.create(
            model=MODEL_NAME,
            input=prompt
        )

        raw_output = (
            response.output_text
            if hasattr(
                response,
                "output_text"
            )
            else str(response)
        )

        parsed = extract_json(
            raw_output
        )

        if not isinstance(
            parsed,
            list
        ):

            raise ValueError(
                "Gemini response was not a JSON list."
            )

        # ----------------------------------------------------
        # VALIDATE BATCH RESPONSE
        # ----------------------------------------------------

        expected_ids = set(
            batch_df[
                "video_id"
            ].astype(str)
        )

        response_ids = set()

        batch_results = []

        for item in parsed:

            if not isinstance(
                item,
                dict
            ):
                continue

            video_id = str(
                item.get(
                    "video_id",
                    ""
                )
            )

            score = clean_score(
                item.get(
                    "teaching_score"
                )
            )

            if (
                video_id in expected_ids
                and score is not None
            ):

                response_ids.add(
                    video_id
                )

                original_row = batch_df[
                    batch_df[
                        "video_id"
                    ].astype(str)
                    == video_id
                ].iloc[0]

                batch_results.append({
                    "video_id": video_id,
                    "domain": original_row[
                        "domain"
                    ],
                    "teaching_score": score,
                    "llm_model": MODEL_NAME,
                    "prompt_version": PROMPT_VERSION,
                    "status": "completed"
                })

        missing_ids = (
            expected_ids -
            response_ids
        )

        extra_ids = (
            response_ids -
            expected_ids
        )

        print(
            f"\nValid scores returned: "
            f"{len(batch_results)}/{len(batch_df)}"
        )

        if extra_ids:

            print(
                f"Unexpected IDs: {extra_ids}"
            )

        if missing_ids:

            print(
                f"Missing IDs: {missing_ids}"
            )

            raise ValueError(
                "Batch did not return a valid score "
                "for every requested video."
            )

        # ----------------------------------------------------
        # SAVE RAW RESPONSE
        # ----------------------------------------------------

        raw_record = {
            "batch_number": batch_number,
            "model": MODEL_NAME,
            "prompt_version": PROMPT_VERSION,
            "video_ids": list(
                expected_ids
            ),
            "raw_output": raw_output
        }

        with open(
            RAW_RESPONSES_PATH,
            "a",
            encoding="utf-8"
        ) as f:

            f.write(
                json.dumps(
                    raw_record,
                    ensure_ascii=False
                )
                + "\n"
            )

        # ----------------------------------------------------
        # SAVE SCORES IMMEDIATELY
        # ----------------------------------------------------

        new_results.extend(
            batch_results
        )

        combined_results = pd.concat(
            [
                existing_results,
                pd.DataFrame(new_results)
            ],
            ignore_index=True
        )

        # Remove accidental duplicate IDs
        combined_results = (
            combined_results
            .drop_duplicates(
                subset=["video_id"],
                keep="last"
            )
        )

        combined_results.to_csv(
            RESULTS_PATH,
            index=False
        )

        print(
            f"✅ Batch {batch_number} saved."
        )

        print(
            f"Total saved so far: "
            f"{len(combined_results)}"
        )

        # Small delay between requests
        time.sleep(1)

    except Exception as e:

        print(
            f"\n❌ Batch {batch_number} failed:"
        )

        print(
            type(e).__name__,
            str(e)
        )

        print(
            "\nProgress already saved."
        )

        print(
            "Stop here and inspect the error "
            "before continuing."
        )

        break


# ------------------------------------------------------------
# 15. LOAD FINAL CURRENT RESULTS
# ------------------------------------------------------------

if os.path.exists(
    RESULTS_PATH
):

    final_scores = pd.read_csv(
        RESULTS_PATH
    )

else:

    final_scores = pd.DataFrame()


# ------------------------------------------------------------
# 16. FINAL STATUS — CURRENT IQS COHORT ONLY
# ------------------------------------------------------------
# The Teaching Score cache may contain extra historical records.
# Completion is measured ONLY against the current input cohort.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9E-B — CURRENT COHORT STATUS")
print("=" * 75)

current_ids = set(teaching_df["video_id"].astype(str).str.strip())

if not final_scores.empty:
    final_scores["video_id"] = (
        final_scores["video_id"].astype(str).str.strip()
    )

    duplicate_cache_ids = (
        final_scores.loc[
            final_scores["video_id"].duplicated(),
            "video_id"
        ].unique().tolist()
    )

    if duplicate_cache_ids:
        raise ValueError(
            "Duplicate Teaching Score video IDs found in cache: "
            f"{duplicate_cache_ids}"
        )

scored_ids = set(
    final_scores["video_id"] if not final_scores.empty else []
)

current_scored_ids = current_ids & scored_ids
missing_current_ids = current_ids - scored_ids
extra_cached_ids = scored_ids - current_ids

print(f"Current input cohort : {len(current_ids)}")
print(f"Cache records        : {len(final_scores)}")
print(f"Current cohort scored: {len(current_scored_ids)}")
print(f"Missing current      : {len(missing_current_ids)}")
print(f"Extra cached records : {len(extra_cached_ids)}")

if missing_current_ids:
    print("\nMissing Teaching Scores for current cohort:")
    for video_id in sorted(missing_current_ids):
        print(f"  - {video_id}")

    print(
        "\n⚠️ Teaching Score run is incomplete for the current cohort."
    )
    print(
        "Rerun this cell; completed current video IDs will be skipped."
    )
else:
    print(
        f"\n✅ ALL {len(current_ids)} CURRENT COHORT "
        "TEACHING SCORES ARE AVAILABLE"
    )

if not final_scores.empty and current_scored_ids:
    current_scores = final_scores[
        final_scores["video_id"].isin(current_ids)
    ].copy()

    print(
        f"\nCurrent-cohort Teaching Score range: "
        f"{current_scores['teaching_score'].min()} "
        f"→ {current_scores['teaching_score'].max()}"
    )
    print(
        f"Current-cohort Mean Teaching Score: "
        f"{current_scores['teaching_score'].mean():.3f}"
    )

    print("\nCurrent-cohort score distribution:")
    display(
        current_scores["teaching_score"]
        .value_counts()
        .sort_index()
        .to_frame("count")
    )

print(f"\nResults file:\n{RESULTS_PATH}")
print(f"\nRaw responses:\n{RAW_RESPONSES_PATH}")
````

````python
# ============================================================
# PHASE 9E-C — FINAL INTRINSIC QUALITY SCORE (IQS)
# ============================================================
#
# IQS =
#   0.30 * Semantic Alignment
# + 0.20 * Depth Score
# + 0.20 * Lexical Diversity
# + 0.30 * Normalized Teaching Score
#
# IMPORTANT:
# The IQS component cohort is the authoritative cohort.
# The Teaching Score file may contain extra historical/cache
# records. Only Teaching Scores belonging to the current IQS
# component cohort are used.
#
# ============================================================


# ------------------------------------------------------------
# 0. IMPORTS
# ------------------------------------------------------------

import os
import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

COMPONENTS_PATH = (
    "/content/revision_phase9e/"
    "phase9e_iqs_components.csv"
)

TEACHING_PATH = (
    "/content/revision_phase9e/"
    "teaching_score_results/"
    "teaching_scores.csv"
)

OUTPUT_ROOT = (
    "/content/revision_phase9e"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. LOAD IQS COMPONENTS
# ------------------------------------------------------------

components_df = pd.read_csv(
    COMPONENTS_PATH
)

print(
    f"IQS component records: "
    f"{len(components_df)}"
)

print(
    f"Unique component video IDs: "
    f"{components_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 3. LOAD TEACHING SCORES
# ------------------------------------------------------------

teaching_df = pd.read_csv(
    TEACHING_PATH
)

print(
    f"\nTeaching Score records: "
    f"{len(teaching_df)}"
)

print(
    f"Unique Teaching Score IDs: "
    f"{teaching_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 4. BASIC COLUMN VALIDATION
# ------------------------------------------------------------

required_component_columns = [
    "video_id",
    "domain",
    "video_title",
    "semantic_alignment",
    "depth_score",
    "lexical_diversity"
]

missing_component_columns = [
    col
    for col in required_component_columns
    if col not in components_df.columns
]

if missing_component_columns:

    raise ValueError(
        "Missing required IQS component columns: "
        f"{missing_component_columns}"
    )


required_teaching_columns = [
    "video_id",
    "teaching_score"
]

missing_teaching_columns = [
    col
    for col in required_teaching_columns
    if col not in teaching_df.columns
]

if missing_teaching_columns:

    raise ValueError(
        "Missing required Teaching Score columns: "
        f"{missing_teaching_columns}"
    )


# ------------------------------------------------------------
# 5. NORMALIZE VIDEO IDs
# ------------------------------------------------------------

components_df["video_id"] = (
    components_df["video_id"]
    .astype(str)
    .str.strip()
)

teaching_df["video_id"] = (
    teaching_df["video_id"]
    .astype(str)
    .str.strip()
)


# ------------------------------------------------------------
# 6. VALIDATE IQS COMPONENT COHORT
# ------------------------------------------------------------

component_ids = set(
    components_df["video_id"]
)

teaching_ids = set(
    teaching_df["video_id"]
)


print("\n" + "=" * 75)
print("COHORT VALIDATION")
print("=" * 75)


intersection_ids = (
    component_ids &
    teaching_ids
)

components_only = (
    component_ids -
    teaching_ids
)

teaching_only = (
    teaching_ids -
    component_ids
)


print(
    f"Components ∩ Teaching: "
    f"{len(intersection_ids)}"
)

print(
    f"Components only: "
    f"{len(components_only)}"
)

print(
    f"Teaching only: "
    f"{len(teaching_only)}"
)


# ------------------------------------------------------------
# 7. CHECK THAT EVERY COMPONENT HAS A TEACHING SCORE
# ------------------------------------------------------------

if components_only:

    print(
        "\nMissing Teaching Scores for these "
        "IQS component videos:"
    )

    for video_id in sorted(components_only):
        print(
            f"  - {video_id}"
        )

    raise ValueError(
        "Some IQS component videos do not have "
        "Teaching Scores."
    )


print(
    "\n✅ Every IQS component video has "
    "a Teaching Score."
)


# ------------------------------------------------------------
# 8. FILTER TEACHING SCORES TO CURRENT IQS COHORT
# ------------------------------------------------------------
#
# IMPORTANT:
# Teaching Score CSV may contain extra historical records.
#
# Example:
#
# IQS components       = 99
# Teaching Scores      = 134
# Current overlap      = 99
# Extra Teaching IDs   = 35
#
# We keep only the 99 required records.
# ------------------------------------------------------------

teaching_df = teaching_df[
    teaching_df["video_id"].isin(component_ids)
].copy()


print(
    f"\nTeaching Score records after "
    f"cohort filtering: {len(teaching_df)}"
)

print(
    f"Unique Teaching Score IDs after "
    f"filtering: {teaching_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 9. CHECK FOR DUPLICATE TEACHING SCORES
# ------------------------------------------------------------

duplicate_teaching_ids = (
    teaching_df[
        teaching_df["video_id"].duplicated(
            keep=False
        )
    ]["video_id"]
    .unique()
)


if len(duplicate_teaching_ids) > 0:

    print(
        "\nDuplicate Teaching Score IDs detected:"
    )

    for video_id in duplicate_teaching_ids:
        print(
            f"  - {video_id}"
        )

    raise ValueError(
        "Multiple Teaching Score records exist "
        "for one or more IQS component videos. "
        "Resolve duplicates before computing IQS."
    )


# ------------------------------------------------------------
# 10. FINAL COHORT MATCH CHECK
# ------------------------------------------------------------

filtered_teaching_ids = set(
    teaching_df["video_id"]
)


if filtered_teaching_ids != component_ids:

    missing_after_filter = (
        component_ids -
        filtered_teaching_ids
    )

    unexpected_after_filter = (
        filtered_teaching_ids -
        component_ids
    )

    raise ValueError(
        "Teaching Score cohort still does not "
        "match the IQS component cohort.\n"
        f"Missing: {sorted(missing_after_filter)}\n"
        f"Unexpected: {sorted(unexpected_after_filter)}"
    )


print(
    "\n" + "=" * 75
)

print(
    "✅ EXACT CURRENT IQS COHORT COVERAGE"
)

print(
    f"IQS component cohort: "
    f"{len(component_ids)} videos"
)

print(
    f"Teaching Score cohort: "
    f"{len(filtered_teaching_ids)} videos"
)

if teaching_only:

    print(
        f"ℹ️ Ignored {len(teaching_only)} "
        "extra Teaching Score records outside "
        "the current IQS cohort."
    )

print(
    "=" * 75
)


# ------------------------------------------------------------
# 11. MERGE
# ------------------------------------------------------------

teaching_merge_columns = [
    "video_id",
    "teaching_score"
]


# Optional metadata columns.
# Include them only if they exist.

for optional_column in [
    "llm_model",
    "prompt_version",
    "status"
]:

    if optional_column in teaching_df.columns:

        teaching_merge_columns.append(
            optional_column
        )


iqs_df = components_df.merge(
    teaching_df[
        teaching_merge_columns
    ],
    on="video_id",
    how="inner",
    validate="one_to_one"
)


# ------------------------------------------------------------
# 12. MERGE VALIDATION
# ------------------------------------------------------------

print(
    f"\nMerged IQS records: "
    f"{len(iqs_df)}"
)

print(
    f"Unique merged IDs: "
    f"{iqs_df['video_id'].nunique()}"
)


expected_n = len(components_df)

expected_unique_n = (
    components_df["video_id"].nunique()
)


if len(iqs_df) != expected_n:

    raise ValueError(
        "Merged record count does not match "
        "the IQS component cohort.\n"
        f"Expected: {expected_n}\n"
        f"Found: {len(iqs_df)}"
    )


if (
    iqs_df["video_id"].nunique()
    != expected_unique_n
):

    raise ValueError(
        "Merged unique video count does not "
        "match the IQS component cohort."
    )


print(
    "✅ Merge validated."
)


# ------------------------------------------------------------
# 13. TEACHING SCORE NUMERICAL VALIDATION
# ------------------------------------------------------------

iqs_df["teaching_score"] = pd.to_numeric(
    iqs_df["teaching_score"],
    errors="coerce"
)


if iqs_df["teaching_score"].isna().any():

    missing_count = (
        iqs_df["teaching_score"]
        .isna()
        .sum()
    )

    raise ValueError(
        f"{missing_count} Teaching Score "
        "values are missing or non-numeric."
    )


if (
    iqs_df["teaching_score"].min() < 0
    or
    iqs_df["teaching_score"].max() > 10
):

    raise ValueError(
        "Teaching Score contains values "
        "outside the valid [0,10] range."
    )


# ------------------------------------------------------------
# 14. TEACHING SCORE NORMALIZATION
# ------------------------------------------------------------

iqs_df[
    "teaching_score_normalized"
] = (
    iqs_df["teaching_score"]
    / 10.0
)


# ------------------------------------------------------------
# 15. FINAL IQS
# ------------------------------------------------------------

iqs_df[
    "intrinsic_quality_score"
] = (

    0.30
    * iqs_df[
        "semantic_alignment"
    ]

    +

    0.20
    * iqs_df[
        "depth_score"
    ]

    +

    0.20
    * iqs_df[
        "lexical_diversity"
    ]

    +

    0.30
    * iqs_df[
        "teaching_score_normalized"
    ]
)


# ------------------------------------------------------------
# 16. GLOBAL IQS RANK
# ------------------------------------------------------------

iqs_df = (
    iqs_df
    .sort_values(
        "intrinsic_quality_score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


iqs_df[
    "global_iqs_rank"
] = np.arange(
    1,
    len(iqs_df) + 1
)


# ------------------------------------------------------------
# 17. DOMAIN-WISE IQS RANK
# ------------------------------------------------------------

iqs_df[
    "domain_iqs_rank"
] = (

    iqs_df
    .groupby(
        "domain"
    )[
        "intrinsic_quality_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 18. COMPONENT CONTRIBUTIONS
# ------------------------------------------------------------

iqs_df[
    "semantic_contribution"
] = (
    0.30
    * iqs_df[
        "semantic_alignment"
    ]
)


iqs_df[
    "depth_contribution"
] = (
    0.20
    * iqs_df[
        "depth_score"
    ]
)


iqs_df[
    "lexical_contribution"
] = (
    0.20
    * iqs_df[
        "lexical_diversity"
    ]
)


iqs_df[
    "teaching_contribution"
] = (
    0.30
    * iqs_df[
        "teaching_score_normalized"
    ]
)


# ------------------------------------------------------------
# 19. DOMINANT CONTRIBUTION
# ------------------------------------------------------------

contribution_columns = [
    "semantic_contribution",
    "depth_contribution",
    "lexical_contribution",
    "teaching_contribution"
]


component_names = {

    "semantic_contribution":
        "Semantic Alignment",

    "depth_contribution":
        "Depth",

    "lexical_contribution":
        "Lexical Diversity",

    "teaching_contribution":
        "Teaching Score"
}


iqs_df[
    "dominant_iqs_component"
] = (

    iqs_df[
        contribution_columns
    ]
    .idxmax(
        axis=1
    )
    .map(
        component_names
    )
)


# ------------------------------------------------------------
# 20. SAVE COMPLETE IQS
# ------------------------------------------------------------

full_output = os.path.join(
    OUTPUT_ROOT,
    "phase9e_iqs_full.csv"
)


iqs_df.to_csv(
    full_output,
    index=False
)


# ------------------------------------------------------------
# 21. SAVE TOP 10
# ------------------------------------------------------------

top10_output = os.path.join(
    OUTPUT_ROOT,
    "phase9e_iqs_top10.csv"
)


iqs_df.head(
    10
).to_csv(
    top10_output,
    index=False
)


# ------------------------------------------------------------
# 22. DOMAIN SUMMARY
# ------------------------------------------------------------

domain_summary = (

    iqs_df
    .groupby(
        "domain"
    )
    .agg(

        videos=(
            "video_id",
            "nunique"
        ),

        mean_iqs=(
            "intrinsic_quality_score",
            "mean"
        ),

        std_iqs=(
            "intrinsic_quality_score",
            "std"
        ),

        min_iqs=(
            "intrinsic_quality_score",
            "min"
        ),

        max_iqs=(
            "intrinsic_quality_score",
            "max"
        ),

        mean_semantic_alignment=(
            "semantic_alignment",
            "mean"
        ),

        mean_depth=(
            "depth_score",
            "mean"
        ),

        mean_lexical_diversity=(
            "lexical_diversity",
            "mean"
        ),

        mean_teaching_score=(
            "teaching_score",
            "mean"
        )
    )
    .reset_index()
)


domain_summary_output = os.path.join(
    OUTPUT_ROOT,
    "phase9e_iqs_domain_summary.csv"
)


domain_summary.to_csv(
    domain_summary_output,
    index=False
)


# ------------------------------------------------------------
# 23. TOP 10 DISPLAY
# ------------------------------------------------------------

print(
    "\n" + "=" * 85
)

print(
    "TOP 10 VIDEOS BY INTRINSIC QUALITY SCORE"
)

print(
    "=" * 85
)


display(
    iqs_df[
        [
            "global_iqs_rank",
            "domain",
            "video_id",
            "video_title",
            "semantic_alignment",
            "depth_score",
            "lexical_diversity",
            "teaching_score",
            "intrinsic_quality_score",
            "dominant_iqs_component"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 24. DOMAIN SUMMARY DISPLAY
# ------------------------------------------------------------

print(
    "\n" + "=" * 85
)

print(
    "DOMAIN-WISE IQS SUMMARY"
)

print(
    "=" * 85
)


display(
    domain_summary
)


# ------------------------------------------------------------
# 25. TEACHING SCORE SUMMARY
# ------------------------------------------------------------

print(
    "\n" + "=" * 85
)

print(
    "TEACHING SCORE SUMMARY"
)

print(
    "=" * 85
)


print(
    f"Mean: "
    f"{iqs_df['teaching_score'].mean():.3f}"
)


print(
    f"Std: "
    f"{iqs_df['teaching_score'].std():.3f}"
)


print(
    f"Min: "
    f"{iqs_df['teaching_score'].min():.1f}"
)


print(
    f"Max: "
    f"{iqs_df['teaching_score'].max():.1f}"
)


# ------------------------------------------------------------
# 26. FINAL VALIDATION
# ------------------------------------------------------------

print(
    "\n" + "=" * 85
)

print(
    "PHASE 9E-C — FINAL VALIDATION"
)

print(
    "=" * 85
)


validation_passed = True


# ------------------------------------------------------------
# Record count
# ------------------------------------------------------------

if len(iqs_df) != expected_n:

    validation_passed = False

    print(
        f"❌ Expected {expected_n} videos, "
        f"found {len(iqs_df)}"
    )

else:

    print(
        f"✅ {expected_n} video records"
    )


# ------------------------------------------------------------
# Unique IDs
# ------------------------------------------------------------

if (
    iqs_df["video_id"].nunique()
    != expected_unique_n
):

    validation_passed = False

    print(
        "❌ Duplicate video IDs"
    )

else:

    print(
        f"✅ {expected_unique_n} unique video IDs"
    )


# ------------------------------------------------------------
# Teaching Score validity
# ------------------------------------------------------------

if (

    iqs_df["teaching_score"].min() < 0

    or

    iqs_df["teaching_score"].max() > 10

):

    validation_passed = False

    print(
        "❌ Teaching Score outside [0,10]"
    )

else:

    print(
        "✅ Teaching Scores within [0,10]"
    )


# ------------------------------------------------------------
# Teaching normalized
# ------------------------------------------------------------

if not (

    iqs_df[
        "teaching_score_normalized"
    ]
    .between(
        0,
        1
    )
    .all()

):

    validation_passed = False

    print(
        "❌ Normalized Teaching Score invalid"
    )

else:

    print(
        "✅ Normalized Teaching Score valid"
    )


# ------------------------------------------------------------
# IQS finite
# ------------------------------------------------------------

if not np.isfinite(
    iqs_df[
        "intrinsic_quality_score"
    ]
).all():

    validation_passed = False

    print(
        "❌ Non-finite IQS values"
    )

else:

    print(
        "✅ All IQS values finite"
    )


# ------------------------------------------------------------
# IQS missing
# ------------------------------------------------------------

if (
    iqs_df[
        "intrinsic_quality_score"
    ]
    .isna()
    .sum()
    > 0
):

    validation_passed = False

    print(
        "❌ Missing IQS values"
    )

else:

    print(
        "✅ No missing IQS values"
    )


# ------------------------------------------------------------
# Final cohort consistency
# ------------------------------------------------------------

if (
    set(iqs_df["video_id"])
    != component_ids
):

    validation_passed = False

    print(
        "❌ Final IQS cohort does not match "
        "component cohort"
    )

else:

    print(
        "✅ Final IQS cohort matches "
        "component cohort"
    )


# ------------------------------------------------------------
# 27. FINAL RESULT
# ------------------------------------------------------------

print(
    "\n" + "=" * 85
)


if validation_passed:

    print(
        "✅ PHASE 9E-C VALIDATION PASSED"
    )

    print(
        "Intrinsic Quality Score successfully "
        f"computed for all {expected_n} videos."
    )

else:

    print(
        "⚠️ PHASE 9E-C VALIDATION REQUIRES REVIEW"
    )


print(
    "=" * 85
)


# ------------------------------------------------------------
# 28. OUTPUT PATHS
# ------------------------------------------------------------

print(
    f"\nFull IQS file:\n"
    f"{full_output}"
)


print(
    f"\nTop-10 IQS file:\n"
    f"{top10_output}"
)


print(
    f"\nDomain summary:\n"
    f"{domain_summary_output}"
)
````

````python
# ============================================================
# PHASE 9F — ADAPTIVE HYBRID RANKING
# ============================================================
#
# Hybrid Score:
#
#   Hybrid = alpha * FES + (1-alpha) * IQS
#
# where:
#
#   EngagementStrength = log1p(relevant_comment_count)
#
#   alpha =
#       EngagementStrength /
#       (EngagementStrength + k)
#
#   k = 5
#
# This phase combines:
#   Phase 9D → Final Engagement Score (FES)
#   Phase 9E → Intrinsic Quality Score (IQS)
#
# ============================================================


import os
import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

FES_PATH = (
    "/content/revision_phase9d/"
    "phase9d_fes_full.csv"
)

IQS_PATH = (
    "/content/revision_phase9e/"
    "phase9e_iqs_full.csv"
)

OUTPUT_ROOT = (
    "/content/revision_phase9f"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. LOAD FES
# ------------------------------------------------------------

fes_df = pd.read_csv(
    FES_PATH
)

print(
    f"FES records: {len(fes_df)}"
)

print(
    f"FES unique videos: "
    f"{fes_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 3. LOAD IQS
# ------------------------------------------------------------

iqs_df = pd.read_csv(
    IQS_PATH
)

print(
    f"\nIQS records: {len(iqs_df)}"
)

print(
    f"IQS unique videos: "
    f"{iqs_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 4. VALIDATE INPUT COHORTS
# ------------------------------------------------------------

fes_ids = set(
    fes_df[
        "video_id"
    ].astype(str)
)

iqs_ids = set(
    iqs_df[
        "video_id"
    ].astype(str)
)


print("\n" + "=" * 75)
print("COHORT VALIDATION")
print("=" * 75)

print(
    f"FES ∩ IQS       : "
    f"{len(fes_ids & iqs_ids)}"
)

print(
    f"FES only        : "
    f"{len(fes_ids - iqs_ids)}"
)

print(
    f"IQS only        : "
    f"{len(iqs_ids - fes_ids)}"
)


if fes_ids != iqs_ids:

    raise ValueError(
        "FES and IQS cohorts do not match."
    )

print(
    f"✅ Exact {len(fes_ids)}-video cohort match."
)


# ------------------------------------------------------------
# 5. SELECT REQUIRED FES COLUMNS
# ------------------------------------------------------------

fes_required = [
    "video_id",
    "domain",
    "query",
    "video_title",
    "final_engagement_score",
    "relevant_comment_count"
]

missing_fes = [
    col
    for col in fes_required
    if col not in fes_df.columns
]

if missing_fes:

    raise ValueError(
        f"Missing FES columns: {missing_fes}"
    )


fes_input = fes_df[
    fes_required
].copy()


# ------------------------------------------------------------
# 6. SELECT REQUIRED IQS COLUMNS
# ------------------------------------------------------------

iqs_required = [
    "video_id",
    "intrinsic_quality_score"
]

missing_iqs = [
    col
    for col in iqs_required
    if col not in iqs_df.columns
]

if missing_iqs:

    raise ValueError(
        f"Missing IQS columns: {missing_iqs}"
    )


iqs_input = iqs_df[
    iqs_required
].copy()


# ------------------------------------------------------------
# 7. MERGE FES + IQS
# ------------------------------------------------------------

hybrid_df = fes_input.merge(
    iqs_input,
    on="video_id",
    how="inner"
)


print(
    f"\nMerged records: "
    f"{len(hybrid_df)}"
)

print(
    f"Merged unique videos: "
    f"{hybrid_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 8. VALIDATE MERGE
# ------------------------------------------------------------

expected_n = len(fes_df)
expected_unique_n = fes_df["video_id"].nunique()

if len(hybrid_df) != expected_n:

    raise ValueError(
        f"Expected {expected_n} merged videos, "
        f"found {len(hybrid_df)}"
    )

if (
    hybrid_df["video_id"].nunique()
    != expected_unique_n
):

    raise ValueError(
        "Duplicate video IDs after merge."
    )

print(
    "✅ Hybrid cohort validation passed."
)


# ------------------------------------------------------------
# 9. NUMERIC CONVERSION
# ------------------------------------------------------------

hybrid_df[
    "final_engagement_score"
] = pd.to_numeric(
    hybrid_df[
        "final_engagement_score"
    ],
    errors="coerce"
)

hybrid_df[
    "intrinsic_quality_score"
] = pd.to_numeric(
    hybrid_df[
        "intrinsic_quality_score"
    ],
    errors="coerce"
)

hybrid_df[
    "relevant_comment_count"
] = pd.to_numeric(
    hybrid_df[
        "relevant_comment_count"
    ],
    errors="coerce"
)


# ------------------------------------------------------------
# 10. CHECK MISSING VALUES
# ------------------------------------------------------------

required_numeric = [
    "final_engagement_score",
    "intrinsic_quality_score",
    "relevant_comment_count"
]

missing_counts = (
    hybrid_df[
        required_numeric
    ]
    .isna()
    .sum()
)

print("\nMissing values:")

display(
    missing_counts.to_frame(
        "missing_count"
    )
)

if missing_counts.sum() > 0:

    raise ValueError(
        "Missing numeric values detected."
    )


# ------------------------------------------------------------
# 11. ENGAGEMENT STRENGTH
# ------------------------------------------------------------
#
# log1p(x) = log(1+x)
#
# This compresses the effect of very large
# comment counts while preserving ordering.
# ------------------------------------------------------------

hybrid_df[
    "engagement_strength"
] = np.log1p(
    hybrid_df[
        "relevant_comment_count"
    ]
)


# ------------------------------------------------------------
# 12. ADAPTIVE ALPHA
# ------------------------------------------------------------

K = 5.0

hybrid_df[
    "alpha"
] = (
    hybrid_df[
        "engagement_strength"
    ]
    /
    (
        hybrid_df[
            "engagement_strength"
        ]
        + K
    )
)


# ------------------------------------------------------------
# 13. IQS WEIGHT
# ------------------------------------------------------------

hybrid_df[
    "iqs_weight"
] = (
    1.0
    -
    hybrid_df[
        "alpha"
    ]
)


# ------------------------------------------------------------
# 14. COMPONENT CONTRIBUTIONS
# ------------------------------------------------------------

hybrid_df[
    "fes_contribution"
] = (
    hybrid_df[
        "alpha"
    ]
    *
    hybrid_df[
        "final_engagement_score"
    ]
)

hybrid_df[
    "iqs_contribution"
] = (
    hybrid_df[
        "iqs_weight"
    ]
    *
    hybrid_df[
        "intrinsic_quality_score"
    ]
)


# ------------------------------------------------------------
# 15. HYBRID SCORE
# ------------------------------------------------------------

hybrid_df[
    "hybrid_score"
] = (
    hybrid_df[
        "fes_contribution"
    ]
    +
    hybrid_df[
        "iqs_contribution"
    ]
)


# ------------------------------------------------------------
# 16. DOMINANT CONTRIBUTION
# ------------------------------------------------------------

hybrid_df[
    "dominant_component"
] = np.where(
    hybrid_df[
        "fes_contribution"
    ]
    >
    hybrid_df[
        "iqs_contribution"
    ],
    "FES",
    "IQS"
)


# ------------------------------------------------------------
# 17. GLOBAL FES RANK
# ------------------------------------------------------------

hybrid_df[
    "fes_rank"
] = (
    hybrid_df[
        "final_engagement_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 18. GLOBAL IQS RANK
# ------------------------------------------------------------

hybrid_df[
    "iqs_rank"
] = (
    hybrid_df[
        "intrinsic_quality_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 19. GLOBAL HYBRID RANK
# ------------------------------------------------------------

hybrid_df = hybrid_df.sort_values(
    "hybrid_score",
    ascending=False
).reset_index(
    drop=True
)

hybrid_df[
    "hybrid_rank"
] = np.arange(
    1,
    len(hybrid_df) + 1
)


# ------------------------------------------------------------
# 20. RANK MOVEMENT
# ------------------------------------------------------------
#
# Positive value:
#   video moved upward relative to source ranking
#
# Negative value:
#   video moved downward
#
# We calculate movement from both FES and IQS.
# ------------------------------------------------------------

hybrid_df[
    "movement_from_fes"
] = (
    hybrid_df[
        "fes_rank"
    ]
    -
    hybrid_df[
        "hybrid_rank"
    ]
)

hybrid_df[
    "movement_from_iqs"
] = (
    hybrid_df[
        "iqs_rank"
    ]
    -
    hybrid_df[
        "hybrid_rank"
    ]
)


# ------------------------------------------------------------
# 21. DOMAIN-WISE HYBRID RANK
# ------------------------------------------------------------

hybrid_df[
    "domain_hybrid_rank"
] = (
    hybrid_df
    .groupby("domain")[
        "hybrid_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 22. DOMAIN-WISE FES / IQS RANK
# ------------------------------------------------------------

hybrid_df[
    "domain_fes_rank"
] = (
    hybrid_df
    .groupby("domain")[
        "final_engagement_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)

hybrid_df[
    "domain_iqs_rank"
] = (
    hybrid_df
    .groupby("domain")[
        "intrinsic_quality_score"
    ]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 23. SAVE COMPLETE HYBRID DATASET
# ------------------------------------------------------------

full_output = os.path.join(
    OUTPUT_ROOT,
    "phase9f_hybrid_full.csv"
)

hybrid_df.to_csv(
    full_output,
    index=False
)


# ------------------------------------------------------------
# 24. SAVE TOP 10
# ------------------------------------------------------------

top10_output = os.path.join(
    OUTPUT_ROOT,
    "phase9f_hybrid_top10.csv"
)

hybrid_df.head(10).to_csv(
    top10_output,
    index=False
)


# ------------------------------------------------------------
# 25. SAVE TOP 20
# ------------------------------------------------------------

top20_output = os.path.join(
    OUTPUT_ROOT,
    "phase9f_hybrid_top20.csv"
)

hybrid_df.head(20).to_csv(
    top20_output,
    index=False
)


# ------------------------------------------------------------
# 26. DOMAIN SUMMARY
# ------------------------------------------------------------

domain_summary = (
    hybrid_df
    .groupby("domain")
    .agg(
        videos=(
            "video_id",
            "nunique"
        ),

        mean_fes=(
            "final_engagement_score",
            "mean"
        ),

        mean_iqs=(
            "intrinsic_quality_score",
            "mean"
        ),

        mean_hybrid=(
            "hybrid_score",
            "mean"
        ),

        std_hybrid=(
            "hybrid_score",
            "std"
        ),

        min_hybrid=(
            "hybrid_score",
            "min"
        ),

        max_hybrid=(
            "hybrid_score",
            "max"
        ),

        mean_alpha=(
            "alpha",
            "mean"
        ),

        mean_engagement_strength=(
            "engagement_strength",
            "mean"
        ),

        fes_dominant_videos=(
            "dominant_component",
            lambda x:
                (x == "FES").sum()
        ),

        iqs_dominant_videos=(
            "dominant_component",
            lambda x:
                (x == "IQS").sum()
        )
    )
    .reset_index()
)


domain_summary_output = os.path.join(
    OUTPUT_ROOT,
    "phase9f_domain_summary.csv"
)

domain_summary.to_csv(
    domain_summary_output,
    index=False
)


# ------------------------------------------------------------
# 27. ALPHA SUMMARY
# ------------------------------------------------------------

alpha_summary = pd.DataFrame({
    "statistic": [
        "count",
        "mean",
        "std",
        "min",
        "median",
        "max"
    ],

    "alpha": [
        hybrid_df["alpha"].count(),
        hybrid_df["alpha"].mean(),
        hybrid_df["alpha"].std(),
        hybrid_df["alpha"].min(),
        hybrid_df["alpha"].median(),
        hybrid_df["alpha"].max()
    ]
})


alpha_output = os.path.join(
    OUTPUT_ROOT,
    "phase9f_alpha_summary.csv"
)

alpha_summary.to_csv(
    alpha_output,
    index=False
)


# ------------------------------------------------------------
# 28. TOP 10 HYBRID RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("TOP 10 VIDEOS — ADAPTIVE HYBRID RANKING")
print("=" * 110)

display(
    hybrid_df[
        [
            "hybrid_rank",
            "domain",
            "video_id",
            "video_title",
            "final_engagement_score",
            "intrinsic_quality_score",
            "relevant_comment_count",
            "engagement_strength",
            "alpha",
            "fes_contribution",
            "iqs_contribution",
            "hybrid_score",
            "dominant_component"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 29. DOMAIN SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("DOMAIN-WISE HYBRID SUMMARY")
print("=" * 110)

display(
    domain_summary
)


# ------------------------------------------------------------
# 30. ALPHA SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("ADAPTIVE ALPHA SUMMARY")
print("=" * 110)

display(
    alpha_summary
)


# ------------------------------------------------------------
# 31. RANK MOVEMENT — TOP MOVERS UP
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("TOP 10 VIDEOS MOVING UP FROM FES RANK")
print("=" * 110)

display(
    hybrid_df[
        [
            "hybrid_rank",
            "fes_rank",
            "movement_from_fes",
            "domain",
            "video_id",
            "video_title",
            "final_engagement_score",
            "intrinsic_quality_score",
            "hybrid_score"
        ]
    ]
    .sort_values(
        "movement_from_fes",
        ascending=False
    )
    .head(10)
)


# ------------------------------------------------------------
# 32. RANK MOVEMENT — TOP MOVERS DOWN
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("TOP 10 VIDEOS MOVING DOWN FROM FES RANK")
print("=" * 110)

display(
    hybrid_df[
        [
            "hybrid_rank",
            "fes_rank",
            "movement_from_fes",
            "domain",
            "video_id",
            "video_title",
            "final_engagement_score",
            "intrinsic_quality_score",
            "hybrid_score"
        ]
    ]
    .sort_values(
        "movement_from_fes",
        ascending=True
    )
    .head(10)
)


# ------------------------------------------------------------
# 33. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("PHASE 9F — FINAL VALIDATION")
print("=" * 110)

validation_passed = True


# Record count
expected_n = len(fes_df)
expected_unique_n = fes_df["video_id"].nunique()

if len(hybrid_df) != expected_n:

    validation_passed = False

    print(
        f"❌ Expected {expected_n} records, "
        f"found {len(hybrid_df)}"
    )

else:

    print(
        f"✅ {expected_n} hybrid records"
    )


# Unique IDs
if (
    hybrid_df["video_id"].nunique()
    != expected_unique_n
):

    validation_passed = False

    print(
        "❌ Duplicate video IDs"
    )

else:

    print(
        f"✅ {expected_unique_n} unique video IDs"
    )


# Alpha range
if not hybrid_df[
    "alpha"
].between(
    0,
    1
).all():

    validation_passed = False

    print(
        "❌ Alpha outside [0,1]"
    )

else:

    print(
        "✅ Alpha values within [0,1]"
    )


# IQS weight range
if not hybrid_df[
    "iqs_weight"
].between(
    0,
    1
).all():

    validation_passed = False

    print(
        "❌ IQS weights outside [0,1]"
    )

else:

    print(
        "✅ IQS weights within [0,1]"
    )


# Hybrid finite
if not np.isfinite(
    hybrid_df[
        "hybrid_score"
    ]
).all():

    validation_passed = False

    print(
        "❌ Non-finite Hybrid scores"
    )

else:

    print(
        "✅ All Hybrid scores finite"
    )


# Hybrid missing
if hybrid_df[
    "hybrid_score"
].isna().sum() > 0:

    validation_passed = False

    print(
        "❌ Missing Hybrid scores"
    )

else:

    print(
        "✅ No missing Hybrid scores"
    )


# Rank integrity
if set(
    hybrid_df["hybrid_rank"]
) != set(
    range(
        1,
        100
    )
):

    validation_passed = False

    print(
        f"❌ Hybrid ranking is not 1–{expected_n}"
    )

else:

    print(
        f"✅ Hybrid ranking is complete 1–{expected_n}"
    )


# Contribution identity
contribution_error = np.max(
    np.abs(
        hybrid_df[
            "hybrid_score"
        ]
        -
        (
            hybrid_df[
                "fes_contribution"
            ]
            +
            hybrid_df[
                "iqs_contribution"
            ]
        )
    )
)

print(
    f"\nMaximum contribution error: "
    f"{contribution_error:.12f}"
)

if contribution_error > 1e-10:

    validation_passed = False

    print(
        "❌ Contribution identity failed"
    )

else:

    print(
        "✅ Hybrid = FES contribution + "
        "IQS contribution"
    )


# Alpha formula validation
expected_alpha = (
    hybrid_df[
        "engagement_strength"
    ]
    /
    (
        hybrid_df[
            "engagement_strength"
        ]
        + K
    )
)

alpha_error = np.max(
    np.abs(
        hybrid_df[
            "alpha"
        ]
        -
        expected_alpha
    )
)

print(
    f"Maximum alpha error: "
    f"{alpha_error:.12f}"
)

if alpha_error > 1e-10:

    validation_passed = False

    print(
        "❌ Alpha formula validation failed"
    )

else:

    print(
        "✅ Alpha formula validated"
    )


# ------------------------------------------------------------
# 34. FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 110)

if validation_passed:

    print(
        "✅ PHASE 9F VALIDATION PASSED"
    )

    print(
        "Adaptive Hybrid Ranking successfully "
        f"computed for all {expected_n} videos."
    )

else:

    print(
        "⚠️ PHASE 9F VALIDATION REQUIRES REVIEW"
    )

print("=" * 110)


# ------------------------------------------------------------
# 35. OUTPUT PATHS
# ------------------------------------------------------------

print(
    f"\nFull Hybrid file:\n{full_output}"
)

print(
    f"\nTop-10 file:\n{top10_output}"
)

print(
    f"\nTop-20 file:\n{top20_output}"
)

print(
    f"\nDomain summary:\n{domain_summary_output}"
)

print(
    f"\nAlpha summary:\n{alpha_output}"
)
````

````python
# ============================================================
# PHASE 9G-A — INDEPENDENT HUMAN EVALUATION DATASET
# ============================================================
#
# PURPOSE:
#   Create a blinded annotation sheet for independent
#   human ground-truth evaluation.
#
# IMPORTANT:
#   FES / IQS / Hybrid scores are NOT included.
#   Human labels must be independent of our system scores.
#
# Relevance:
#   0 = Not relevant
#   1 = Marginally relevant
#   2 = Clearly relevant
#   3 = Highly relevant
#
# Educational Quality:
#   1 = Very poor
#   2 = Poor
#   3 = Adequate
#   4 = Good
#   5 = Excellent
#
# ============================================================

import os
import uuid
import hashlib
import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

FES_PATH = (
    "/content/revision_phase9d/"
    "phase9d_fes_full.csv"
)

OUTPUT_ROOT = (
    "/content/revision_phase9g"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. LOAD VIDEO METADATA
# ------------------------------------------------------------

fes_df = pd.read_csv(
    FES_PATH
)

print(
    f"Loaded records: {len(fes_df)}"
)

print(
    f"Unique videos: "
    f"{fes_df['video_id'].nunique()}"
)


# ------------------------------------------------------------
# 3. REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "video_id",
    "domain",
    "query",
    "video_title",
    "description"
]

missing_columns = [
    c
    for c in required_columns
    if c not in fes_df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )


# ------------------------------------------------------------
# 4. VALIDATE COHORT
# ------------------------------------------------------------

expected_n = len(fes_df)
expected_unique_n = fes_df["video_id"].nunique()

if len(fes_df) != expected_n:

    raise ValueError(
        f"Expected {expected_n} videos, found {len(fes_df)}"
    )

if fes_df[
    "video_id"
].nunique() != expected_unique_n:

    raise ValueError(
        "Video IDs are not unique."
    )

print(
    f"✅ {expected_n}-video evaluation cohort validated."
)


# ------------------------------------------------------------
# 5. BUILD BLINDED DATASET
# ------------------------------------------------------------
#
# We intentionally exclude:
#
#   final_engagement_score
#   intrinsic_quality_score
#   hybrid_score
#   alpha
#   automated ranking
#
# ------------------------------------------------------------

annotation_df = fes_df[
    [
        "video_id",
        "domain",
        "query",
        "video_title",
        "description"
    ]
].copy()


# ------------------------------------------------------------
# 6. CREATE ANONYMOUS EVALUATION ID
# ------------------------------------------------------------
#
# The evaluator works with an anonymous ID rather than
# the actual YouTube video ID.
#
# This reduces direct association with our system.
# ------------------------------------------------------------

def make_eval_id(video_id):

    digest = hashlib.sha256(
        str(video_id).encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "EV"
        + digest[:10].upper()
    )


annotation_df[
    "evaluation_id"
] = (
    annotation_df[
        "video_id"
    ]
    .apply(make_eval_id)
)


# ------------------------------------------------------------
# 7. SHUFFLE EVALUATION ORDER
# ------------------------------------------------------------

RANDOM_SEED = 2026

annotation_df = annotation_df.sample(
    frac=1,
    random_state=RANDOM_SEED
).reset_index(
    drop=True
)

annotation_df[
    "evaluation_order"
] = np.arange(
    1,
    len(annotation_df) + 1
)


# ------------------------------------------------------------
# 8. ADD HUMAN-ANNOTATION COLUMNS
# ------------------------------------------------------------

annotation_df[
    "relevance_score"
] = ""

annotation_df[
    "educational_quality_score"
] = ""

annotation_df[
    "justification"
] = ""


# ------------------------------------------------------------
# 9. ADD ANNOTATOR INFORMATION
# ------------------------------------------------------------

annotation_df[
    "annotator_id"
] = ""

annotation_df[
    "evaluation_date"
] = ""


# ------------------------------------------------------------
# 10. COLUMN ORDER
# ------------------------------------------------------------

annotation_df = annotation_df[
    [
        "evaluation_order",
        "evaluation_id",
        "domain",
        "query",
        "video_title",
        "description",
        "relevance_score",
        "educational_quality_score",
        "justification",
        "annotator_id",
        "evaluation_date"
    ]
]


# ------------------------------------------------------------
# 11. SAVE BLINDED ANNOTATION SHEET
# ------------------------------------------------------------

annotation_path = os.path.join(
    OUTPUT_ROOT,
    "phase9g_human_annotation_sheet.csv"
)

annotation_df.to_csv(
    annotation_path,
    index=False
)


# ------------------------------------------------------------
# 12. SAVE CODEBOOK
# ------------------------------------------------------------

codebook = pd.DataFrame({
    "field": [
        "relevance_score",
        "educational_quality_score",
        "justification"
    ],

    "allowed_values": [
        "0, 1, 2, 3",
        "1, 2, 3, 4, 5",
        "Free text"
    ],

    "meaning": [
        (
            "0=Not relevant; "
            "1=Marginally relevant; "
            "2=Clearly relevant; "
            "3=Highly relevant"
        ),
        (
            "1=Very poor; "
            "2=Poor; "
            "3=Adequate; "
            "4=Good; "
            "5=Excellent"
        ),
        (
            "Brief reason for the assigned scores"
        )
    ]
})

codebook_path = os.path.join(
    OUTPUT_ROOT,
    "phase9g_annotation_codebook.csv"
)

codebook.to_csv(
    codebook_path,
    index=False
)


# ------------------------------------------------------------
# 13. DOMAIN DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9G-A — EVALUATION COHORT")
print("=" * 75)

display(
    annotation_df[
        "domain"
    ]
    .value_counts()
    .sort_index()
    .to_frame(
        "videos"
    )
)


# ------------------------------------------------------------
# 14. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 9G-A — VALIDATION")
print("=" * 75)

validation_passed = True


if len(annotation_df) != expected_n:

    validation_passed = False

    print(
        f"❌ Expected {expected_n} videos."
    )

else:

    print(
        f"✅ {expected_n} videos present."
    )


if annotation_df[
    "evaluation_id"
].nunique() != expected_unique_n:

    validation_passed = False

    print(
        "❌ Duplicate evaluation IDs."
    )

else:

    print(
        f"✅ {expected_unique_n} unique evaluation IDs."
    )


# Ensure system scores are not present
for forbidden_column in [
    "final_engagement_score",
    "intrinsic_quality_score",
    "hybrid_score",
    "alpha",
    "fes_rank",
    "iqs_rank",
    "hybrid_rank"
]:

    if forbidden_column in annotation_df.columns:

        validation_passed = False

        print(
            f"❌ System score leaked: "
            f"{forbidden_column}"
        )


if validation_passed:

    print(
        "✅ No automated ranking scores "
        "are present in the annotation sheet."
    )


print("\n" + "=" * 75)

if validation_passed:

    print(
        "✅ PHASE 9G-A VALIDATION PASSED"
    )

else:

    print(
        "⚠️ PHASE 9G-A REQUIRES REVIEW"
    )

print("=" * 75)


print(
    f"\nAnnotation sheet:\n"
    f"{annotation_path}"
)

print(
    f"\nCodebook:\n"
    f"{codebook_path}"
)
````

````python
# ============================================================
# PHASE 9F — COHORT MANIFEST / FREEZE POINT
# ============================================================
# This cell freezes the exact Phase 9 cohort used by all downstream
# ablation and sensitivity analyses. It prevents accidental mixing of
# outputs from different YouTube retrieval runs.

from pathlib import Path
import hashlib
import json
import pandas as pd

phase9_files = {
    "9C": Path("/content/revision_phase9c/all_domains_video_emotional_engagement.csv"),
    "9D": Path("/content/revision_phase9d/phase9d_fes_full.csv"),
    "9E": Path("/content/revision_phase9e/phase9e_iqs_full.csv"),
    "9F": Path("/content/revision_phase9f/phase9f_hybrid_full.csv"),
}

for label, path in phase9_files.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing {label} artifact: {path}")

def normalized_ids(df):
    if "video_id" not in df.columns:
        raise ValueError("Artifact is missing video_id.")
    ids = df["video_id"].astype(str).str.strip()
    if ids.duplicated().any():
        raise ValueError("Duplicate video IDs detected in cohort artifact.")
    return set(ids)

p9c = pd.read_csv(phase9_files["9C"])
p9d = pd.read_csv(phase9_files["9D"])
p9e = pd.read_csv(phase9_files["9E"])
p9f = pd.read_csv(phase9_files["9F"])

cohort_ids = normalized_ids(p9c)
for label, df in [("9D", p9d), ("9E", p9e), ("9F", p9f)]:
    if normalized_ids(df) != cohort_ids:
        raise ValueError(f"{label} cohort does not exactly match Phase 9C.")

expected_domains = {"Programming", "Mathematics", "Physics", "Biology", "Data Science"}
domain_counts = p9c.groupby("domain")["video_id"].nunique().to_dict()
if set(domain_counts) != expected_domains:
    raise ValueError(f"Unexpected domain set: {domain_counts}")
if any(v < 1 or v > 20 for v in domain_counts.values()):
    raise ValueError(f"Domain count outside 1–20: {domain_counts}")

# Stable fingerprint based on video IDs and the final Phase 9F hybrid scores.
fingerprint_df = p9f[["video_id", "hybrid_score"]].copy()
fingerprint_df["video_id"] = fingerprint_df["video_id"].astype(str).str.strip()
fingerprint_df = fingerprint_df.sort_values("video_id")
fingerprint_payload = fingerprint_df.to_json(orient="records", double_precision=15)
cohort_fingerprint = hashlib.sha256(fingerprint_payload.encode("utf-8")).hexdigest()

manifest = {
    "cohort_size": len(cohort_ids),
    "domain_counts": domain_counts,
    "cohort_fingerprint": cohort_fingerprint,
    "phase9c_rows": len(p9c),
    "phase9d_rows": len(p9d),
    "phase9e_rows": len(p9e),
    "phase9f_rows": len(p9f),
}

manifest_path = Path("/content/revision_phase9f/phase9_cohort_manifest.json")
manifest_path.write_text(json.dumps(manifest, indent=2), encoding="utf-8")

print("=" * 90)
print("PHASE 9F COHORT MANIFEST")
print("=" * 90)
print(f"Frozen cohort size : {manifest['cohort_size']}")
print(f"Domain counts      : {domain_counts}")
print(f"Cohort fingerprint : {cohort_fingerprint}")
print(f"Saved              : {manifest_path}")
print("\nDownstream Phase 9H–9K must use this exact cohort.")
````

````python
# ============================================================
# PHASE 9H — TRUE ABLATION STUDY
# ============================================================
#
# Ablation variants:
#   1. Full adaptive hybrid
#   2. IQS-only / no engagement
#   3. FES-only / no IQS
#   4. Fixed 50/50 FES + IQS
#
# Uses the actual Phase 9F column:
#   intrinsic_quality_score
#
# Cohort size is derived dynamically.
# ============================================================

import os
import numpy as np
import pandas as pd
from scipy.stats import spearmanr

# ------------------------------------------------------------
# 1. LOAD PHASE 9F
# ------------------------------------------------------------

HYBRID_PATH = (
    "/content/revision_phase9f/"
    "phase9f_hybrid_full.csv"
)

hybrid_df = pd.read_csv(HYBRID_PATH)

print("=" * 90)
print("PHASE 9H — TRUE ABLATION STUDY")
print("=" * 90)

print(f"Phase 9F rows: {len(hybrid_df)}")
print(
    f"Unique video IDs: "
    f"{hybrid_df['video_id'].nunique()}"
)

# ------------------------------------------------------------
# 2. REQUIRED COLUMNS
# ------------------------------------------------------------

required = [
    "video_id",
    "final_engagement_score",
    "intrinsic_quality_score",
    "hybrid_score",
    "alpha",
    "relevant_comment_count"
]

missing = [
    c for c in required
    if c not in hybrid_df.columns
]

if missing:
    raise ValueError(
        f"Missing required Phase 9F columns: {missing}"
    )

# ------------------------------------------------------------
# 3. COHORT VALIDATION
# ------------------------------------------------------------

if hybrid_df["video_id"].duplicated().any():
    raise ValueError(
        "Duplicate video IDs detected in Phase 9F."
    )

N_VIDEOS = hybrid_df["video_id"].nunique()

if len(hybrid_df) != N_VIDEOS:
    raise ValueError(
        "Phase 9F contains duplicate video rows."
    )

print(
    f"Dynamic cohort size: {N_VIDEOS} videos"
)

if not np.isfinite(hybrid_df[required[1:4]]).all().all():
    raise ValueError("Non-finite Phase 9F score values detected.")

# ------------------------------------------------------------
# 4. RECONSTRUCT COMPONENT SCORES
# ------------------------------------------------------------

hybrid_df["iqs_only"] = (
    hybrid_df["intrinsic_quality_score"]
)

hybrid_df["fes_only"] = (
    hybrid_df["final_engagement_score"]
)

# ------------------------------------------------------------
# 5. FIXED 50/50 HYBRID
# ------------------------------------------------------------

hybrid_df["fixed_50_50"] = (
    0.5 * hybrid_df["final_engagement_score"]
    +
    0.5 * hybrid_df["intrinsic_quality_score"]
)

# ------------------------------------------------------------
# 6. NO-ENGAGEMENT VARIANT
# ------------------------------------------------------------
#
# Ranking driven entirely by educational quality.
# ------------------------------------------------------------

hybrid_df["no_engagement"] = (
    hybrid_df["intrinsic_quality_score"]
)

# ------------------------------------------------------------
# 7. NO-IQS VARIANT
# ------------------------------------------------------------
#
# Ranking driven entirely by engagement.
# ------------------------------------------------------------

hybrid_df["no_iqs"] = (
    hybrid_df["final_engagement_score"]
)

# ------------------------------------------------------------
# 8. NO-ADAPTIVE-WEIGHTING VARIANT
# ------------------------------------------------------------

hybrid_df["fixed_weight_hybrid"] = (
    hybrid_df["fixed_50_50"]
)

# ------------------------------------------------------------
# 9. RANK ALL VARIANTS
# ------------------------------------------------------------

score_columns = [
    "hybrid_score",
    "no_engagement",
    "no_iqs",
    "fixed_weight_hybrid",
    "iqs_only",
    "fes_only"
]

for column in score_columns:

    hybrid_df[f"{column}_rank"] = (
        hybrid_df[column]
        .rank(
            method="min",
            ascending=False
        )
        .astype(int)
    )

# ------------------------------------------------------------
# 10. ABLATION COMPARISON
# ------------------------------------------------------------

ablation_rows = []

for variant in score_columns[1:]:

    rho, p_value = spearmanr(
        hybrid_df["hybrid_score"],
        hybrid_df[variant]
    )

    movement = (
        hybrid_df[f"{variant}_rank"]
        -
        hybrid_df["hybrid_score_rank"]
    )

    full_top10 = set(
        hybrid_df
        .nlargest(
            10,
            "hybrid_score"
        )["video_id"]
    )

    variant_top10 = set(
        hybrid_df
        .nlargest(
            10,
            variant
        )["video_id"]
    )

    top10_overlap = len(
        full_top10 & variant_top10
    )

    ablation_rows.append({

        "cohort_size": N_VIDEOS,
        "variant": variant,

        "spearman_vs_full": rho,

        "p_value": p_value,

        "mean_abs_rank_change":
            movement.abs().mean(),

        "max_abs_rank_change":
            movement.abs().max(),

        "top10_overlap":
            top10_overlap,

        "top10_overlap_fraction":
            top10_overlap / 10.0

    })

# ------------------------------------------------------------
# 11. SAVE RESULTS
# ------------------------------------------------------------

OUTPUT_ROOT = (
    "/content/revision_phase9h"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

ablation_results = pd.DataFrame(
    ablation_rows
)

ablation_results.to_csv(
    os.path.join(
        OUTPUT_ROOT,
        "true_ablation_results.csv"
    ),
    index=False
)

hybrid_df.to_csv(
    os.path.join(
        OUTPUT_ROOT,
        "ablation_rankings_full.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 12. DISPLAY RESULTS
# ------------------------------------------------------------

print()
print("=" * 90)
print("ABLATION RESULTS")
print("=" * 90)

print(
    ablation_results.to_string(
        index=False
    )
)

print()
print("=" * 90)
print("SAVED OUTPUTS")
print("=" * 90)

print(
    os.path.join(
        OUTPUT_ROOT,
        "true_ablation_results.csv"
    )
)

print(
    os.path.join(
        OUTPUT_ROOT,
        "ablation_rankings_full.csv"
    )
)

print()
print(
    "✅ PHASE 9H TRUE ABLATION COMPLETED"
)
````

````python
# ============================================================
# PHASE 9I — WEIGHT SENSITIVITY ANALYSIS
# ============================================================
#
# Vary the adaptive-weight smoothing parameter k and measure
# ranking stability relative to the final k=5 configuration.
#
# Hybrid(k) =
#     alpha_k * FES
#     + (1 - alpha_k) * IQS
#
# where:
#     EngagementStrength = log1p(relevant_comment_count)
#     alpha_k = EngagementStrength /
#               (EngagementStrength + k)
#
# ============================================================

import os
import numpy as np
import pandas as pd
from scipy.stats import spearmanr

# ------------------------------------------------------------
# 1. VALIDATE REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "video_id",
    "final_engagement_score",
    "intrinsic_quality_score",
    "relevant_comment_count",
    "hybrid_score"
]

missing = [
    c for c in required_columns
    if c not in hybrid_df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns for Phase 9I: {missing}"
    )

# ------------------------------------------------------------
# 2. COHORT VALIDATION
# ------------------------------------------------------------

if hybrid_df["video_id"].duplicated().any():
    raise ValueError(
        "Duplicate video IDs detected in hybrid_df."
    )

N_VIDEOS = hybrid_df["video_id"].nunique()

print("=" * 90)
print("PHASE 9I — WEIGHT SENSITIVITY ANALYSIS")
print("=" * 90)
print(f"Videos analysed: {N_VIDEOS}")

# ------------------------------------------------------------
# 3. k VALUES
# ------------------------------------------------------------

k_values = [
    1, 2, 3, 5,
    7, 10, 15, 20
]

sensitivity_rows = []

# ------------------------------------------------------------
# 4. BASE ENGAGEMENT STRENGTH
# ------------------------------------------------------------

engagement_strength = np.log1p(
    pd.to_numeric(
        hybrid_df["relevant_comment_count"],
        errors="coerce"
    )
)

if engagement_strength.isna().any():
    raise ValueError(
        "Invalid relevant_comment_count values detected."
    )

# ------------------------------------------------------------
# 5. TEST EACH k
# ------------------------------------------------------------

for k in k_values:

    # Adaptive engagement weight
    alpha_k = (
        engagement_strength
        /
        (engagement_strength + k)
    )

    # IQS receives the remaining weight
    iqs_weight_k = 1.0 - alpha_k

    # --------------------------------------------------------
    # Hybrid score for this k
    # --------------------------------------------------------

    hybrid_k = (
        alpha_k
        * hybrid_df["final_engagement_score"]
        +
        iqs_weight_k
        * hybrid_df["intrinsic_quality_score"]
    )

    # --------------------------------------------------------
    # Ranking
    # --------------------------------------------------------

    rank_k = (
        hybrid_k
        .rank(
            method="first",
            ascending=False
        )
        .astype(int)
    )

    # --------------------------------------------------------
    # Compare with final k=5 ranking
    # --------------------------------------------------------

    rho, p_value = spearmanr(
        hybrid_df["hybrid_score"],
        hybrid_k
    )

    # --------------------------------------------------------
    # Rank movement
    # --------------------------------------------------------

    final_rank = (
        hybrid_df["hybrid_score"]
        .rank(
            method="first",
            ascending=False
        )
        .astype(int)
    )

    rank_change = (
        rank_k - final_rank
    )

    # --------------------------------------------------------
    # Top-10 overlap
    # --------------------------------------------------------

    final_top10 = set(
        hybrid_df
        .loc[
            hybrid_df["hybrid_score"]
            .nlargest(10)
            .index,
            "video_id"
        ]
    )

    k_top10 = set(
        hybrid_df
        .loc[
            hybrid_k
            .nlargest(10)
            .index,
            "video_id"
        ]
    )

    top10_overlap = len(
        final_top10 & k_top10
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    sensitivity_rows.append({

        "cohort_size": N_VIDEOS,

        "k": k,

        "mean_alpha":
            alpha_k.mean(),

        "min_alpha":
            alpha_k.min(),

        "max_alpha":
            alpha_k.max(),

        "mean_iqs_weight":
            iqs_weight_k.mean(),

        "spearman_vs_final_k5":
            rho,

        "p_value":
            p_value,

        "mean_abs_rank_change":
            rank_change.abs().mean(),

        "max_abs_rank_change":
            rank_change.abs().max(),

        "top10_overlap":
            top10_overlap,

        "top10_overlap_fraction":
            top10_overlap / 10.0
    })

# ------------------------------------------------------------
# 6. CREATE RESULT TABLE
# ------------------------------------------------------------

sensitivity_df = pd.DataFrame(
    sensitivity_rows
)

# ------------------------------------------------------------
# 7. SAVE
# ------------------------------------------------------------

OUTPUT_ROOT = (
    "/content/revision_phase9h"
)

os.makedirs(
    OUTPUT_ROOT,
    exist_ok=True
)

SENSITIVITY_PATH = os.path.join(
    OUTPUT_ROOT,
    "weight_sensitivity.csv"
)

sensitivity_df.to_csv(
    SENSITIVITY_PATH,
    index=False
)

# ------------------------------------------------------------
# 8. DISPLAY
# ------------------------------------------------------------

print()
print("=" * 90)
print("WEIGHT SENSITIVITY RESULTS")
print("=" * 90)

print(
    sensitivity_df.to_string(
        index=False
    )
)

print()
print("=" * 90)
print("PHASE 9I COMPLETED")
print("=" * 90)

reference_row = sensitivity_df[sensitivity_df["k"] == 5]
if len(reference_row) != 1:
    raise ValueError("Sensitivity output must contain exactly one k=5 row.")
if not np.isclose(float(reference_row["spearman_vs_final_k5"].iloc[0]), 1.0, atol=1e-12):
    raise ValueError("k=5 sensitivity row is not identical to the Phase 9F reference ranking.")

print(
    f"Reference configuration: k = 5"
)

print(
    f"Output: {SENSITIVITY_PATH}"
)

print(
    "✅ WEIGHT SENSITIVITY ANALYSIS COMPLETED"
)
````

````python
# ============================================================
# PHASE 9K — STATISTICAL SIGNIFICANCE OF RANKING DIFFERENCES
# ============================================================
#
# These tests compare paired per-video component scores/rankings.
# They are NOT presented as ground-truth effectiveness tests.
# They quantify whether component removals produce systematic
# score differences across the same per-video cohort.
# ============================================================

from scipy.stats import wilcoxon

significance_rows = []

for variant in ["no_engagement", "no_iqs", "fixed_weight_hybrid"]:
    stat, p = wilcoxon(
        hybrid_df["hybrid_score"],
        hybrid_df[variant],
        alternative="two-sided"
    )

    significance_rows.append({
        "cohort_size": len(hybrid_df),
        "comparison": f"Full Hybrid vs {variant}",
        "wilcoxon_statistic": stat,
        "p_value": p
    })

significance_df = pd.DataFrame(significance_rows)

if significance_df["p_value"].isna().any():
    raise ValueError("Phase 9K produced missing p-values.")

significance_df.to_csv(
    "/content/revision_phase9h/paired_significance_tests.csv",
    index=False
)

print(significance_df.to_string(index=False))
print()
print("Saved: /content/revision_phase9h/paired_significance_tests.csv")
````

# PHASE 10A — LAK 2026 BENCHMARK PREPARATION (RESTART-SAFE)

The benchmark is reused from the validated CSV when available. If it is absent,
the notebook restores/extracts `db.sqlite3` from the Drive backup and rebuilds the
same expert-labeled evaluation CSV. No manual upload or third-party JavaScript
reverse-engineering is required for normal execution.

````python
# ============================================================================
# PHASE 10A — RESTART-SAFE LAK 2026 BENCHMARK PREPARATION
# ============================================================================
#
# Priority:
#   1. Reuse an already validated /content/revision_phase10/lak2026_benchmark_eval.csv
#   2. Otherwise restore/extract db.sqlite3.zip and build the clean CSV.
#
# No manual file upload and no reverse-engineering of a third-party web bundle
# is required for a normal run.
# ============================================================================

from pathlib import Path
import sqlite3
import zipfile
import pandas as pd
import re
import shutil
import os

BASE = Path("/content/revision_phase10")
BASE.mkdir(parents=True, exist_ok=True)

OUTPUT = BASE / "lak2026_benchmark_eval.csv"
DB_ROOT = Path("/content/lak2026_benchmark")
DB_PATH = DB_ROOT / "db.sqlite3"
ZIP_PATH = Path("/content/db.sqlite3.zip")

print("=" * 100)
print("PHASE 10A — LAK 2026 BENCHMARK PREPARATION")
print("=" * 100)

# ---------------------------------------------------------------------------
# 1. If the clean validated benchmark already exists, reuse it.
# ---------------------------------------------------------------------------
if OUTPUT.exists():
    benchmark_eval = pd.read_csv(OUTPUT)

    required = [
        "topic_id", "domain", "topic",
        "video_id", "content", "relevance"
    ]

    missing = [c for c in required if c not in benchmark_eval.columns]

    if missing:
        raise RuntimeError(
            f"Existing benchmark CSV is missing required columns: {missing}"
        )

    print("✅ Reusing existing clean benchmark:")
    print("   ", OUTPUT)

else:
    # -----------------------------------------------------------------------
    # 2. Extract database if necessary.
    # -----------------------------------------------------------------------
    if not DB_PATH.exists():
        if ZIP_PATH.exists():
            print("Extracting:", ZIP_PATH)
            DB_ROOT.mkdir(parents=True, exist_ok=True)

            with zipfile.ZipFile(ZIP_PATH, "r") as z:
                z.extractall(DB_ROOT)

            # Some archives contain a nested db.sqlite3.
            candidates = list(DB_ROOT.rglob("db.sqlite3"))

            if candidates:
                if candidates[0] != DB_PATH:
                    shutil.copy2(candidates[0], DB_PATH)
            else:
                raise FileNotFoundError(
                    "db.sqlite3.zip was found, but no db.sqlite3 was extracted."
                )
        else:
            raise FileNotFoundError(
                "LAK benchmark data is unavailable after restart. "
                "Either keep /content/revision_phase10/lak2026_benchmark_eval.csv "
                "in the Drive backup or back up db.sqlite3.zip before restarting."
            )

    # -----------------------------------------------------------------------
    # 3. Build the clean expert-labeled benchmark.
    # -----------------------------------------------------------------------
    conn = sqlite3.connect(DB_PATH)

    benchmark = pd.read_sql_query(
        """
        SELECT
            t.id AS topic_id,
            c.title AS domain,
            t.title AS topic,
            r.id AS resource_id,
            r.url,
            r.content,
            r.resource_order,
            trl.status
        FROM TopicResourceLink trl
        JOIN Topics t
            ON t.id = trl.topic_id
        JOIN Courses c
            ON c.id = t.course_id
        JOIN Resources r
            ON r.id = trl.resource_id
        WHERE trl.status IN ('Accepted', 'Rejected')
        ORDER BY t.id, r.resource_order
        """,
        conn
    )

    conn.close()

    benchmark["video_id"] = benchmark["url"].astype(str).str.extract(
        r"[?&]v=([A-Za-z0-9_-]{11})",
        expand=False
    )

    benchmark["has_content"] = (
        benchmark["content"].notna()
        & benchmark["content"].astype(str).str.strip().ne("")
    )

    benchmark_eval = benchmark[
        benchmark["has_content"]
        & benchmark["video_id"].notna()
    ].copy()

    benchmark_eval["relevance"] = (
        benchmark_eval["status"] == "Accepted"
    ).astype(int)

    # Keep only topics with both positive and negative expert labels.
    topic_counts = benchmark_eval.groupby("topic_id")["relevance"].agg(
        positives="sum",
        total="count"
    )
    topic_counts["negatives"] = (
        topic_counts["total"] - topic_counts["positives"]
    )

    usable_topics = topic_counts.index[
        (topic_counts["positives"] > 0)
        & (topic_counts["negatives"] > 0)
    ]

    benchmark_eval = benchmark_eval[
        benchmark_eval["topic_id"].isin(usable_topics)
    ].copy()

    benchmark_eval.to_csv(OUTPUT, index=False)

    print("✅ Benchmark CSV created:", OUTPUT)

print("\nBenchmark summary:")
print("Pairs :", len(benchmark_eval))
print("Topics:", benchmark_eval["topic_id"].nunique())
print("Videos:", benchmark_eval["video_id"].nunique())
print("Statuses:")
print(benchmark_eval["relevance"].value_counts().sort_index())

print("\nDomains:")
print(benchmark_eval["domain"].value_counts().to_string())

print("\n✅ PHASE 10A READY")
````

````python
# ============================================================================
# PHASE 10A — FINAL BENCHMARK SANITY CHECK
# ============================================================================

from pathlib import Path
import pandas as pd

path = Path("/content/revision_phase10/lak2026_benchmark_eval.csv")

if not path.exists():
    raise FileNotFoundError(f"Benchmark CSV not found: {path}")

benchmark_eval = pd.read_csv(path)

required = [
    "topic_id", "domain", "topic",
    "video_id", "content", "relevance"
]

missing = [c for c in required if c not in benchmark_eval.columns]

if missing:
    raise ValueError(f"Missing benchmark columns: {missing}")

if benchmark_eval["video_id"].isna().any():
    raise ValueError("Benchmark contains missing video IDs.")

topic_counts = benchmark_eval.groupby("topic_id")["relevance"].agg(
    positives="sum",
    total="count"
)
topic_counts["negatives"] = topic_counts["total"] - topic_counts["positives"]

if (
    (topic_counts["positives"] == 0).any()
    or (topic_counts["negatives"] == 0).any()
):
    raise ValueError("Benchmark contains a topic without both relevance classes.")

print("=" * 90)
print("PHASE 10A SANITY CHECK PASSED")
print("=" * 90)
print(f"Topics : {benchmark_eval['topic_id'].nunique()}")
print(f"Pairs  : {len(benchmark_eval)}")
print(f"Videos : {benchmark_eval['video_id'].nunique()}")
````

# PHASE 10B — BM25 + E5 EXTERNAL BENCHMARK BASELINES

Evaluate BM25 and E5 against the independent LAK expert labels using
NDCG@5/10, MAP@10, MRR, Precision@5/10, and Recall@5/10.

The active Phase 10B code is self-contained and uses the benchmark CSV produced
by Phase 10A.

````python
from sklearn.metrics import ndcg_score
from scipy.stats import spearmanr
# ============================================================
# PHASE 10B — BM25 + E5 EXTERNAL BENCHMARK BASELINES
# ============================================================

import os
import re
import numpy as np
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm

from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer


# ============================================================
# PATHS
# ============================================================

BASE = Path("/content/revision_phase10")
INPUT = BASE / "lak2026_benchmark_eval.csv"

if not INPUT.exists():
    raise FileNotFoundError(
        f"Benchmark file not found: {INPUT}"
    )

benchmark = pd.read_csv(INPUT)

print("=" * 90)
print("PHASE 10B — BM25 + E5 EXTERNAL BENCHMARK")
print("=" * 90)

print("Loaded pairs:", len(benchmark))
print("Topics:", benchmark["topic_id"].nunique())
print("Unique videos:", benchmark["video_id"].nunique())


# ============================================================
# VALIDATE REQUIRED COLUMNS
# ============================================================

required_columns = [
    "topic_id",
    "domain",
    "topic",
    "video_id",
    "content",
    "relevance",
]

missing = [
    c for c in required_columns
    if c not in benchmark.columns
]

if missing:
    raise RuntimeError(
        f"Missing required benchmark columns: {missing}"
    )


# ============================================================
# CLEAN TEXT
# ============================================================

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


benchmark["content_clean"] = (
    benchmark["content"]
    .apply(clean_text)
)


# ============================================================
# REMOVE EMPTY DOCUMENTS
# ============================================================

before = len(benchmark)

benchmark = benchmark[
    benchmark["content_clean"].str.len() > 0
].copy()

after = len(benchmark)

print("\nContent filtering:")
print("Before:", before)
print("After :", after)
print("Removed:", before - after)


# ============================================================
# CHECK TOPIC LABEL VALIDITY
# ============================================================

topic_check = (
    benchmark
    .groupby("topic_id")
    .agg(
        positives=("relevance", "sum"),
        total=("relevance", "count")
    )
)

topic_check["negatives"] = (
    topic_check["total"] -
    topic_check["positives"]
)

invalid_topics = topic_check[
    (topic_check["positives"] == 0) |
    (topic_check["negatives"] == 0)
]

if len(invalid_topics) > 0:
    print(
        "\nWARNING — topics without both positive and negative labels:"
    )
    print(invalid_topics)

    benchmark = benchmark[
        ~benchmark["topic_id"].isin(
            invalid_topics.index
        )
    ].copy()


print("\nFinal evaluation topics:", benchmark["topic_id"].nunique())
print("Final evaluation pairs:", len(benchmark))
print("Final unique videos:", benchmark["video_id"].nunique())


# ============================================================
# METRIC FUNCTIONS
# ============================================================

def dcg(relevances, k):
    relevances = np.asarray(
        relevances[:k],
        dtype=float
    )

    if len(relevances) == 0:
        return 0.0

    discounts = np.log2(
        np.arange(2, len(relevances) + 2)
    )

    return np.sum(
        (2 ** relevances - 1) / discounts
    )


def ndcg_at_k(y_true, ranking, k):

    ranked_relevance = [
        y_true.get(video_id, 0)
        for video_id in ranking[:k]
    ]

    ideal_relevance = sorted(
        y_true.values(),
        reverse=True
    )[:k]

    ideal_dcg = dcg(
        ideal_relevance,
        k
    )

    if ideal_dcg == 0:
        return np.nan

    return (
        dcg(ranked_relevance, k)
        / ideal_dcg
    )


def precision_at_k(y_true, ranking, k):

    ranking = ranking[:k]

    if not ranking:
        return np.nan

    hits = sum(
        y_true.get(video_id, 0)
        for video_id in ranking
    )

    return hits / len(ranking)


def recall_at_k(y_true, ranking, k):

    ranking = ranking[:k]

    total_relevant = sum(
        y_true.values()
    )

    if total_relevant == 0:
        return np.nan

    hits = sum(
        y_true.get(video_id, 0)
        for video_id in ranking
    )

    return hits / total_relevant


def average_precision_at_k(
    y_true,
    ranking,
    k
):

    total_relevant = sum(
        y_true.values()
    )

    if total_relevant == 0:
        return np.nan

    hits = 0
    precision_sum = 0.0

    for rank, video_id in enumerate(
        ranking[:k],
        start=1
    ):

        if y_true.get(video_id, 0) == 1:

            hits += 1

            precision_sum += (
                hits / rank
            )

    return (
        precision_sum /
        min(total_relevant, k)
    )


def reciprocal_rank(
    y_true,
    ranking
):

    for rank, video_id in enumerate(
        ranking,
        start=1
    ):

        if y_true.get(video_id, 0) == 1:
            return 1.0 / rank

    return 0.0


# ============================================================
# LOAD E5
# ============================================================

print("\nLoading E5 model...")

e5_model = SentenceTransformer(
    "intfloat/e5-base-v2"
)

print("E5 model loaded.")


# ============================================================
# TOPIC-BY-TOPIC EVALUATION
# ============================================================

results = []

for topic_id, group in tqdm(
    benchmark.groupby("topic_id"),
    desc="Evaluating LAK topics"
):

    group = group.copy()

    topic = str(
        group["topic"].iloc[0]
    )

    domain = str(
        group["domain"].iloc[0]
    )

    video_ids = (
        group["video_id"]
        .astype(str)
        .tolist()
    )

    documents = (
        group["content_clean"]
        .tolist()
    )

    relevance = (
        group["relevance"]
        .astype(int)
        .tolist()
    )

    y_true = dict(
        zip(
            video_ids,
            relevance
        )
    )


    # ========================================================
    # BM25
    # ========================================================

    tokenized_documents = [
        document.lower().split()
        for document in documents
    ]

    bm25 = BM25Okapi(
        tokenized_documents
    )

    query_tokens = (
        topic.lower().split()
    )

    bm25_scores = bm25.get_scores(
        query_tokens
    )

    bm25_order = np.argsort(
        bm25_scores
    )[::-1]

    bm25_ranking = [
        video_ids[i]
        for i in bm25_order
    ]


    # ========================================================
    # E5
    # ========================================================

    query_embedding = e5_model.encode(
        ["query: " + topic],
        normalize_embeddings=True,
        show_progress_bar=False
    )[0]

    document_embeddings = (
        e5_model.encode(
            [
                "passage: " + document
                for document in documents
            ],
            normalize_embeddings=True,
            show_progress_bar=False
        )
    )

    e5_scores = np.dot(
        document_embeddings,
        query_embedding
    )

    e5_order = np.argsort(
        e5_scores
    )[::-1]

    e5_ranking = [
        video_ids[i]
        for i in e5_order
    ]


    # ========================================================
    # STORE METRICS
    # ========================================================

    for method, ranking in [
        ("BM25", bm25_ranking),
        ("E5", e5_ranking)
    ]:

        results.append({

            "topic_id": topic_id,
            "domain": domain,
            "topic": topic,
            "method": method,

            "n_candidates": len(group),

            "n_relevant": sum(relevance),

            "NDCG@5":
                ndcg_at_k(
                    y_true,
                    ranking,
                    5
                ),

            "NDCG@10":
                ndcg_at_k(
                    y_true,
                    ranking,
                    10
                ),

            "MAP@10":
                average_precision_at_k(
                    y_true,
                    ranking,
                    10
                ),

            "MRR":
                reciprocal_rank(
                    y_true,
                    ranking
                ),

            "Precision@5":
                precision_at_k(
                    y_true,
                    ranking,
                    5
                ),

            "Precision@10":
                precision_at_k(
                    y_true,
                    ranking,
                    10
                ),

            "Recall@5":
                recall_at_k(
                    y_true,
                    ranking,
                    5
                ),

            "Recall@10":
                recall_at_k(
                    y_true,
                    ranking,
                    10
                )
        })


# ============================================================
# SAVE TOPIC-LEVEL RESULTS
# ============================================================

results_df = pd.DataFrame(
    results
)

topic_results_path = (
    BASE /
    "phase10b_baseline_topic_results.csv"
)

results_df.to_csv(
    topic_results_path,
    index=False
)


# ============================================================
# MACRO-AVERAGED SUMMARY
# ============================================================

metric_columns = [
    "NDCG@5",
    "NDCG@10",
    "MAP@10",
    "MRR",
    "Precision@5",
    "Precision@10",
    "Recall@5",
    "Recall@10"
]

summary = (
    results_df
    .groupby("method")[metric_columns]
    .mean()
    .reset_index()
)

summary_path = (
    BASE /
    "phase10b_baseline_summary.csv"
)

summary.to_csv(
    summary_path,
    index=False
)


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "=" * 90)
print("PHASE 10B COMPLETE")
print("=" * 90)

print(
    "\nTopic-level results:"
)
print(topic_results_path)

print(
    "\nSummary:"
)
print(summary_path)

print(
    "\nMACRO-AVERAGED RESULTS:"
)

display(
    summary.round(4)
)
````

# PHASE 10C — EXTERNAL BENCHMARK RANKING METRICS

The metrics are computed **per benchmark topic** and then macro-averaged,
which avoids allowing topics with more resources to dominate the result.

For this external benchmark:

- Relevance = expert accepted/rejected label.
- NDCG@K uses the binary relevance labels.
- Precision@K and Recall@K use the same independent labels.
- MRR@K and MAP@K are calculated per topic.
- Results are reported with the number of evaluated topics.

This is the part of the revision that supplies an independent reference
for ranking evaluation; it must not be confused with the 99-video
five-domain case study. citeturn1search0turn1search2


````python
# ============================================================================
# PHASE 10C — PROPOSED METHOD COMPONENT ON LAK 2026 EXTERNAL BENCHMARK
# ============================================================================
# Purpose:
# Evaluate the content-based component of the proposed ranking framework
# against the same human-validated LAK 2026 benchmark used in Phase 10B.
#
# IMPORTANT:
# - No fabricated Teaching Scores
# - No fabricated engagement data
# - No FES reconstruction
# - Uses only information actually available in the LAK benchmark
#
# Components evaluated:
#   1. E5 Semantic Alignment
#   2. Content Depth proxy
#   3. Lexical Diversity (TTR)
#   4. Content-based Intrinsic Quality Score
#
# Metrics:
#   NDCG@5, NDCG@10, MAP@10, MRR,
#   Precision@5, Precision@10, Recall@5, Recall@10
# ============================================================================

import os
import re
import numpy as np
import pandas as pd
from tqdm.auto import tqdm
from sentence_transformers import SentenceTransformer
from sklearn.metrics import ndcg_score
from scipy.stats import spearmanr

print("=" * 90)
print("PHASE 10C — PROPOSED CONTENT-BASED METHOD")
print("=" * 90)

# ----------------------------------------------------------------------------
# 1. LOAD CLEAN LAK BENCHMARK
# ----------------------------------------------------------------------------

BENCHMARK_PATH = "/content/revision_phase10/lak2026_benchmark_eval.csv"

df = pd.read_csv(BENCHMARK_PATH)

required_cols = [
    "topic_id",
    "domain",
    "topic",
    "video_id",
    "content",
    "relevance"
]

missing = [c for c in required_cols if c not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

print(f"Loaded pairs : {len(df)}")
print(f"Topics       : {df['topic_id'].nunique()}")
print(f"Videos       : {df['video_id'].nunique()}")

# ----------------------------------------------------------------------------
# 2. CLEAN CONTENT
# ----------------------------------------------------------------------------

df["content_clean"] = (
    df["content"]
    .fillna("")
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

before = len(df)

df = df[df["content_clean"].str.len() > 0].copy()

print("\nContent filtering:")
print(f"Before: {before}")
print(f"After : {len(df)}")
print(f"Removed: {before - len(df)}")

# ----------------------------------------------------------------------------
# 3. VALIDATE TOPICS
# ----------------------------------------------------------------------------

topic_stats = (
    df.groupby("topic_id")
      .agg(
          topic=("topic", "first"),
          n_pairs=("video_id", "count"),
          positives=("relevance", "sum"),
          negatives=("relevance", lambda x: (x == 0).sum())
      )
      .reset_index()
)

valid_topics = topic_stats[
    (topic_stats["positives"] > 0) &
    (topic_stats["negatives"] > 0)
]["topic_id"].tolist()

df = df[df["topic_id"].isin(valid_topics)].copy()

print(f"\nFinal evaluation topics: {len(valid_topics)}")
print(f"Final evaluation pairs : {len(df)}")
print(f"Final unique videos     : {df['video_id'].nunique()}")

# ----------------------------------------------------------------------------
# 4. LOAD E5
# ----------------------------------------------------------------------------

print("\nLoading E5 model...")

model = SentenceTransformer("intfloat/e5-base-v2")

print("E5 model loaded.")

# ----------------------------------------------------------------------------
# 5. HELPER FUNCTIONS
# ----------------------------------------------------------------------------

def clean_words(text):
    return re.findall(r"\b[a-zA-Z0-9]+\b", str(text).lower())


def lexical_diversity(text):
    """
    Type-Token Ratio (TTR):
        unique tokens / total tokens
    """
    words = clean_words(text)

    if not words:
        return 0.0

    return len(set(words)) / len(words)


def minmax(values):
    values = np.asarray(values, dtype=float)

    vmin = values.min()
    vmax = values.max()

    if vmax == vmin:
        return np.zeros_like(values)

    return (values - vmin) / (vmax - vmin)


def average_precision_at_k(y_true, k=10):
    y_true = np.asarray(y_true)[:k]

    total_relevant = np.sum(y_true)

    if total_relevant == 0:
        return 0.0

    score = 0.0
    hits = 0

    for i, rel in enumerate(y_true, start=1):

        if rel == 1:
            hits += 1
            score += hits / i

    return score / min(total_relevant, k)


def reciprocal_rank(y_true):
    for i, rel in enumerate(y_true, start=1):
        if rel == 1:
            return 1.0 / i

    return 0.0


def precision_at_k(y_true, k):
    y_true = np.asarray(y_true)[:k]

    if len(y_true) == 0:
        return 0.0

    return np.mean(y_true)


def recall_at_k(y_true, k):
    y_true = np.asarray(y_true)

    total_relevant = np.sum(y_true)

    if total_relevant == 0:
        return 0.0

    return np.sum(y_true[:k]) / total_relevant


def ndcg_at_k(y_true, k):
    y_true = np.asarray(y_true)

    if np.sum(y_true) == 0:
        return 0.0

    return ndcg_score(
        [y_true],
        [np.arange(len(y_true), 0, -1)],
        k=k
    )


# ----------------------------------------------------------------------------
# 6. EVALUATE TOPIC BY TOPIC
# ----------------------------------------------------------------------------

results = []
ranking_records = []

for topic_id, group in tqdm(
    df.groupby("topic_id"),
    desc="Evaluating LAK topics"
):

    group = group.copy()

    topic = group["topic"].iloc[0]
    domain = group["domain"].iloc[0]

    contents = group["content_clean"].tolist()

    # ------------------------------------------------------------------------
    # E5 semantic alignment
    # ------------------------------------------------------------------------

    query_embedding = model.encode(
        ["query: " + topic],
        normalize_embeddings=True,
        show_progress_bar=False
    )[0]

    document_embeddings = model.encode(
        ["passage: " + x for x in contents],
        normalize_embeddings=True,
        show_progress_bar=False,
        batch_size=32
    )

    semantic_scores = np.dot(
        document_embeddings,
        query_embedding
    )

    group["semantic_alignment"] = semantic_scores

    # ------------------------------------------------------------------------
    # Content depth
    #
    # We deliberately call this a STRUCTURAL PROXY.
    # Longer text does NOT imply higher educational quality.
    # ------------------------------------------------------------------------

    group["content_length"] = group["content_clean"].str.len()

    group["depth_proxy"] = minmax(
        group["content_length"].values
    )

    # ------------------------------------------------------------------------
    # Lexical diversity
    # ------------------------------------------------------------------------

    group["lexical_diversity"] = group[
        "content_clean"
    ].apply(lexical_diversity)

    # Normalize semantic alignment within topic
    group["semantic_alignment_norm"] = minmax(
        group["semantic_alignment"].values
    )

    # Normalize lexical diversity within topic
    group["lexical_diversity_norm"] = minmax(
        group["lexical_diversity"].values
    )

    # ------------------------------------------------------------------------
    # CONTENT-BASED INTRINSIC QUALITY SCORE
    #
    # Original manuscript structure:
    #
    # 0.30 Semantic Alignment
    # 0.20 Depth
    # 0.20 Lexical Diversity
    # 0.30 Teaching Score
    #
    # Teaching Score is unavailable in LAK benchmark.
    #
    # Therefore we DO NOT invent it.
    #
    # Instead, the available components are renormalized:
    #
    # Semantic = 0.30 / 0.70
    # Depth    = 0.20 / 0.70
    # Lexical  = 0.20 / 0.70
    # ------------------------------------------------------------------------

    group["intrinsic_quality_score"] = (
        (0.30 / 0.70) * group["semantic_alignment_norm"] +
        (0.20 / 0.70) * group["depth_proxy"] +
        (0.20 / 0.70) * group["lexical_diversity_norm"]
    )

    # ------------------------------------------------------------------------
    # Ranking
    # ------------------------------------------------------------------------

    group = group.sort_values(
        "intrinsic_quality_score",
        ascending=False
    ).reset_index(drop=True)

    y_true = group["relevance"].astype(int).values

    # ------------------------------------------------------------------------
    # Metrics
    # ------------------------------------------------------------------------

    ndcg5 = ndcg_at_k(y_true, 5)
    ndcg10 = ndcg_at_k(y_true, 10)

    map10 = average_precision_at_k(y_true, 10)

    mrr = reciprocal_rank(y_true)

    p5 = precision_at_k(y_true, 5)
    p10 = precision_at_k(y_true, 10)

    r5 = recall_at_k(y_true, 5)
    r10 = recall_at_k(y_true, 10)

    results.append({
        "topic_id": topic_id,
        "domain": domain,
        "topic": topic,
        "n_pairs": len(group),
        "n_relevant": int(np.sum(y_true)),
        "NDCG@5": ndcg5,
        "NDCG@10": ndcg10,
        "MAP@10": map10,
        "MRR": mrr,
        "Precision@5": p5,
        "Precision@10": p10,
        "Recall@5": r5,
        "Recall@10": r10
    })

    # Save ranking information for later error analysis
    for rank, (_, row) in enumerate(group.iterrows(), start=1):

        ranking_records.append({
            "topic_id": topic_id,
            "domain": domain,
            "topic": topic,
            "rank": rank,
            "video_id": row["video_id"],
            "relevance": int(row["relevance"]),
            "semantic_alignment": row["semantic_alignment"],
            "depth_proxy": row["depth_proxy"],
            "lexical_diversity": row["lexical_diversity"],
            "intrinsic_quality_score": row["intrinsic_quality_score"]
        })


# ----------------------------------------------------------------------------
# 7. RESULTS
# ----------------------------------------------------------------------------

results_df = pd.DataFrame(results)

ranking_df = pd.DataFrame(ranking_records)

# ----------------------------------------------------------------------------
# 8. SAVE OUTPUTS
# ----------------------------------------------------------------------------

OUTPUT_DIR = "/content/revision_phase10"

os.makedirs(OUTPUT_DIR, exist_ok=True)

topic_output = os.path.join(
    OUTPUT_DIR,
    "phase10c_proposed_content_topic_results.csv"
)

ranking_output = os.path.join(
    OUTPUT_DIR,
    "phase10c_proposed_content_rankings.csv"
)

summary_output = os.path.join(
    OUTPUT_DIR,
    "phase10c_proposed_content_summary.csv"
)

results_df.to_csv(topic_output, index=False)

ranking_df.to_csv(ranking_output, index=False)

macro_summary = pd.DataFrame([{
    "method": "Proposed_Content_IQS",
    "NDCG@5": results_df["NDCG@5"].mean(),
    "NDCG@10": results_df["NDCG@10"].mean(),
    "MAP@10": results_df["MAP@10"].mean(),
    "MRR": results_df["MRR"].mean(),
    "Precision@5": results_df["Precision@5"].mean(),
    "Precision@10": results_df["Precision@10"].mean(),
    "Recall@5": results_df["Recall@5"].mean(),
    "Recall@10": results_df["Recall@10"].mean()
}])

macro_summary.to_csv(summary_output, index=False)

# ----------------------------------------------------------------------------
# 9. DISPLAY
# ----------------------------------------------------------------------------

print("\n" + "=" * 90)
print("PHASE 10C COMPLETE")
print("=" * 90)

print(f"\nTopic-level results:")
print(topic_output)

print(f"\nRanking results:")
print(ranking_output)

print(f"\nSummary:")
print(summary_output)

print("\nMACRO-AVERAGED RESULTS:")

display(
    macro_summary.round(4)
)

print("\nComponent statistics:")

display(
    results_df[
        [
            "NDCG@5",
            "NDCG@10",
            "MAP@10",
            "MRR",
            "Precision@5",
            "Precision@10",
            "Recall@5",
            "Recall@10"
        ]
    ].describe().round(4)
)

print("\n" + "=" * 90)
print("IMPORTANT:")
print("This benchmark evaluation excludes FES and LLM Teaching Score")
print("because those signals are not available in the LAK benchmark.")
print("No ground-truth labels or engagement values were fabricated.")
print("=" * 90)
````

````python
# ============================================================================
# PHASE 10D — STATISTICAL SIGNIFICANCE ANALYSIS
# ============================================================================
#
# Compares:
#   BM25 vs E5
#   BM25 vs Proposed Content-IQS
#   E5   vs Proposed Content-IQS
#
# Uses the same 54 LAK topics for every comparison.
#
# Statistical test:
#   Paired Wilcoxon signed-rank test
#
# Multiple-comparison correction:
#   Benjamini-Hochberg FDR
#
# Metrics:
#   NDCG@5
#   NDCG@10
#   MAP@10
#   MRR
#   Precision@5
#   Precision@10
#   Recall@5
#   Recall@10
# ============================================================================

import os
import numpy as np
import pandas as pd

from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests


print("=" * 90)
print("PHASE 10D — STATISTICAL SIGNIFICANCE ANALYSIS")
print("=" * 90)


# ============================================================================
# 1. PATHS
# ============================================================================

PHASE10_DIR = "/content/revision_phase10"

BASELINE_PATH = os.path.join(
    PHASE10_DIR,
    "phase10b_baseline_topic_results.csv"
)

PROPOSED_PATH = os.path.join(
    PHASE10_DIR,
    "phase10c_proposed_content_topic_results.csv"
)


# ============================================================================
# 2. LOAD RESULTS
# ============================================================================

baseline_df = pd.read_csv(BASELINE_PATH)
proposed_df = pd.read_csv(PROPOSED_PATH)

print(f"\nPhase 10B rows: {len(baseline_df)}")
print(f"Phase 10C rows: {len(proposed_df)}")


# ============================================================================
# 3. REQUIRED COLUMNS
# ============================================================================

metrics = [
    "NDCG@5",
    "NDCG@10",
    "MAP@10",
    "MRR",
    "Precision@5",
    "Precision@10",
    "Recall@5",
    "Recall@10"
]

required_baseline = ["topic_id", "method"] + metrics
required_proposed = ["topic_id"] + metrics

for col in required_baseline:
    if col not in baseline_df.columns:
        raise ValueError(
            f"Missing column in Phase 10B: {col}"
        )

for col in required_proposed:
    if col not in proposed_df.columns:
        raise ValueError(
            f"Missing column in Phase 10C: {col}"
        )


# ============================================================================
# 4. SPLIT BM25 AND E5
# ============================================================================

baseline_df["method"] = (
    baseline_df["method"]
    .astype(str)
    .str.strip()
)

bm25_df = baseline_df[
    baseline_df["method"].str.upper() == "BM25"
].copy()

e5_df = baseline_df[
    baseline_df["method"].str.upper() == "E5"
].copy()


print("\nPhase 10B method counts:")
print(baseline_df["method"].value_counts())


if len(bm25_df) != 54:
    raise ValueError(
        f"Expected 54 BM25 topic rows, found {len(bm25_df)}"
    )

if len(e5_df) != 54:
    raise ValueError(
        f"Expected 54 E5 topic rows, found {len(e5_df)}"
    )

if len(proposed_df) != 54:
    raise ValueError(
        f"Expected 54 Proposed topic rows, found {len(proposed_df)}"
    )


# ============================================================================
# 5. CHECK DUPLICATES
# ============================================================================

for name, frame in [
    ("BM25", bm25_df),
    ("E5", e5_df),
    ("Proposed_Content_IQS", proposed_df)
]:

    if frame["topic_id"].duplicated().any():

        duplicate_ids = (
            frame.loc[
                frame["topic_id"].duplicated(keep=False),
                "topic_id"
            ]
            .unique()
            .tolist()
        )

        raise ValueError(
            f"{name} contains duplicate topic IDs: "
            f"{duplicate_ids}"
        )


# ============================================================================
# 6. CREATE ONE ROW PER TOPIC
# ============================================================================

bm25 = bm25_df[
    ["topic_id"] + metrics
].copy()

e5 = e5_df[
    ["topic_id"] + metrics
].copy()

proposed = proposed_df[
    ["topic_id"] + metrics
].copy()


bm25 = bm25.rename(
    columns={
        m: f"BM25_{m}"
        for m in metrics
    }
)

e5 = e5.rename(
    columns={
        m: f"E5_{m}"
        for m in metrics
    }
)

proposed = proposed.rename(
    columns={
        m: f"Proposed_{m}"
        for m in metrics
    }
)


# ============================================================================
# 7. MATCH THE SAME TOPICS
# ============================================================================

topic_scores = (
    bm25
    .merge(
        e5,
        on="topic_id",
        how="inner"
    )
    .merge(
        proposed,
        on="topic_id",
        how="inner"
    )
)


print(
    f"\nMatched topics available for statistical testing: "
    f"{len(topic_scores)}"
)


if len(topic_scores) != 54:

    raise ValueError(
        f"Expected 54 matched topics, "
        f"but found {len(topic_scores)}."
    )


print("✓ All 54 topics matched across all three methods.")


# ============================================================================
# 8. WILCOXON TEST FUNCTION
# ============================================================================

def run_wilcoxon(x, y):

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    valid = (
        np.isfinite(x) &
        np.isfinite(y)
    )

    x = x[valid]
    y = y[valid]

    differences = x - y

    # If all paired differences are zero
    if np.all(differences == 0):

        return 0.0, 1.0, len(x)

    try:

        statistic, p_value = wilcoxon(
            x,
            y,
            alternative="two-sided",
            zero_method="wilcox",
            correction=False,
            mode="auto"
        )

        return (
            float(statistic),
            float(p_value),
            len(x)
        )

    except ValueError:

        return (
            np.nan,
            np.nan,
            len(x)
        )


# ============================================================================
# 9. PAIRWISE COMPARISONS
# ============================================================================

comparisons = [
    ("BM25", "E5"),
    ("BM25", "Proposed"),
    ("E5", "Proposed")
]


test_rows = []


for method_a, method_b in comparisons:

    for metric in metrics:

        col_a = f"{method_a}_{metric}"

        col_b = f"{method_b}_{metric}"

        x = topic_scores[col_a].values
        y = topic_scores[col_b].values

        statistic, p_value, n = run_wilcoxon(
            x,
            y
        )

        mean_a = np.mean(x)
        mean_b = np.mean(y)

        median_a = np.median(x)
        median_b = np.median(y)

        mean_difference = mean_a - mean_b

        if mean_difference > 0:
            direction = "A > B"

        elif mean_difference < 0:
            direction = "A < B"

        else:
            direction = "Equal"

        test_rows.append({

            "comparison":
                f"{method_a} vs {method_b}",

            "method_a":
                method_a,

            "method_b":
                method_b,

            "metric":
                metric,

            "n_topics":
                n,

            "mean_a":
                mean_a,

            "mean_b":
                mean_b,

            "mean_difference":
                mean_difference,

            "median_a":
                median_a,

            "median_b":
                median_b,

            "wilcoxon_statistic":
                statistic,

            "p_value":
                p_value,

            "direction":
                direction
        })


significance_df = pd.DataFrame(test_rows)


# ============================================================================
# 10. BENJAMINI-HOCHBERG FDR CORRECTION
# ============================================================================
#
# 24 statistical tests:
#   3 method comparisons × 8 metrics
#
# We correct all 24 p-values together.
# ============================================================================

valid = significance_df["p_value"].notna()

significance_df["p_value_fdr"] = np.nan

significance_df["significant_fdr_0.05"] = False


if valid.sum() > 0:

    corrected_p = multipletests(
        significance_df.loc[
            valid,
            "p_value"
        ],
        alpha=0.05,
        method="fdr_bh"
    )

    significance_df.loc[
        valid,
        "p_value_fdr"
    ] = corrected_p[1]

    significance_df.loc[
        valid,
        "significant_fdr_0.05"
    ] = corrected_p[0]


# ============================================================================
# 11. EFFECT SIZE — RANK-BISERIAL CORRELATION
# ============================================================================
#
# For paired Wilcoxon:
#
# r_rb = (W_positive - W_negative) /
#        (W_positive + W_negative)
#
# This gives an interpretable effect direction.
# ============================================================================

def rank_biserial_effect(x, y):

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    differences = x - y

    differences = differences[
        differences != 0
    ]

    if len(differences) == 0:
        return 0.0

    abs_diff = np.abs(differences)

    ranks = pd.Series(abs_diff).rank(
        method="average"
    ).to_numpy()

    positive_sum = np.sum(
        ranks[differences > 0]
    )

    negative_sum = np.sum(
        ranks[differences < 0]
    )

    denominator = (
        positive_sum +
        negative_sum
    )

    if denominator == 0:
        return 0.0

    return (
        (positive_sum - negative_sum) /
        denominator
    )


effect_sizes = []


for _, row in significance_df.iterrows():

    method_a = row["method_a"]
    method_b = row["method_b"]
    metric = row["metric"]

    x = topic_scores[
        f"{method_a}_{metric}"
    ].values

    y = topic_scores[
        f"{method_b}_{metric}"
    ].values

    effect = rank_biserial_effect(
        x,
        y
    )

    effect_sizes.append(effect)


significance_df[
    "rank_biserial_effect"
] = effect_sizes


# ============================================================================
# 12. SAVE RESULTS
# ============================================================================

SIGNIFICANCE_PATH = os.path.join(
    PHASE10_DIR,
    "phase10d_statistical_significance.csv"
)

TOPIC_SCORES_PATH = os.path.join(
    PHASE10_DIR,
    "phase10d_topic_level_scores.csv"
)


significance_df.to_csv(
    SIGNIFICANCE_PATH,
    index=False
)

topic_scores.to_csv(
    TOPIC_SCORES_PATH,
    index=False
)


# ============================================================================
# 13. DISPLAY
# ============================================================================

print("\n" + "=" * 90)
print("PHASE 10D COMPLETE")
print("=" * 90)


print("\nSTATISTICAL SIGNIFICANCE RESULTS:")

display(
    significance_df[
        [
            "comparison",
            "metric",
            "n_topics",
            "mean_a",
            "mean_b",
            "mean_difference",
            "wilcoxon_statistic",
            "p_value",
            "p_value_fdr",
            "significant_fdr_0.05",
            "rank_biserial_effect",
            "direction"
        ]
    ].round(6)
)


# ============================================================================
# 14. ONLY SIGNIFICANT RESULTS
# ============================================================================

significant_results = significance_df[
    significance_df[
        "significant_fdr_0.05"
    ]
].copy()


print("\n" + "=" * 90)
print("FDR-SIGNIFICANT RESULTS (q < 0.05)")
print("=" * 90)


if len(significant_results) == 0:

    print(
        "No comparisons remain statistically significant "
        "after Benjamini-Hochberg FDR correction."
    )

else:

    display(
        significant_results[
            [
                "comparison",
                "metric",
                "mean_a",
                "mean_b",
                "mean_difference",
                "p_value",
                "p_value_fdr",
                "rank_biserial_effect",
                "direction"
            ]
        ].round(6)
    )


# ============================================================================
# 15. SAVE A COMPACT PAPER-READY TABLE
# ============================================================================

paper_table = significance_df[
    [
        "comparison",
        "metric",
        "mean_a",
        "mean_b",
        "mean_difference",
        "p_value",
        "p_value_fdr",
        "rank_biserial_effect",
        "significant_fdr_0.05"
    ]
].copy()


PAPER_TABLE_PATH = os.path.join(
    PHASE10_DIR,
    "phase10d_paper_significance_table.csv"
)


paper_table.to_csv(
    PAPER_TABLE_PATH,
    index=False
)


# ============================================================================
# 16. FINAL FILES
# ============================================================================

print("\n" + "=" * 90)
print("FILES SAVED")
print("=" * 90)

print(
    "\nFull statistical results:"
)

print(SIGNIFICANCE_PATH)

print(
    "\nMatched topic-level scores:"
)

print(TOPIC_SCORES_PATH)

print(
    "\nPaper-ready significance table:"
)

print(PAPER_TABLE_PATH)


print("\n" + "=" * 90)
print("ANALYSIS DETAILS")
print("=" * 90)

print("""
Evaluation unit:
    54 matched educational topics

Statistical test:
    Paired Wilcoxon signed-rank test

Number of pairwise method comparisons:
    3

Metrics per comparison:
    8

Total statistical tests:
    24

Multiple-comparison correction:
    Benjamini-Hochberg FDR

Effect size:
    Rank-biserial correlation

The analysis compares topic-level performance across the same
54 educational topics for every method.
""")
````

# PHASE 10E — EVALUATION INTERPRETATION

Do **not** claim that the external benchmark proves the full Hybrid model is
superior unless the exact Hybrid pipeline is executed on the benchmark
resources.

The defensible interpretation is:

1. The external benchmark provides independent expert reference labels.
2. Our E5 semantic retrieval component can be evaluated against those labels.
3. The five-domain dataset evaluates the complete FES/IQS/adaptive
   Hybrid framework through multi-domain analysis, ablation, sensitivity,
   ranking movement, and statistical testing.
4. These are complementary evaluations, not one substituted for the other.

The LAK 2026 benchmark itself used two domain experts per topic and reported
standard top-k ranking measures, making it a relevant methodological precedent
for this study. citeturn1search0turn1search2


# PHASE 10F — EXTERNAL BASELINE CONTEXT

For manuscript comparison, cite the benchmark as a **related external
evaluation**, not as our own result.

The LAK 2026 study reports a YouTube baseline and several embedding models
on 53 topics, with its best reported embedding approach achieving 0.79 ± 0.14
ranking accuracy and the YouTube baseline 0.59 ± 0.18. It also reports
Precision@3 and Precision@5 and uses a Friedman test across the evaluated
methods. citeturn2view0

Our notebook will report **our own scores** on the downloaded reference data;
we should not copy the paper's scores into our experimental tables.


````python
# ============================================================================
# PHASE 10E — EXTERNAL BENCHMARK INTERPRETATION
# ============================================================================
#
# IMPORTANT:
# The LAK 2026 benchmark provides independent expert reference labels.
#
# We evaluated:
#   - BM25
#   - E5 semantic retrieval
#   - Proposed Content-IQS component
#
# We did NOT execute the complete FES + LLM Teaching Score +
# Adaptive Hybrid pipeline on the LAK benchmark.
#
# Therefore:
#   - LAK results are NOT presented as proof that the complete Hybrid
#     framework is superior.
#   - LAK is an independent external validation of the content-based
#     ranking component.
#   - The complete Hybrid framework is evaluated separately on the
#     five-domain experimental cohort.
# ============================================================================

print("=" * 100)
print("PHASE 10E — EXTERNAL BENCHMARK INTERPRETATION")
print("=" * 100)

print("""
EXTERNAL BENCHMARK INTERPRETATION
----------------------------------

1. The LAK 2026 benchmark provides independently generated expert
   reference labels for educational-resource relevance.

2. Our BM25 and E5 retrieval baselines were evaluated directly against
   these reference labels.

3. The proposed Content-IQS component was also evaluated on the same
   54-topic LAK benchmark.

4. The complete FES + LLM Teaching Score + Adaptive Hybrid framework
   was NOT treated as evaluated on the LAK benchmark because the
   benchmark does not provide the required engagement and LLM-derived
   signals.

5. Therefore, the external benchmark and the five-domain Hybrid
   experiment are complementary evaluations rather than substitutes.

6. No model-generated score, engagement score, or IQS value is treated
   as ground truth.
""")


# ============================================================================
# PHASE 10F — EXTERNAL BENCHMARK ARTIFACT VALIDATION
# ============================================================================

print("\n" + "=" * 100)
print("PHASE 10F — EXTERNAL BENCHMARK ARTIFACT VALIDATION")
print("=" * 100)


# ---------------------------------------------------------------------------
# Core experimental artifacts
# ---------------------------------------------------------------------------

artifact_checks = [

    # Phase 9 — complete Hybrid experiment
    (
        "Phase 9F — Hybrid cohort",
        "/content/revision_phase9f/phase9f_hybrid_full.csv"
    ),

    # Phase 9 — true ablation
    (
        "Phase 9H — true ablation",
        "/content/revision_phase9h/true_ablation_results.csv"
    ),

    # Phase 9 — sensitivity analysis
    (
        "Phase 9I — weight sensitivity",
        "/content/revision_phase9h/weight_sensitivity.csv"
    ),

    # Phase 10 — external benchmark
    (
        "Phase 10A — LAK benchmark",
        "/content/revision_phase10/lak2026_benchmark_eval.csv"
    ),

    # Phase 10B
    (
        "Phase 10B — BM25 + E5 baseline results",
        "/content/revision_phase10/phase10b_baseline_summary.csv"
    ),

    (
        "Phase 10B — topic-level baseline results",
        "/content/revision_phase10/phase10b_baseline_topic_results.csv"
    ),

    # Phase 10C
    (
        "Phase 10C — Proposed Content-IQS summary",
        "/content/revision_phase10/phase10c_proposed_content_summary.csv"
    ),

    (
        "Phase 10C — topic-level results",
        "/content/revision_phase10/phase10c_proposed_content_topic_results.csv"
    ),

    (
        "Phase 10C — ranking results",
        "/content/revision_phase10/phase10c_proposed_content_rankings.csv"
    ),

    # Phase 10D
    (
        "Phase 10D — statistical significance",
        "/content/revision_phase10/phase10d_statistical_significance.csv"
    ),

    (
        "Phase 10D — topic-level scores",
        "/content/revision_phase10/phase10d_topic_level_scores.csv"
    ),

    (
        "Phase 10D — paper significance table",
        "/content/revision_phase10/phase10d_paper_significance_table.csv"
    ),

    # Phase 11 — popularity + LambdaMART
    (
        "Phase 11 — popularity topic results",
        "/content/revision_phase11/phase11_popularity_topic_results.csv"
    ),

    (
        "Phase 11 — LambdaMART topic results",
        "/content/revision_phase11/phase11_lambdamart_topic_results.csv"
    ),

    (
        "Phase 11 — LambdaMART predictions",
        "/content/revision_phase11/phase11_lambdamart_predictions.csv"
    ),

    (
        "Phase 11 — LambdaMART feature importance",
        "/content/revision_phase11/phase11_lambdamart_feature_importance.csv"
    ),

    (
        "Phase 11 — final baseline summary",
        "/content/revision_phase11/phase11_summary.csv"
    )
]


print("\nArtifact validation:\n")


all_artifacts_ok = True


for name, path in artifact_checks:

    exists = os.path.exists(path)

    if exists:
        print(f"✅ {name}")
    else:
        print(f"❌ {name}")
        all_artifacts_ok = False


# ============================================================================
# VALIDATE LAK BENCHMARK NUMBERS
# ============================================================================

benchmark_path = (
    "/content/revision_phase10/"
    "lak2026_benchmark_eval.csv"
)


if os.path.exists(benchmark_path):

    benchmark_check = pd.read_csv(
        benchmark_path
    )

    print("\n" + "-" * 100)
    print("LAK BENCHMARK VALIDATION")
    print("-" * 100)

    print(
        f"Topics       : "
        f"{benchmark_check['topic_id'].nunique()}"
    )

    print(
        f"Pairs        : "
        f"{len(benchmark_check)}"
    )

    print(
        f"Unique videos: "
        f"{benchmark_check['video_id'].nunique()}"
    )

    # Expected finalized benchmark cohort
    expected_topics = 54
    expected_pairs = 914
    expected_videos = 866

    if (
        benchmark_check["topic_id"].nunique()
        == expected_topics
    ):
        print("✅ 54 evaluation topics")

    else:
        print(
            "⚠️ Unexpected topic count"
        )
        all_artifacts_ok = False


    if len(benchmark_check) == expected_pairs:
        print("✅ 914 labeled topic-resource pairs")

    else:
        print(
            "⚠️ Unexpected pair count"
        )
        all_artifacts_ok = False


    if (
        benchmark_check["video_id"].nunique()
        == expected_videos
    ):
        print("✅ 866 unique videos")

    else:
        print(
            "⚠️ Unexpected video count"
        )
        all_artifacts_ok = False


# ============================================================================
# VALIDATE STATISTICAL RESULTS
# ============================================================================

significance_path = (
    "/content/revision_phase10/"
    "phase10d_statistical_significance.csv"
)


if os.path.exists(significance_path):

    significance = pd.read_csv(
        significance_path
    )

    print("\n" + "-" * 100)
    print("STATISTICAL ANALYSIS VALIDATION")
    print("-" * 100)

    print(
        f"Statistical tests: "
        f"{len(significance)}"
    )

    print(
        f"Method comparisons: "
        f"{significance['comparison'].nunique()}"
    )

    print(
        f"Metrics: "
        f"{significance['metric'].nunique()}"
    )

    significant_count = (
        significance[
            "significant_fdr_0.05"
        ]
        .sum()
    )

    print(
        f"FDR-significant comparisons: "
        f"{significant_count}"
    )

    if len(significance) == 24:
        print(
            "✅ 24 statistical tests "
            "(3 comparisons × 8 metrics)"
        )

    else:
        print(
            "⚠️ Unexpected number of statistical tests"
        )
        all_artifacts_ok = False


# ============================================================================
# FINAL STATUS
# ============================================================================

print("\n" + "=" * 100)
print("FINAL REVISION PIPELINE STATUS")
print("=" * 100)


if all_artifacts_ok:

    print("""
✅ ALL REQUIRED REVISION ARTIFACTS ARE PRESENT.

Core five-domain Hybrid experiment:
    ✓ Hybrid ranking
    ✓ True ablation
    ✓ Weight sensitivity

External LAK 2026 benchmark:
    ✓ Clean benchmark cohort
    ✓ BM25 baseline
    ✓ E5 baseline
    ✓ Proposed Content-IQS evaluation
    ✓ Statistical significance testing

Learning-to-Rank baselines:
    ✓ Popularity (views, likes, comments)
    ✓ LambdaMART with topic-grouped cross-validation

Methodological safeguard:
    ✓ No fabricated human labels
    ✓ No engagement values treated as ground truth
    ✓ No LAK result presented as proof of full Hybrid superiority
    ✓ External benchmark and internal Hybrid evaluation remain separate
""")

else:

    print("""
⚠️ SOME REQUIRED ARTIFACTS ARE MISSING.

Review the ❌ items above before treating the revision
pipeline as complete.
""")


print("=" * 100)
````

# PHASE 11 — POPULARITY + LAMBDAMART LEARNING-TO-RANK BASELINES

This phase adds the remaining established ranking baselines requested by the reviewers.

- Popularity by views, likes, and comment count
- LambdaMART Learning-to-Rank
- Topic-grouped 5-fold evaluation to prevent topic leakage
- NDCG, MAP, MRR, Precision@K, Recall@K

The LambdaMART implementation uses LightGBM's `lambdarank` objective and topic-level
groups; LightGBM documents `lambdarank` as a ranking objective and supports NDCG
evaluation at specified cutoffs. citeturn0search0turn0search1

````python
# ============================================================================
# PHASE 11 — POPULARITY + LAMBDAMART LEARNING-TO-RANK BASELINES
# ============================================================================
#
# Reviewer requirements addressed:
#   • established ranking baselines
#   • popularity baseline
#   • Learning-to-Rank baseline
#   • standard ranking metrics
#   • leakage-safe evaluation
#
# Evaluation source:
#   LAK 2026 expert-labeled benchmark prepared in Phase 10A.
#
# Popularity:
#   YouTube Views, Likes, and Comment Count are retrieved from the YouTube
#   Data API and cached. Each is evaluated as a separate popularity ranking.
#
# LambdaMART:
#   5-fold GroupKFold by TOPIC. A topic never appears in both training and
#   test folds. The model uses:
#       BM25 score
#       E5 semantic similarity
#       content length
#       lexical diversity (TTR)
#       log(views)
#       log(likes)
#       log(comment count)
#
# The target is the independent LAK expert label:
#       1 = Accepted
#       0 = Rejected
#
# LambdaMART uses LightGBM's "lambdarank" objective, which is specifically
# designed for learning-to-rank and supports NDCG-oriented evaluation.
# ============================================================================

import os
import re
import time
from pathlib import Path

import numpy as np
import pandas as pd

from tqdm.auto import tqdm
from scipy.stats import spearmanr
from sklearn.model_selection import GroupKFold
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi
from lightgbm import LGBMRanker

from googleapiclient.discovery import build

# ---------------------------------------------------------------------------
# 1. Paths
# ---------------------------------------------------------------------------
BASE = Path("/content/revision_phase11")
BASE.mkdir(parents=True, exist_ok=True)

BENCHMARK_PATH = Path("/content/revision_phase10/lak2026_benchmark_eval.csv")
STATS_PATH = BASE / "youtube_popularity_stats.csv"

POPULARITY_TOPIC_PATH = BASE / "phase11_popularity_topic_results.csv"
LAMBDA_TOPIC_PATH = BASE / "phase11_lambdamart_topic_results.csv"
LAMBDA_PRED_PATH = BASE / "phase11_lambdamart_predictions.csv"
FEATURE_IMPORTANCE_PATH = BASE / "phase11_lambdamart_feature_importance.csv"
SUMMARY_PATH = BASE / "phase11_summary.csv"

if not BENCHMARK_PATH.exists():
    raise FileNotFoundError(
        f"Phase 10A benchmark is missing: {BENCHMARK_PATH}"
    )

benchmark = pd.read_csv(BENCHMARK_PATH)

required = [
    "topic_id", "domain", "topic",
    "video_id", "content", "relevance"
]
missing = [c for c in required if c not in benchmark.columns]

if missing:
    raise ValueError(f"Missing benchmark columns: {missing}")

benchmark["video_id"] = benchmark["video_id"].astype(str)
benchmark["content_clean"] = (
    benchmark["content"]
    .fillna("")
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

benchmark = benchmark[
    benchmark["content_clean"].str.len() > 0
].copy()

# ---------------------------------------------------------------------------
# 2. Ranking metric functions — self-contained so Phase 11 survives restart.
# ---------------------------------------------------------------------------
def dcg_binary(relevances, k):
    rel = np.asarray(relevances[:k], dtype=float)
    if len(rel) == 0:
        return 0.0
    discounts = np.log2(np.arange(2, len(rel) + 2))
    return float(np.sum((2 ** rel - 1) / discounts))


def ndcg_at_k(y_true, ranking, k):
    ranked = [y_true.get(v, 0) for v in ranking[:k]]
    ideal = sorted(y_true.values(), reverse=True)[:k]
    ideal_dcg = dcg_binary(ideal, k)
    if ideal_dcg == 0:
        return np.nan
    return dcg_binary(ranked, k) / ideal_dcg


def precision_at_k(y_true, ranking, k):
    top = ranking[:k]
    if not top:
        return np.nan
    return sum(y_true.get(v, 0) for v in top) / len(top)


def recall_at_k(y_true, ranking, k):
    total = sum(y_true.values())
    if total == 0:
        return np.nan
    top = ranking[:k]
    return sum(y_true.get(v, 0) for v in top) / total


def average_precision_at_k(y_true, ranking, k):
    total = sum(y_true.values())
    if total == 0:
        return np.nan

    hits = 0
    precision_sum = 0.0

    for rank, video_id in enumerate(ranking[:k], start=1):
        if y_true.get(video_id, 0) == 1:
            hits += 1
            precision_sum += hits / rank

    return precision_sum / min(total, k)


def reciprocal_rank(y_true, ranking):
    for rank, video_id in enumerate(ranking, start=1):
        if y_true.get(video_id, 0) == 1:
            return 1.0 / rank
    return 0.0


METRICS = [
    "NDCG@5", "NDCG@10", "MAP@10", "MRR",
    "Precision@5", "Precision@10",
    "Recall@5", "Recall@10"
]


def evaluate_ranking(group, ranking):
    video_ids = group["video_id"].astype(str).tolist()
    relevance = group["relevance"].astype(int).tolist()
    y_true = dict(zip(video_ids, relevance))

    return {
        "NDCG@5": ndcg_at_k(y_true, ranking, 5),
        "NDCG@10": ndcg_at_k(y_true, ranking, 10),
        "MAP@10": average_precision_at_k(y_true, ranking, 10),
        "MRR": reciprocal_rank(y_true, ranking),
        "Precision@5": precision_at_k(y_true, ranking, 5),
        "Precision@10": precision_at_k(y_true, ranking, 10),
        "Recall@5": recall_at_k(y_true, ranking, 5),
        "Recall@10": recall_at_k(y_true, ranking, 10),
    }


# ---------------------------------------------------------------------------
# 3. Retrieve/cache current YouTube popularity statistics.
# ---------------------------------------------------------------------------
unique_video_ids = sorted(
    benchmark["video_id"].dropna().astype(str).unique()
)

if STATS_PATH.exists():
    popularity = pd.read_csv(STATS_PATH)
    popularity["video_id"] = popularity["video_id"].astype(str)

    missing_ids = sorted(set(unique_video_ids) - set(popularity["video_id"]))
else:
    popularity = pd.DataFrame(
        columns=["video_id", "views", "likes", "comment_count"]
    )
    missing_ids = unique_video_ids

print("=" * 100)
print("PHASE 11 — POPULARITY + LAMBDAMART")
print("=" * 100)
print("Benchmark videos:", len(unique_video_ids))
print("Cached popularity rows:", len(popularity))
print("Videos requiring API retrieval:", len(missing_ids))

if missing_ids:
    if not YOUTUBE_API_KEY:
        raise RuntimeError(
            "YOUTUBE_API_KEY is required to retrieve missing popularity statistics. "
            "Add it to Colab Secrets, or restore phase11/youtube_popularity_stats.csv "
            "from the Drive backup."
        )

    youtube = build(
        "youtube",
        "v3",
        developerKey=YOUTUBE_API_KEY
    )

    new_rows = []

    for start in range(0, len(missing_ids), 50):
        batch = missing_ids[start:start + 50]

        response = youtube.videos().list(
            part="statistics",
            id=",".join(batch)
        ).execute()

        returned = {
            item["id"]: item.get("statistics", {})
            for item in response.get("items", [])
        }

        for video_id in batch:
            stats = returned.get(video_id, {})

            new_rows.append({
                "video_id": video_id,
                "views": int(stats.get("viewCount", 0)),
                "likes": int(stats.get("likeCount", 0)),
                "comment_count": int(stats.get("commentCount", 0)),
            })

        print(
            f"  Retrieved {min(start + 50, len(missing_ids))}/"
            f"{len(missing_ids)} videos"
        )

    popularity = pd.concat(
        [popularity, pd.DataFrame(new_rows)],
        ignore_index=True
    )

    popularity = (
        popularity
        .drop_duplicates("video_id", keep="last")
        .sort_values("video_id")
        .reset_index(drop=True)
    )

    popularity.to_csv(STATS_PATH, index=False)

print("Popularity cache:", STATS_PATH)

# Validate complete coverage.
missing_after = sorted(
    set(unique_video_ids) - set(popularity["video_id"])
)

if missing_after:
    raise RuntimeError(
        f"Popularity retrieval incomplete. Missing {len(missing_after)} videos."
    )

# ---------------------------------------------------------------------------
# 4. Merge popularity features.
# ---------------------------------------------------------------------------
for col in ["views", "likes", "comment_count"]:
    popularity[col] = pd.to_numeric(
        popularity[col], errors="coerce"
    ).fillna(0)

data = benchmark.merge(
    popularity[
        ["video_id", "views", "likes", "comment_count"]
    ],
    on="video_id",
    how="left",
    validate="many_to_one"
)

if data[["views", "likes", "comment_count"]].isna().any().any():
    raise RuntimeError("Popularity merge produced missing values.")

# ---------------------------------------------------------------------------
# 5. Popularity baselines.
# ---------------------------------------------------------------------------
popularity_rows = []

for topic_id, group in data.groupby("topic_id"):
    group = group.copy()

    for feature, method_name in [
        ("views", "Popularity_Views"),
        ("likes", "Popularity_Likes"),
        ("comment_count", "Popularity_Comments"),
    ]:
        ranking = (
            group.sort_values(
                [feature, "video_id"],
                ascending=[False, True]
            )["video_id"]
            .astype(str)
            .tolist()
        )

        metrics = evaluate_ranking(group, ranking)

        popularity_rows.append({
            "topic_id": topic_id,
            "domain": group["domain"].iloc[0],
            "topic": group["topic"].iloc[0],
            "method": method_name,
            **metrics
        })

popularity_topic = pd.DataFrame(popularity_rows)
popularity_topic.to_csv(
    POPULARITY_TOPIC_PATH,
    index=False
)

print("\nPOPULARITY BASELINES")
print(
    popularity_topic
    .groupby("method")[METRICS]
    .mean()
    .round(4)
    .to_string()
)

# ---------------------------------------------------------------------------
# 6. Build BM25 + E5 + content features for LambdaMART.
# ---------------------------------------------------------------------------
print("\nBuilding LambdaMART features...")

e5_model = SentenceTransformer("intfloat/e5-base-v2")

feature_frames = []

for topic_id, group in tqdm(
    data.groupby("topic_id"),
    desc="Feature extraction"
):
    group = group.copy().reset_index(drop=True)

    documents = group["content_clean"].tolist()
    topic = str(group["topic"].iloc[0])

    # BM25
    tokenized_documents = [
        doc.lower().split()
        for doc in documents
    ]

    bm25 = BM25Okapi(tokenized_documents)
    bm25_scores = bm25.get_scores(topic.lower().split())

    # E5
    query_embedding = e5_model.encode(
        ["query: " + topic],
        normalize_embeddings=True,
        show_progress_bar=False
    )[0]

    doc_embeddings = e5_model.encode(
        ["passage: " + doc for doc in documents],
        normalize_embeddings=True,
        show_progress_bar=False,
        batch_size=32
    )

    e5_scores = np.dot(
        doc_embeddings,
        query_embedding
    )

    # TTR
    def ttr(text):
        words = re.findall(
            r"\b[a-zA-Z0-9]+\b",
            str(text).lower()
        )
        return (
            len(set(words)) / len(words)
            if words else 0.0
        )

    group["bm25_score"] = bm25_scores
    group["e5_score"] = e5_scores
    group["content_length"] = group["content_clean"].str.len().astype(float)
    group["lexical_diversity"] = group["content_clean"].apply(ttr)

    # Log-transform popularity features to reduce scale dominance.
    group["log_views"] = np.log1p(group["views"])
    group["log_likes"] = np.log1p(group["likes"])
    group["log_comment_count"] = np.log1p(group["comment_count"])

    feature_frames.append(group)

feature_data = pd.concat(
    feature_frames,
    ignore_index=True
)

FEATURES = [
    "bm25_score",
    "e5_score",
    "content_length",
    "lexical_diversity",
    "log_views",
    "log_likes",
    "log_comment_count",
]

feature_data[FEATURES] = (
    feature_data[FEATURES]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0.0)
)

print("\nFeatures:", FEATURES)
print("Feature rows:", len(feature_data))

# ---------------------------------------------------------------------------
# 7. Leakage-safe 5-fold LambdaMART.
# ---------------------------------------------------------------------------
#
# GroupKFold is performed on topic IDs. Therefore no topic is present in both
# train and test in a fold.
# ---------------------------------------------------------------------------
topics = feature_data["topic_id"].unique()

if len(topics) < 5:
    raise RuntimeError(
        f"At least 5 topics are required for 5-fold GroupKFold; found {len(topics)}."
    )

gkf = GroupKFold(n_splits=5)

ltr_topic_rows = []
prediction_rows = []
importance_rows = []

for fold, (train_idx, test_idx) in enumerate(
    gkf.split(
        feature_data[FEATURES],
        feature_data["relevance"],
        groups=feature_data["topic_id"]
    ),
    start=1
):
    train = feature_data.iloc[train_idx].copy()
    test = feature_data.iloc[test_idx].copy()

    # Sort by topic because LightGBM's group sizes refer to contiguous rows.
    train = train.sort_values("topic_id").reset_index(drop=True)
    test = test.sort_values("topic_id").reset_index(drop=True)

    train_groups = (
        train.groupby("topic_id", sort=False)
        .size()
        .tolist()
    )

    test_groups = (
        test.groupby("topic_id", sort=False)
        .size()
        .tolist()
    )

    model = LGBMRanker(
        objective="lambdarank",
        metric="ndcg",
        label_gain=[0, 1],
        lambdarank_truncation_level=10,
        n_estimators=250,
        learning_rate=0.05,
        num_leaves=31,
        min_child_samples=10,
        subsample=0.9,
        subsample_freq=1,
        colsample_bytree=0.9,
        reg_lambda=1.0,
        random_state=42,
        verbosity=-1,
        n_jobs=-1,
    )

    model.fit(
        train[FEATURES],
        train["relevance"].astype(int),
        group=train_groups,
        eval_set=[(test[FEATURES], test["relevance"].astype(int))],
        eval_group=[test_groups],
        eval_at=[5, 10],
    )

    test["lambdamart_score"] = model.predict(
        test[FEATURES]
    )

    for topic_id, group in test.groupby("topic_id"):
        group = group.copy()

        ranking = (
            group.sort_values(
                ["lambdamart_score", "video_id"],
                ascending=[False, True]
            )["video_id"]
            .astype(str)
            .tolist()
        )

        metrics = evaluate_ranking(group, ranking)

        ltr_topic_rows.append({
            "fold": fold,
            "topic_id": topic_id,
            "domain": group["domain"].iloc[0],
            "topic": group["topic"].iloc[0],
            "method": "LambdaMART",
            **metrics
        })

        for rank, (_, row) in enumerate(
            group.sort_values(
                ["lambdamart_score", "video_id"],
                ascending=[False, True]
            ).iterrows(),
            start=1
        ):
            prediction_rows.append({
                "fold": fold,
                "topic_id": topic_id,
                "domain": row["domain"],
                "topic": row["topic"],
                "video_id": row["video_id"],
                "relevance": int(row["relevance"]),
                "lambdamart_score": float(row["lambdamart_score"]),
                "rank": rank,
            })

    gain = pd.DataFrame({
        "feature": FEATURES,
        "importance": model.feature_importances_,
    })
    gain["fold"] = fold
    importance_rows.append(gain)

lambda_topic = pd.DataFrame(ltr_topic_rows)
lambda_predictions = pd.DataFrame(prediction_rows)
feature_importance = pd.concat(
    importance_rows,
    ignore_index=True
)

lambda_topic.to_csv(LAMBDA_TOPIC_PATH, index=False)
lambda_predictions.to_csv(LAMBDA_PRED_PATH, index=False)
feature_importance.to_csv(FEATURE_IMPORTANCE_PATH, index=False)

# ---------------------------------------------------------------------------
# 8. Combine baseline summaries.
# ---------------------------------------------------------------------------
summary_rows = []

for method, frame in [
    ("Popularity_Views", popularity_topic),
    ("Popularity_Likes", popularity_topic),
    ("Popularity_Comments", popularity_topic),
]:
    subset = frame[frame["method"] == method]
    summary_rows.append({
        "method": method,
        **subset[METRICS].mean().to_dict()
    })

summary_rows.append({
    "method": "LambdaMART",
    **lambda_topic[METRICS].mean().to_dict()
})

summary = pd.DataFrame(summary_rows)

# Also include the already-computed Phase 10 BM25/E5/Content-IQS results.
phase10_summary = Path(
    "/content/revision_phase10/phase10b_baseline_summary.csv"
)

phase10_proposed = Path(
    "/content/revision_phase10/phase10c_proposed_content_summary.csv"
)

if phase10_summary.exists():
    p10 = pd.read_csv(phase10_summary)
    summary = pd.concat([summary, p10], ignore_index=True)

if phase10_proposed.exists():
    p10p = pd.read_csv(phase10_proposed)
    summary = pd.concat([summary, p10p], ignore_index=True)

# Remove accidental duplicates while preserving the first occurrence.
summary = summary.drop_duplicates(
    subset=["method"],
    keep="first"
)

summary.to_csv(SUMMARY_PATH, index=False)

print("\n" + "=" * 100)
print("PHASE 11 COMPLETE")
print("=" * 100)

print("\nFINAL BASELINE SUMMARY:")
print(summary[["method"] + METRICS].round(4).to_string(index=False))

print("\nSaved:")
for path in [
    STATS_PATH,
    POPULARITY_TOPIC_PATH,
    LAMBDA_TOPIC_PATH,
    LAMBDA_PRED_PATH,
    FEATURE_IMPORTANCE_PATH,
    SUMMARY_PATH,
]:
    print("  ✅", path)

print("\nLambdaMART evaluation:")
print("  Folds: 5")
print("  Grouping variable: topic_id")
print("  Test topics:", lambda_topic["topic_id"].nunique())
print("  Feature count:", len(FEATURES))
````

# PHASE 12 — EXTENDED REVISION ANALYSIS (v10 Extended)

This extension preserves the validated **99-video Phase 9 frozen cohort** and adds the remaining non-human analyses requested by the reviewers:

1. **Component-level ablation refinement** — isolates comment-semantic filtering, sentiment/emotional weighting, intrinsic quality, semantic content alignment, and adaptive weighting.
2. **LLM explanation evaluation** — generates grounded explanations for the top-ranked videos and performs an automatic consistency/faithfulness audit against the numerical evidence used by the ranking.
3. **Error analysis** — identifies ranking cases where engagement and intrinsic quality disagree and produces interpretable error categories.
4. **Runtime / complexity / cost analysis** — measures representative embedding and ranking runtime and records API-call accounting without inventing provider prices.
5. **Optional DPR / RankNet baselines** — disabled by default; enable explicitly when additional baseline strengthening is desired.

**Human educator evaluation remains intentionally deferred. No human labels are fabricated or inferred.**

````python
# ============================================================================
# PHASE 12A — COMPONENT-LEVEL ABLATION REFINEMENT
# ============================================================================
# Purpose:
#   Extend the existing Phase 9H ablation so individual components are
#   explicitly isolated rather than only removing FES/IQS as whole blocks.
#
# Variants:
#   1. no_comment_semantic_filter : sentiment is applied to all collected
#      comments for the frozen cohort, rather than only semantically filtered
#      comments. This removes the comment-level semantic filtering gate.
#   2. no_sentiment                : uses semantic relevance without sentiment
#      weighting (emotion_weighted_score -> relevance_score).
#   3. no_iqs                      : engagement-only ranking.
#   4. no_iqs_semantic_alignment   : IQS with the semantic-alignment component
#      removed and the remaining IQS weights renormalized.
#   5. fixed_50_50                 : removes adaptive weighting.
#
# IMPORTANT:
#   This remains a component/ranking ablation, NOT a ground-truth effectiveness
#   test. The independent LAK benchmark is evaluated separately in Phase 10/11.
# ============================================================================

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from scipy.stats import spearmanr

BASE9A = Path('/content/revision_phase9a')
BASE9C = Path('/content/revision_phase9c')
BASE9E = Path('/content/revision_phase9e')
BASE9F = Path('/content/revision_phase9f')
OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

hyb = pd.read_csv(BASE9F / 'phase9f_hybrid_full.csv')
iqs = pd.read_csv(BASE9E / 'phase9e_iqs_full.csv')

required = ['video_id','final_engagement_score','intrinsic_quality_score','relevant_comment_count','alpha','hybrid_score']
missing = [c for c in required if c not in hyb.columns]
if missing:
    raise ValueError(f'Missing Phase 9F columns: {missing}')

cohort = set(hyb.video_id.astype(str))
if len(cohort) != len(hyb):
    raise ValueError('Phase 9F cohort is not one-row-per-video.')

# --------------------------------------------------------------------------
# Load raw comments from Phase 9A. Phase 9C contains the already-scored
# semantically filtered comments, but does not contain the raw comments.csv files.
# --------------------------------------------------------------------------
raw_parts = []
rel_parts = []
for p in BASE9A.glob('*/comments.csv'):
    d = pd.read_csv(p)
    raw_parts.append(d[d.video_id.astype(str).isin(cohort)].copy())
for p in BASE9C.glob('*/relevant_comments_sentiment.csv'):
    d = pd.read_csv(p)
    rel_parts.append(d[d.video_id.astype(str).isin(cohort)].copy())

raw = pd.concat(raw_parts, ignore_index=True) if raw_parts else pd.DataFrame()
rel = pd.concat(rel_parts, ignore_index=True) if rel_parts else pd.DataFrame()

if raw.empty:
    raise FileNotFoundError(
        'Raw Phase 9A comment files were not found. Expected /content/revision_phase9a/*/comments.csv.'
    )
if rel.empty:
    raise FileNotFoundError(
        'Phase 9C relevant comment sentiment files were not found. Expected /content/revision_phase9c/*/relevant_comments_sentiment.csv.'
    )
print(f'Loaded raw comments: {len(raw):,}; semantically filtered + sentiment-scored comments: {len(rel):,}')

# --------------------------------------------------------------------------
# 1) No sentiment: preserve the original semantically-filtered comments but
#    remove sentiment weighting. Relevance alone becomes the comment signal.
# --------------------------------------------------------------------------
no_sent = (
    rel.groupby('video_id')
       .agg(no_sent_emotional_engagement=('relevance_score','mean'),
            no_sent_emotional_variance=('relevance_score','var'))
       .reset_index()
)

# Phase 9C can contain one comment per video in edge cases; variance is then NaN.
no_sent['no_sent_emotional_variance'] = no_sent['no_sent_emotional_variance'].fillna(0.0)

# --------------------------------------------------------------------------
# 2) No comment-semantic-filter: use every collected comment in the frozen
#    cohort. Sentiment scores must be recomputed because Phase 9C sentiment was
#    intentionally applied only after semantic filtering.
# --------------------------------------------------------------------------
RUN_FULL_COMMENT_SENTIMENT = False

if RUN_FULL_COMMENT_SENTIMENT:
    from transformers import AutoTokenizer, AutoModelForSequenceClassification
    import torch
    import re

    MODEL_NAME = 'cardiffnlp/twitter-roberta-base-sentiment-latest'
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device).eval()

    def clean_text(x):
        x = str(x)
        x = re.sub(r'http\\S+|www\\.\\S+', ' ', x)
        x = re.sub(r'[^A-Za-z0-9\\s!?.,\'"-]', ' ', x)
        return re.sub(r'\\s+', ' ', x).strip().lower()

    raw['cleaned_comment'] = raw['comment_text'].fillna('').map(clean_text)
    scores = []
    batch_size = 32
    labels = {0: -1.0, 1: 0.0, 2: 1.0}
    for start in range(0, len(raw), batch_size):
        batch = raw['cleaned_comment'].iloc[start:start+batch_size].tolist()
        enc = tokenizer(batch, padding=True, truncation=True, max_length=256, return_tensors='pt')
        enc = {k:v.to(device) for k,v in enc.items()}
        with torch.no_grad():
            logits = model(**enc).logits
            probs = torch.softmax(logits, dim=-1).cpu().numpy()
        scores.extend((probs[:,2] - probs[:,0]).tolist())
    raw['sentiment_score_full'] = np.asarray(scores, dtype=float)
    raw['emotion_weighted_score_full'] = raw['sentiment_score_full']
    no_sem = (
        raw.groupby('video_id')
           .agg(no_sem_emotional_engagement=('emotion_weighted_score_full','mean'),
                no_sem_emotional_variance=('emotion_weighted_score_full','var'),
                no_sem_comment_count=('comment_id','nunique'))
           .reset_index()
    )
    no_sem['no_sem_emotional_variance'] = no_sem['no_sem_emotional_variance'].fillna(0.0)
else:
    # Safe default: do not silently pretend the full-comment sentiment pass ran.
    # A placeholder is created so the report clearly marks this variant pending.
    no_sem = pd.DataFrame({'video_id': sorted(cohort)})
    no_sem['no_sem_emotional_engagement'] = np.nan
    no_sem['no_sem_emotional_variance'] = np.nan
    no_sem['no_sem_comment_count'] = np.nan

# --------------------------------------------------------------------------
# Helper for the exact FES formula used by Phase 9D.
# --------------------------------------------------------------------------
def minmax(series):
    series = pd.to_numeric(series, errors='coerce').astype(float)
    lo, hi = series.min(), series.max()
    if hi == lo:
        return pd.Series(np.zeros(len(series)), index=series.index)
    return (series - lo) / (hi - lo)

base = hyb[['video_id','final_engagement_score','intrinsic_quality_score','relevant_comment_count','alpha','hybrid_score']].copy()
base = base.merge(iqs[['video_id','semantic_alignment','depth_score','lexical_diversity','teaching_score_normalized']], on='video_id', how='left', validate='one_to_one')
base = base.merge(no_sent, on='video_id', how='left', validate='one_to_one')
base = base.merge(no_sem, on='video_id', how='left', validate='one_to_one')

# No-sentiment FES: only sentiment weighting is removed; comment set stays fixed.
base['no_sent_likes_norm'] = minmax(iqs.set_index('video_id').loc[base.video_id, 'likes'].reset_index(drop=True))
base['no_sent_comment_norm'] = minmax(base['relevant_comment_count'])
base['fes_no_sentiment'] = (
    0.4 * base['no_sent_emotional_engagement']
    + 0.2 * (1.0 - base['no_sent_emotional_variance'])
    + 0.2 * base['no_sent_likes_norm']
    + 0.2 * base['no_sent_comment_norm']
)

# No semantic alignment inside IQS. Remaining weights 0.20/0.20/0.30 are
# renormalized to sum to 1: depth=.285714, TTR=.285714, teaching=.428571.
base['iqs_no_semantic_alignment'] = (
    (0.20/0.70) * base['depth_score']
    + (0.20/0.70) * base['lexical_diversity']
    + (0.30/0.70) * base['teaching_score_normalized']
)

# Hybrid variants.
base['hybrid_no_sentiment'] = base['alpha'] * base['fes_no_sentiment'] + (1-base['alpha']) * base['intrinsic_quality_score']
base['hybrid_no_iqs'] = base['final_engagement_score']
base['hybrid_no_iqs_semantic_alignment'] = base['alpha'] * base['final_engagement_score'] + (1-base['alpha']) * base['iqs_no_semantic_alignment']
base['hybrid_fixed_50_50'] = 0.5 * base['final_engagement_score'] + 0.5 * base['intrinsic_quality_score']

if RUN_FULL_COMMENT_SENTIMENT:
    # With semantic filtering removed, both the engagement signal and the
    # adaptive weight's comment-count input are allowed to change.
    base['no_sem_comment_norm'] = minmax(base['no_sem_comment_count'])
    base['fes_no_comment_semantic_filter'] = (
        0.4 * base['no_sem_emotional_engagement']
        + 0.2 * (1.0 - base['no_sem_emotional_variance'])
        + 0.2 * base['no_sent_likes_norm']
        + 0.2 * base['no_sem_comment_norm']
    )
    es = np.log1p(base['no_sem_comment_count'].clip(lower=0))
    alpha_no_sem = es / (es + 5.0)
    base['hybrid_no_comment_semantic_filter'] = alpha_no_sem * base['fes_no_comment_semantic_filter'] + (1-alpha_no_sem) * base['intrinsic_quality_score']

variants = {
    'no_sentiment': 'hybrid_no_sentiment',
    'no_iqs': 'hybrid_no_iqs',
    'no_iqs_semantic_alignment': 'hybrid_no_iqs_semantic_alignment',
    'fixed_50_50': 'hybrid_fixed_50_50'
}
if RUN_FULL_COMMENT_SENTIMENT:
    variants['no_comment_semantic_filter'] = 'hybrid_no_comment_semantic_filter'

rows = []
for name, col in variants.items():
    rho, p = spearmanr(base['hybrid_score'], base[col], nan_policy='omit')
    full_rank = base['hybrid_score'].rank(method='min', ascending=False)
    var_rank = base[col].rank(method='min', ascending=False)
    top_full = set(base.nlargest(10, 'hybrid_score').video_id)
    top_var = set(base.nlargest(10, col).video_id)
    rows.append({
        'cohort_size': len(base), 'variant': name, 'score_column': col,
        'spearman_vs_full': rho, 'p_value': p,
        'mean_abs_rank_change': (var_rank-full_rank).abs().mean(),
        'max_abs_rank_change': (var_rank-full_rank).abs().max(),
        'top10_overlap': len(top_full & top_var),
        'top10_overlap_fraction': len(top_full & top_var)/10.0,
        'executed': bool(name != 'no_comment_semantic_filter' or RUN_FULL_COMMENT_SENTIMENT)
    })

ablation = pd.DataFrame(rows)
ablation.to_csv(OUT / 'phase12_component_ablation.csv', index=False)
base.to_csv(OUT / 'phase12_component_ablation_rankings.csv', index=False)

print(ablation.to_string(index=False))
print(f'\\nSaved: {OUT / "phase12_component_ablation.csv"}')
print('NOTE: no_comment_semantic_filter is disabled by default because it requires a fresh sentiment pass over all raw comments.')
````

## PHASE 12B — LLM EXPLANATION GENERATION + AUTOMATIC GROUNDING AUDIT

The explanation module is evaluated separately from the ranking benchmark. Explanations are generated only from the evidence actually available to the system: FES, IQS, adaptive weight, contributions, relevant-comment count, and dominant component.

The automatic audit checks:
- numerical consistency when values are mentioned;
- dominant-factor consistency;
- presence of evidence terms;
- explanation length/readability constraints.

This is an **automatic grounding/faithfulness audit**, not a substitute for human usefulness judgments. Human educator evaluation remains deferred.

````python

# ============================================================================
# PHASE 12B — GROUNDED, VIDEO-SPECIFIC LLM EXPLANATIONS + AUTOMATIC AUDIT
# ============================================================================
from pathlib import Path
import json, re, time
import numpy as np
import pandas as pd

BASE = Path('/content/revision_phase9f')
IQS_PATH = Path('/content/revision_phase9e/phase9e_iqs_full.csv')
OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

rank = (
    pd.read_csv(BASE / 'phase9f_hybrid_full.csv')
    .sort_values('hybrid_score', ascending=False)
    .head(10)
    .copy()
)

iqs = pd.read_csv(IQS_PATH)

extra_cols = [
    'video_id', 'semantic_alignment', 'depth_score',
    'lexical_diversity', 'teaching_score',
    'emotional_engagement', 'duration_minutes'
]
extra_cols = [c for c in extra_cols if c in iqs.columns]

rank = rank.merge(
    iqs[extra_cols].drop_duplicates('video_id'),
    on='video_id',
    how='left',
    validate='one_to_one'
)

required = [
    'video_id', 'domain', 'video_title',
    'final_engagement_score', 'intrinsic_quality_score',
    'alpha', 'fes_contribution', 'iqs_contribution',
    'relevant_comment_count', 'dominant_component'
]
missing = [c for c in required if c not in rank.columns]
if missing:
    raise ValueError(f'Missing explanation evidence columns: {missing}')

RUN_LLM_EXPLANATIONS = True
MODEL_NAME = 'gemini-3.5-flash-lite'
PROMPT_VERSION = 'Phase12B_grounded_explanation_v2_specific'

if RUN_LLM_EXPLANATIONS:
    from google import genai
    from google.colab import userdata

    api_key = userdata.get('GEMINI_API_KEY')
    if not api_key:
        raise RuntimeError('GEMINI_API_KEY not found in Colab Secrets.')

    client = genai.Client(api_key=api_key)

    records = []
    for _, r in rank.iterrows():
        rec = {
            'video_id': str(r.video_id),
            'domain': str(r.domain),
            'title': str(r.video_title),
            'FES': round(float(r.final_engagement_score), 4),
            'IQS': round(float(r.intrinsic_quality_score), 4),
            'alpha': round(float(r.alpha), 4),
            'FES_contribution': round(float(r.fes_contribution), 4),
            'IQS_contribution': round(float(r.iqs_contribution), 4),
            'relevant_comment_count': int(r.relevant_comment_count),
            'dominant_component': str(r.dominant_component)
        }

        for c in [
            'semantic_alignment', 'depth_score', 'lexical_diversity',
            'teaching_score', 'emotional_engagement', 'duration_minutes'
        ]:
            if c in rank.columns and pd.notna(r[c]):
                rec[c] = round(float(r[c]), 4)

        records.append(rec)

    prompt = f'''
You are generating evidence-grounded explanations for an educational
video ranking system.

Use ONLY the supplied evidence. Do not invent views, likes, comments,
timestamps, learning outcomes, claims about video content, or teaching
quality beyond the supplied numerical fields.

For EACH video, explain why its ranking is produced by:
1. the FES/IQS balance and adaptive alpha;
2. the dominant contribution (FES or IQS);
3. at least TWO video-specific evidence fields beyond the generic rank
   statement, selected from semantic_alignment, depth_score,
   lexical_diversity, teaching_score, emotional_engagement,
   duration_minutes, and relevant_comment_count when available.

Mention the video's distinctive title/topic phrase when useful, but do not
invent content that is not in the evidence.

If FES_contribution > IQS_contribution, engagement is dominant.
If IQS_contribution > FES_contribution, intrinsic quality is dominant.
Do not reverse these relationships.

Return ONLY a JSON array with objects containing:
{{"video_id": "...", "explanation": "..."}}

Each explanation must be 50-100 words, be specific to that video, and
contain at least three concrete evidence references, including at least
one non-FES/IQS content-specific field when available.

Evidence:
{json.dumps(records, ensure_ascii=False, indent=2)}
'''

    t_llm = time.perf_counter()
    response = client.interactions.create(model=MODEL_NAME, input=prompt)
    llm_elapsed = time.perf_counter() - t_llm

    raw = response.output_text if hasattr(response, 'output_text') else str(response)
    m = re.search(r'\[.*\]', raw, flags=re.S)
    if not m:
        raise ValueError('Could not locate JSON array in Gemini explanation response.')

    parsed = json.loads(m.group(0))
    exp_df = pd.DataFrame(parsed)

    usage = getattr(response, 'usage', None)
    usage_dict = {}
    if usage is not None:
        try:
            if hasattr(usage, 'model_dump'):
                usage_dict = usage.model_dump()
            elif hasattr(usage, '__dict__'):
                usage_dict = dict(usage.__dict__)
            elif isinstance(usage, dict):
                usage_dict = usage
        except Exception:
            usage_dict = {}

    with open(OUT / 'phase12_llm_runtime.json', 'w', encoding='utf-8') as f:
        json.dump({
            'model': MODEL_NAME,
            'prompt_version': PROMPT_VERSION,
            'explanations_generated': int(len(exp_df)),
            'elapsed_seconds': float(llm_elapsed),
            'usage': usage_dict
        }, f, indent=2, default=str)

else:
    exp_df = pd.DataFrame(columns=['video_id', 'explanation'])
    with open(OUT / 'phase12_llm_runtime.json', 'w', encoding='utf-8') as f:
        json.dump({
            'model': MODEL_NAME,
            'prompt_version': PROMPT_VERSION,
            'explanations_generated': 0,
            'elapsed_seconds': None,
            'usage': {}
        }, f, indent=2)

if 'video_id' not in exp_df.columns or 'explanation' not in exp_df.columns:
    raise ValueError('LLM output must contain video_id and explanation fields.')

exp_df['video_id'] = exp_df['video_id'].astype(str)

exp_df = rank.merge(
    exp_df[['video_id', 'explanation']],
    on='video_id',
    how='left',
    validate='one_to_one'
)

def dominant_ok(row, text):
    t = str(text).lower()
    dom = str(row.dominant_component).lower()
    if 'engagement' in dom or 'fes' in dom:
        return any(k in t for k in ['engagement', 'audience', 'feedback', 'fes'])
    return any(k in t for k in ['intrinsic', 'quality', 'iqs', 'content'])

def mentioned_numbers_consistent(row, text, tol=0.03):
    nums = [
        float(x) for x in re.findall(
            r'(?<![A-Za-z])0?\.\d{2,4}(?!\d)', str(text)
        )
    ]

    evidence = []
    for c in [
        'final_engagement_score', 'intrinsic_quality_score', 'alpha',
        'fes_contribution', 'iqs_contribution', 'semantic_alignment',
        'depth_score', 'lexical_diversity', 'teaching_score',
        'emotional_engagement'
    ]:
        if c in row.index and pd.notna(row[c]):
            evidence.append(float(row[c]))

    return bool(nums) and any(
        abs(n - e) <= tol for n in nums for e in evidence
    )

def content_specific_ok(row, text):
    t = str(text).lower()
    terms = [
        'semantic', 'alignment', 'teaching', 'depth', 'lexical',
        'diversity', 'duration', 'emotional', 'engagement',
        'relevant comments', 'comments'
    ]
    return int(sum(term in t for term in terms) >= 1)

rows = []
for _, r in exp_df.iterrows():
    text = str(r.get('explanation', ''))
    words = re.findall(r'\b\w+\b', text)

    rows.append({
        'video_id': r.video_id,
        'explanation_present': bool(text.strip()),
        'word_count': len(words),
        'length_ok': 50 <= len(words) <= 120,
        'dominant_factor_consistent': dominant_ok(r, text) if text.strip() else False,
        'numeric_evidence_consistent': mentioned_numbers_consistent(r, text) if text.strip() else False,
        'content_specific_evidence_present': bool(content_specific_ok(r, text)) if text.strip() else False,
        'evidence_grounded': (
            bool(text.strip())
            and dominant_ok(r, text)
            and mentioned_numbers_consistent(r, text)
            and bool(content_specific_ok(r, text))
        ),
        'human_usefulness_evaluation': 'DEFERRED'
    })

audit = pd.DataFrame(rows)

merged = exp_df.merge(
    audit, on='video_id', how='left', validate='one_to_one'
)

exp_path = OUT / 'phase12_explanations.csv'
audit_path = OUT / 'phase12_explanation_audit.csv'

merged.to_csv(exp_path, index=False)
audit.to_csv(audit_path, index=False)

print('Explanation audit:')
print(audit.to_string(index=False))
print(f'\nGrounded explanations: {audit.evidence_grounded.mean():.3f}')
print(f'Saved: {exp_path}')
print(f'Saved: {audit_path}')
print('Saved measured LLM timing:', OUT / 'phase12_llm_runtime.json')
````

## PHASE 12C — ERROR ANALYSIS

This analysis does not label the system as “wrong” without ground truth. Instead, it identifies **diagnostic disagreement cases** where engagement and intrinsic quality pull the hybrid ranking in different directions. These are useful for the Results/Discussion and Limitations sections.

````python
# ============================================================================
# PHASE 12C — ERROR / FAILURE-MODE ANALYSIS
# ============================================================================
#
# Purpose:
#   Identify representative cases where FES and IQS disagree strongly,
#   including:
#     1. High engagement / lower intrinsic quality
#     2. High intrinsic quality / lower engagement
#     3. Large FES-IQS rank disagreement
#     4. Engagement-dominant adaptive weighting
#     5. Quality-dominant adaptive weighting
#
# Important:
#   This is diagnostic/error analysis only.
#   It does NOT claim that a video is objectively "wrong" or "bad".
# ============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


# ============================================================================
# 1. PATHS
# ============================================================================

BASE = Path('/content/revision_phase9f')
OUT = Path('/content/revision_phase12')

OUT.mkdir(parents=True, exist_ok=True)

INPUT_FILE = BASE / 'phase9f_hybrid_full.csv'

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f'Phase 9F file not found:\n{INPUT_FILE}'
    )


# ============================================================================
# 2. LOAD PHASE 9F DATA
# ============================================================================

df = pd.read_csv(INPUT_FILE)

print(f'Loaded Phase 9F data: {len(df)} rows')
print(f'Columns available: {list(df.columns)}')


# ============================================================================
# 3. CHECK REQUIRED CORE COLUMNS
# ============================================================================

required_columns = [
    'video_id',
    'final_engagement_score',
    'intrinsic_quality_score',
    'alpha',
    'hybrid_score'
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f'Missing required Phase 9F columns: {missing}'
    )


# ============================================================================
# 4. ENSURE NUMERIC TYPES
# ============================================================================

numeric_columns = [
    'final_engagement_score',
    'intrinsic_quality_score',
    'alpha',
    'hybrid_score'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )


# Remove rows that cannot be analyzed
before = len(df)

df = df.dropna(
    subset=[
        'final_engagement_score',
        'intrinsic_quality_score',
        'alpha',
        'hybrid_score'
    ]
).copy()

after = len(df)

if before != after:
    print(
        f'Removed {before - after} rows with missing core scores.'
    )


# ============================================================================
# 5. CREATE / RECREATE RANKS
# ============================================================================
#
# Do not assume these columns already exist in Phase 9F.
# We explicitly calculate them from the scores.
# ============================================================================

df['fes_rank'] = (
    df['final_engagement_score']
    .rank(method='min', ascending=False)
    .astype(int)
)

df['iqs_rank'] = (
    df['intrinsic_quality_score']
    .rank(method='min', ascending=False)
    .astype(int)
)

df['final_rank'] = (
    df['hybrid_score']
    .rank(method='min', ascending=False)
    .astype(int)
)


# ============================================================================
# 6. FES-IQS RANK DIFFERENCE
# ============================================================================

df['fes_rank_gap'] = (
    df['fes_rank'] - df['iqs_rank']
)

df['abs_fes_iqs_rank_gap'] = (
    df['fes_rank_gap'].abs()
)


# ============================================================================
# 7. CREATE / RECREATE DOMINANT COMPONENT
# ============================================================================
#
# The adaptive hybrid score is:
#
# Hybrid = alpha * FES + (1-alpha) * IQS
#
# Therefore:
#   alpha >= 0.5 -> engagement contribution is at least as large
#   alpha <  0.5 -> IQS contribution is larger
#
# We calculate actual weighted contributions as well.
# ============================================================================

df['engagement_contribution'] = (
    df['alpha'] *
    df['final_engagement_score']
)

df['quality_contribution'] = (
    (1.0 - df['alpha']) *
    df['intrinsic_quality_score']
)

df['dominant_component'] = np.where(
    df['engagement_contribution'] >= df['quality_contribution'],
    'Engagement',
    'Intrinsic Quality'
)


# ============================================================================
# 8. OPTIONAL METADATA COLUMNS
# ============================================================================
#
# The diagnostic analysis should still work if some descriptive columns
# are absent from Phase 9F.
# ============================================================================

if 'domain' not in df.columns:
    df['domain'] = 'Unknown'

if 'video_title' not in df.columns:
    df['video_title'] = df['video_id'].astype(str)


# ============================================================================
# 9. DATA-DRIVEN QUARTILES
# ============================================================================

fes_q75 = df['final_engagement_score'].quantile(0.75)
fes_q25 = df['final_engagement_score'].quantile(0.25)

iqs_q75 = df['intrinsic_quality_score'].quantile(0.75)
iqs_q25 = df['intrinsic_quality_score'].quantile(0.25)


print('\nQuartile thresholds:')
print(f'FES 25th percentile: {fes_q25:.6f}')
print(f'FES 75th percentile: {fes_q75:.6f}')
print(f'IQS 25th percentile: {iqs_q25:.6f}')
print(f'IQS 75th percentile: {iqs_q75:.6f}')


# ============================================================================
# 10. DIAGNOSTIC CATEGORY FUNCTION
# ============================================================================

def assign_diagnostic_category(row):

    fes = row['final_engagement_score']
    iqs = row['intrinsic_quality_score']

    fes_hi = fes >= fes_q75
    iqs_hi = iqs >= iqs_q75

    fes_lo = fes <= fes_q25
    iqs_lo = iqs <= iqs_q25

    rank_gap = abs(
        row['fes_rank'] -
        row['iqs_rank']
    )

    # --------------------------------------------------------
    # Case 1: High engagement but lower intrinsic quality
    # --------------------------------------------------------
    if fes_hi and iqs_lo:
        return 'High engagement / lower intrinsic quality'

    # --------------------------------------------------------
    # Case 2: High intrinsic quality but lower engagement
    # --------------------------------------------------------
    if iqs_hi and fes_lo:
        return 'High intrinsic quality / lower engagement'

    # --------------------------------------------------------
    # Case 3: Large disagreement between FES and IQS rankings
    # --------------------------------------------------------
    if rank_gap >= 25:
        return 'Large FES-IQS rank disagreement'

    # --------------------------------------------------------
    # Case 4: Strong engagement dominance
    # --------------------------------------------------------
    if row['alpha'] >= 0.80:
        return 'Engagement-dominant adaptive weight'

    # --------------------------------------------------------
    # Case 5: Strong quality dominance
    # --------------------------------------------------------
    if row['alpha'] <= 0.40:
        return 'Quality-dominant adaptive weight'

    # --------------------------------------------------------
    # Default
    # --------------------------------------------------------
    return 'No major diagnostic disagreement'


df['diagnostic_category'] = df.apply(
    assign_diagnostic_category,
    axis=1
)


# ============================================================================
# 11. CATEGORY SUMMARY
# ============================================================================

summary = (
    df['diagnostic_category']
    .value_counts()
    .rename_axis('diagnostic_category')
    .reset_index(name='videos')
)

summary['percentage'] = (
    100.0 *
    summary['videos'] /
    len(df)
)


# ============================================================================
# 12. SELECT MOST INFORMATIVE CASES
# ============================================================================
#
# We select the largest FES-IQS rank disagreements first.
# ============================================================================

example_columns = [
    'video_id',
    'domain',
    'video_title',
    'final_engagement_score',
    'intrinsic_quality_score',
    'alpha',
    'engagement_contribution',
    'quality_contribution',
    'hybrid_score',
    'fes_rank',
    'iqs_rank',
    'fes_rank_gap',
    'abs_fes_iqs_rank_gap',
    'final_rank',
    'dominant_component',
    'diagnostic_category'
]

examples = (
    df.sort_values(
        'abs_fes_iqs_rank_gap',
        ascending=False
    )[example_columns]
    .head(20)
    .copy()
)


# ============================================================================
# 13. SELECT REPRESENTATIVE CASES BY CATEGORY
# ============================================================================
#
# This is useful later when writing the manuscript's Error Analysis section.
# ============================================================================

representative_cases = (
    df.sort_values(
        'abs_fes_iqs_rank_gap',
        ascending=False
    )
    .groupby(
        'diagnostic_category',
        group_keys=False
    )
    .head(5)
    .copy()
)


# ============================================================================
# 14. SAVE OUTPUTS
# ============================================================================

summary_file = (
    OUT /
    'phase12_error_analysis_summary.csv'
)

examples_file = (
    OUT /
    'phase12_error_analysis_examples.csv'
)

representative_file = (
    OUT /
    'phase12_error_analysis_representative_cases.csv'
)

full_file = (
    OUT /
    'phase12_error_analysis_full.csv'
)


summary.to_csv(
    summary_file,
    index=False
)

examples.to_csv(
    examples_file,
    index=False
)

representative_cases.to_csv(
    representative_file,
    index=False
)

df.to_csv(
    full_file,
    index=False
)


# ============================================================================
# 15. DISPLAY RESULTS
# ============================================================================

print('\n' + '=' * 80)
print('PHASE 12C — ERROR / FAILURE-MODE ANALYSIS')
print('=' * 80)

print(f'\nAnalyzed cohort: {len(df)} videos')

print('\nDiagnostic categories:')
print(
    summary.to_string(
        index=False
    )
)

print('\n' + '-' * 80)
print('Top 20 FES-IQS disagreement cases:')
print('-' * 80)

display_columns = [
    'video_id',
    'domain',
    'final_engagement_score',
    'intrinsic_quality_score',
    'alpha',
    'fes_rank',
    'iqs_rank',
    'final_rank',
    'diagnostic_category'
]

print(
    examples[display_columns]
    .to_string(index=False)
)


print('\n' + '-' * 80)
print('Representative cases by diagnostic category:')
print('-' * 80)

print(
    representative_cases[
        display_columns
    ].to_string(index=False)
)


# ============================================================================
# 16. OUTPUT LOCATIONS
# ============================================================================

print('\nSaved files:')

print(f'  Summary:')
print(f'    {summary_file}')

print(f'  Top disagreement examples:')
print(f'    {examples_file}')

print(f'  Representative cases:')
print(f'    {representative_file}')

print(f'  Full diagnostic dataset:')
print(f'    {full_file}')

print('\nPhase 12C completed successfully.')
````

## PHASE 12D — RUNTIME, COMPLEXITY, SCALABILITY, AND COST ACCOUNTING (REVISED)

This phase reports measured local runtime, memory, theoretical complexity, API request/token accounting, and a provider-list-price cost calculation from the measured Gemini token usage.
It does **not** infer or fabricate an actual invoice amount. The monetary calculation uses the official Gemini 3.5 Flash-Lite Standard text-token list prices captured on 2026-09-30 ($0.30 per 1M input tokens and $2.50 per 1M output tokens). If the account used a free tier, credits, discounts, or another billing plan, the actual billed amount can differ.

The experiment distinguishes:
- deterministic local ranking runtime;
- E5 model loading/encoding runtime;
- measured process memory where available;
- LLM request count/token accounting when available;
- measured paid-list-price equivalent for the recorded token usage;
- optional live LLM latency measurement.


````python
# ============================================================================
# PHASE 12D — RUNTIME / COMPLEXITY / MEMORY / COST ANALYSIS (FINAL)
# ============================================================================

from pathlib import Path
import time
import math
import os
import json
import gc

import numpy as np
import pandas as pd

# ---------------------------------------------------------------------------
# Paths
# ---------------------------------------------------------------------------

OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

HYB_PATH = Path('/content/revision_phase9f/phase9f_hybrid_full.csv')
IQS_PATH = Path('/content/revision_phase9e/phase9e_iqs_full.csv')

if not HYB_PATH.exists():
    raise FileNotFoundError(f"Missing required file: {HYB_PATH}")

if not IQS_PATH.exists():
    raise FileNotFoundError(f"Missing required file: {IQS_PATH}")

hyb = pd.read_csv(HYB_PATH)
iqs = pd.read_csv(IQS_PATH)

print("=" * 80)
print("PHASE 12D — RUNTIME / COMPLEXITY / MEMORY / COST ANALYSIS")
print("=" * 80)
print(f"Hybrid rows: {len(hyb)}")
print(f"IQS rows:    {len(iqs)}")


# ---------------------------------------------------------------------------
# Memory helper
# ---------------------------------------------------------------------------

def rss_mb():
    """
    Return current process resident memory in MB.
    Returns NaN if psutil is unavailable.
    """
    try:
        import psutil
        process = psutil.Process(os.getpid())
        return process.memory_info().rss / (1024 ** 2)
    except Exception:
        return np.nan


# ---------------------------------------------------------------------------
# Runtime measurements
# ---------------------------------------------------------------------------

rows = []


# ===========================================================================
# 1. FES / IQS SCORE FUSION + SORTING
# ===========================================================================

required_hyb_cols = [
    'final_engagement_score',
    'intrinsic_quality_score'
]

missing_hyb = [c for c in required_hyb_cols if c not in hyb.columns]

if missing_hyb:
    raise KeyError(
        f"Missing required Hybrid columns: {missing_hyb}"
    )

rss_before = rss_mb()

t0 = time.perf_counter()

hybrid_scores = (
    0.4 * hyb['final_engagement_score'].astype(float)
    + 0.6 * hyb['intrinsic_quality_score'].astype(float)
)

_ = hybrid_scores.sort_values(ascending=False)

elapsed = time.perf_counter() - t0

rss_after = rss_mb()

rows.append({
    'stage': 'FES/IQS score fusion + sorting',
    'n_videos': len(hyb),
    'elapsed_seconds': elapsed,
    'memory_before_mb': rss_before,
    'memory_after_mb': rss_after,
    'memory_delta_mb': (
        rss_after - rss_before
        if not np.isnan(rss_before) and not np.isnan(rss_after)
        else np.nan
    ),
    'measured': True,
    'notes': 'Local deterministic computation; ranking sort is O(N log N)'
})


# ===========================================================================
# 2. E5 RUNTIME
# ===========================================================================

RUN_E5_RUNTIME = True

if RUN_E5_RUNTIME:

    try:
        from sentence_transformers import SentenceTransformer
    except ImportError as e:
        raise ImportError(
            "sentence-transformers is required for the E5 runtime measurement."
        ) from e

    required_iqs_cols = [
        'video_title',
        'description',
        'semantic_query'
    ]

    missing_iqs = [
        c for c in required_iqs_cols
        if c not in iqs.columns
    ]

    if missing_iqs:
        raise KeyError(
            f"Missing required IQS columns: {missing_iqs}"
        )

    texts = (
        iqs['video_title'].fillna('').astype(str)
        + ' '
        + iqs['description'].fillna('').astype(str)
    ).tolist()

    queries = (
        iqs['semantic_query']
        .fillna('')
        .astype(str)
        .tolist()
    )

    # -----------------------------------------------------------------------
    # E5 model loading
    # -----------------------------------------------------------------------

    rss_before = rss_mb()

    t0 = time.perf_counter()

    model = SentenceTransformer('intfloat/e5-base-v2')

    load_time = time.perf_counter() - t0

    rss_after = rss_mb()

    rows.append({
        'stage': 'E5 model load',
        'n_videos': len(hyb),
        'elapsed_seconds': load_time,
        'memory_before_mb': rss_before,
        'memory_after_mb': rss_after,
        'memory_delta_mb': (
            rss_after - rss_before
            if not np.isnan(rss_before) and not np.isnan(rss_after)
            else np.nan
        ),
        'measured': True,
        'notes': 'intfloat/e5-base-v2'
    })

    # -----------------------------------------------------------------------
    # E5 content encoding
    # -----------------------------------------------------------------------

    rss_before = rss_mb()

    t0 = time.perf_counter()

    _ = model.encode(
        ['passage: ' + x for x in texts],
        batch_size=16,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    content_time = time.perf_counter() - t0

    rss_after = rss_mb()

    rows.append({
        'stage': 'E5 content encoding',
        'n_videos': len(hyb),
        'elapsed_seconds': content_time,
        'memory_before_mb': rss_before,
        'memory_after_mb': rss_after,
        'memory_delta_mb': (
            rss_after - rss_before
            if not np.isnan(rss_before) and not np.isnan(rss_after)
            else np.nan
        ),
        'measured': True,
        'notes': 'Frozen cohort title + description passages'
    })

    # -----------------------------------------------------------------------
    # E5 query encoding
    # -----------------------------------------------------------------------

    rss_before = rss_mb()

    t0 = time.perf_counter()

    _ = model.encode(
        ['query: ' + x for x in queries],
        batch_size=16,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    query_time = time.perf_counter() - t0

    rss_after = rss_mb()

    rows.append({
        'stage': 'E5 query encoding',
        'n_videos': len(hyb),
        'elapsed_seconds': query_time,
        'memory_before_mb': rss_before,
        'memory_after_mb': rss_after,
        'memory_delta_mb': (
            rss_after - rss_before
            if not np.isnan(rss_before) and not np.isnan(rss_after)
            else np.nan
        ),
        'measured': True,
        'notes': 'Frozen cohort domain-specific queries'
    })

    del model
    gc.collect()


# ===========================================================================
# 3. LLM EXPLANATION RUNTIME
# ===========================================================================

# IMPORTANT:
# Use the actual timing captured during Phase 12B.
# Do NOT make a second Gemini request.
# Do NOT infer latency from request count.

llm_path = OUT / 'phase12_llm_runtime.json'

if llm_path.exists():

    with open(llm_path, 'r', encoding='utf-8') as f:
        llm_info = json.load(f)

    elapsed_llm = llm_info.get('elapsed_seconds')

    rows.append({
        'stage': 'LLM explanation generation',
        'n_videos': llm_info.get('explanations_generated', 0),
        'elapsed_seconds': elapsed_llm,
        'memory_before_mb': np.nan,
        'memory_after_mb': np.nan,
        'memory_delta_mb': np.nan,
        'measured': elapsed_llm is not None,
        'notes': (
            f"Measured in Phase 12B using "
            f"{llm_info.get('model', 'unknown')}; "
            f"prompt_version="
            f"{llm_info.get('prompt_version', 'unknown')}; "
            f"usage="
            f"{json.dumps(llm_info.get('usage', {}), default=str)}"
        )
    })

else:

    rows.append({
        'stage': 'LLM explanation generation',
        'n_videos': 0,
        'elapsed_seconds': np.nan,
        'memory_before_mb': np.nan,
        'memory_after_mb': np.nan,
        'memory_delta_mb': np.nan,
        'measured': False,
        'notes': (
            'Run Phase 12B first. '
            'No latency is inferred or fabricated.'
        )
    })


# ===========================================================================
# 4. SAVE RUNTIME MEASUREMENTS
# ===========================================================================

runtime_df = pd.DataFrame(rows)

runtime_path = OUT / 'phase12_runtime_measurements.csv'

runtime_df.to_csv(
    runtime_path,
    index=False
)


# ===========================================================================
# 5. COMPLEXITY SUMMARY
# ===========================================================================

complexity = pd.DataFrame([
    [
        'Comment semantic filtering',
        'O(C*d)',
        'C comments; d embedding dimension'
    ],
    [
        'Sentiment inference',
        'O(C*t)',
        'C comments; t transformer inference cost'
    ],
    [
        'FES aggregation',
        'O(C)',
        'Aggregate comment-level scores to videos'
    ],
    [
        'IQS computation',
        'O(N)',
        'N videos'
    ],
    [
        'Adaptive hybrid fusion',
        'O(N)',
        'N videos'
    ],
    [
        'Ranking/sorting',
        'O(N log N)',
        'N videos'
    ],
    [
        'LLM explanations',
        'O(B) API batches',
        'Provider-side cost additionally depends on prompt/output tokens'
    ]
], columns=[
    'component',
    'complexity',
    'interpretation'
])

complexity_path = OUT / 'phase12_complexity_summary.csv'

complexity.to_csv(
    complexity_path,
    index=False
)


# ===========================================================================
# 6. PRINT RESULTS
# ===========================================================================

print("\n" + "=" * 80)
print("RUNTIME MEASUREMENTS")
print("=" * 80)

print(
    runtime_df.to_string(index=False)
)

print("\n" + "=" * 80)
print("COMPLEXITY SUMMARY")
print("=" * 80)

print(
    complexity.to_string(index=False)
)

print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)

print(f"Saved: {runtime_path}")
print(f"Saved: {complexity_path}")

print("\nPhase 12D completed successfully.")
````

## PHASE 12D-A — EXACT TOKEN-BASED MONETARY COST (GEMINI 3.5 FLASH-LITE)

This cell converts the **measured token usage recorded by Phase 12B** into a deterministic provider-list-price equivalent.

Official Google Gemini API Standard pricing captured on **2026-09-30**:
- Input: **$0.30 / 1,000,000 tokens**
- Output: **$2.50 / 1,000,000 tokens**

Source: https://ai.google.dev/gemini-api/docs/pricing

The calculation does not make another API call. It reads `phase12_llm_runtime.json`. It reports the exact mathematical cost at the captured list prices. Actual billing can be lower or zero when free-tier quota, credits, discounts, or another billing plan applies.


````python
# ============================================================================
# PHASE 12D-A — TOKEN-BASED MONETARY COST (FINAL)
# ============================================================================
from pathlib import Path
import json, math
import pandas as pd

OUT = Path('/content/revision_phase12')
USAGE_PATH = OUT / 'phase12_llm_runtime.json'
COST_PATH = OUT / 'phase12_monetary_cost.csv'

if not USAGE_PATH.exists():
    raise FileNotFoundError(
        f'{USAGE_PATH} not found. Run Phase 12B first so measured token usage is recorded.'
    )

with open(USAGE_PATH, 'r', encoding='utf-8') as f:
    info = json.load(f)

usage = info.get('usage', {}) or {}

# Gemini 3.5 Flash-Lite Standard text-token list prices captured 2026-09-30.
# Source: https://ai.google.dev/gemini-api/docs/pricing
INPUT_USD_PER_1M = 0.30
OUTPUT_USD_PER_1M = 2.50
PRICING_DATE = '2026-09-30'
PRICING_TIER = 'Standard paid tier'

# Prefer the explicit aggregate token fields returned by the Gemini API.
input_tokens = usage.get('total_input_tokens')
output_tokens = usage.get('total_output_tokens')

# Fallback extraction for future SDK response-shape changes.
if input_tokens is None:
    input_tokens = usage.get('input_tokens')
if output_tokens is None:
    output_tokens = usage.get('output_tokens')

if input_tokens is None or output_tokens is None:
    raise ValueError(
        'Measured Gemini input/output token counts are unavailable. '
        'The notebook refuses to estimate monetary cost from character/word counts.'
    )

input_tokens = int(input_tokens)
output_tokens = int(output_tokens)

input_cost_usd = (input_tokens / 1_000_000.0) * INPUT_USD_PER_1M
output_cost_usd = (output_tokens / 1_000_000.0) * OUTPUT_USD_PER_1M
paid_list_price_cost_usd = input_cost_usd + output_cost_usd

explanations = int(info.get('explanations_generated', 0) or 0)
cost_per_explanation = (
    paid_list_price_cost_usd / explanations if explanations > 0 else math.nan
)

cost_df = pd.DataFrame([{
    'model': info.get('model', 'gemini-3.5-flash-lite'),
    'pricing_date': PRICING_DATE,
    'pricing_tier': PRICING_TIER,
    'input_tokens': input_tokens,
    'output_tokens': output_tokens,
    'total_tokens': input_tokens + output_tokens,
    'input_price_usd_per_1m': INPUT_USD_PER_1M,
    'output_price_usd_per_1m': OUTPUT_USD_PER_1M,
    'input_cost_usd': input_cost_usd,
    'output_cost_usd': output_cost_usd,
    'paid_list_price_cost_usd': paid_list_price_cost_usd,
    'explanations_generated': explanations,
    'paid_list_price_cost_per_explanation_usd': cost_per_explanation,
    'free_tier_cost_if_fully_covered_usd': 0.0,
    'pricing_source': 'https://ai.google.dev/gemini-api/docs/pricing'
}])

cost_df.to_csv(COST_PATH, index=False)

print('=' * 80)
print('PHASE 12D-A — GEMINI MONETARY COST')
print('=' * 80)
print(f'Model: {info.get("model", "unknown")}')
print(f'Input tokens: {input_tokens:,}')
print(f'Output tokens: {output_tokens:,}')
print(f'Input cost: ${input_cost_usd:.8f}')
print(f'Output cost: ${output_cost_usd:.8f}')
print(f'Paid-list-price equivalent: ${paid_list_price_cost_usd:.8f}')
if explanations:
    print(f'Cost per explanation at paid list price: ${cost_per_explanation:.8f}')
print('\nIMPORTANT: this is the exact mathematical list-price equivalent for the measured tokens.')
print('Actual billing may be lower/zero under free-tier quota, credits, discounts, or another billing plan.')
print(f'Saved: {COST_PATH}')
````

## PHASE 12E — OPTIONAL DPR / RANKNET BASELINES

These baselines are intentionally **disabled by default** because they add model-download and training time. They can be enabled independently.

- **DPR**: `facebook/dpr-question_encoder-single-nq-base` + `facebook/dpr-ctx_encoder-single-nq-base`, evaluated against the independent LAK-2026 labels.
- **RankNet**: a lightweight pairwise neural Learning-to-Rank model trained only on training topics using content/popularity features; validation topics are held out with GroupKFold to avoid topic leakage.

The output format matches the existing Phase 10/11 ranking metrics so the methods can be inserted into the comparison table.

````python
# ============================================================================
# PHASE 12E-A — OPTIONAL DPR BASELINE
# ============================================================================
# Dense Passage Retrieval baseline for the LAK-2026 benchmark.
#
# Query  : topic
# Context: educational video content
# Score  : DPR context embedding · DPR question embedding
#
# The prediction file is evaluated later using the same topic-level ranking
# metrics as the other LAK benchmark baselines.
# ============================================================================

RUN_DPR = True

from pathlib import Path
import numpy as np
import pandas as pd

OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

BENCH_PATH = Path('/content/revision_phase10/lak2026_benchmark_eval.csv')

if not BENCH_PATH.exists():
    raise FileNotFoundError(f'Benchmark file not found: {BENCH_PATH}')

BENCH = pd.read_csv(BENCH_PATH)

required_cols = [
    'topic_id',
    'domain',
    'topic',
    'video_id',
    'content',
    'relevance'
]

missing = [c for c in required_cols if c not in BENCH.columns]
if missing:
    raise ValueError(f'Missing benchmark columns: {missing}')

if RUN_DPR:

    import torch
    from transformers import (
        DPRQuestionEncoder,
        DPRQuestionEncoderTokenizer,
        DPRContextEncoder,
        DPRContextEncoderTokenizer
    )

    # ------------------------------------------------------------------------
    # Models
    # ------------------------------------------------------------------------

    Q_MODEL = 'facebook/dpr-question_encoder-single-nq-base'
    C_MODEL = 'facebook/dpr-ctx_encoder-single-nq-base'

    print('Loading DPR question encoder...')
    qtok = DPRQuestionEncoderTokenizer.from_pretrained(Q_MODEL)
    qenc = DPRQuestionEncoder.from_pretrained(Q_MODEL)

    print('Loading DPR context encoder...')
    ctok = DPRContextEncoderTokenizer.from_pretrained(C_MODEL)
    cenc = DPRContextEncoder.from_pretrained(C_MODEL)

    device = torch.device(
        'cuda' if torch.cuda.is_available() else 'cpu'
    )

    print('DPR device:', device)

    qenc.to(device).eval()
    cenc.to(device).eval()

    # ------------------------------------------------------------------------
    # Embedding helpers
    # ------------------------------------------------------------------------

    def embed_questions(texts):
        """
        Encode query/topic text using the DPR question encoder.
        """
        outputs = []

        for text in texts:
            tokens = qtok(
                [str(text)],
                padding=True,
                truncation=True,
                max_length=256,
                return_tensors='pt'
            )

            tokens = {
                k: v.to(device)
                for k, v in tokens.items()
            }

            with torch.no_grad():
                emb = qenc(**tokens).pooler_output

            outputs.append(
                emb.detach().cpu().numpy()[0]
            )

        return np.vstack(outputs)


    def embed_contexts(texts, batch_size=16):
        """
        Encode video content using the DPR context encoder.
        """
        outputs = []

        texts = [str(x) for x in texts]

        for start in range(0, len(texts), batch_size):

            batch = texts[start:start + batch_size]

            tokens = ctok(
                batch,
                padding=True,
                truncation=True,
                max_length=256,
                return_tensors='pt'
            )

            tokens = {
                k: v.to(device)
                for k, v in tokens.items()
            }

            with torch.no_grad():
                emb = cenc(**tokens).pooler_output

            outputs.append(
                emb.detach().cpu().numpy()
            )

        return np.vstack(outputs)


    # ------------------------------------------------------------------------
    # Generate topic-level DPR rankings
    # ------------------------------------------------------------------------

    predictions = []

    topic_groups = list(BENCH.groupby('topic_id', sort=True))

    print('Number of benchmark topics:', len(topic_groups))

    for topic_id, g in topic_groups:

        topic = str(g['topic'].iloc[0])

        # Encode topic/query
        qv = embed_questions([topic])[0]

        # Encode all video contents for this topic
        contexts = (
            g['content']
            .fillna('')
            .astype(str)
            .tolist()
        )

        cv = embed_contexts(
            contexts,
            batch_size=16
        )

        # Standard DPR relevance score:
        # question embedding · context embedding
        scores = cv @ qv

        tmp = g[
            [
                'topic_id',
                'domain',
                'topic',
                'video_id',
                'relevance'
            ]
        ].copy()

        tmp['dpr_score'] = scores

        # Explicit ranking from the actual DPR scores
        tmp = tmp.sort_values(
            'dpr_score',
            ascending=False,
            kind='mergesort'
        ).reset_index(drop=True)

        tmp['rank'] = np.arange(
            1,
            len(tmp) + 1
        )

        predictions.append(tmp)

        print(
            f'Topic {topic_id}: '
            f'{len(tmp)} resources ranked'
        )


    # ------------------------------------------------------------------------
    # Save predictions
    # ------------------------------------------------------------------------

    dpr_pred = pd.concat(
        predictions,
        ignore_index=True
    )

    expected_rows = len(BENCH)

    if len(dpr_pred) != expected_rows:
        raise ValueError(
            f'DPR row mismatch: '
            f'expected {expected_rows}, got {len(dpr_pred)}'
        )

    if dpr_pred['video_id'].nunique() != BENCH['video_id'].nunique():
        raise ValueError(
            'DPR video coverage does not match benchmark.'
        )

    if dpr_pred['topic_id'].nunique() != BENCH['topic_id'].nunique():
        raise ValueError(
            'DPR topic coverage does not match benchmark.'
        )

    if dpr_pred['dpr_score'].isna().any():
        raise ValueError(
            'DPR predictions contain NaN scores.'
        )

    dpr_path = OUT / 'phase12_dpr_predictions.csv'

    dpr_pred.to_csv(
        dpr_path,
        index=False
    )

    print()
    print('DPR predictions saved:')
    print(dpr_path)
    print('Rows:', len(dpr_pred))
    print('Topics:', dpr_pred['topic_id'].nunique())
    print('Videos:', dpr_pred['video_id'].nunique())

else:

    print(
        'DPR baseline is DISABLED. '
        'Set RUN_DPR=True to execute it.'
    )
````

````python
# ============================================================================
# PHASE 12E-B — OPTIONAL RANKNET BASELINE
# ============================================================================
# Pairwise RankNet baseline evaluated using 5-fold GroupKFold by topic.
#
# Important:
# - Topics are kept completely separated between training and validation.
# - The same broad feature family used by the LambdaMART baseline is used.
# - Predictions are generated only for held-out topics.
# ============================================================================

RUN_RANKNET = True

from pathlib import Path
import numpy as np
import pandas as pd

OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

BENCH_PATH = Path('/content/revision_phase10/lak2026_benchmark_eval.csv')
POP_PATH = Path('/content/revision_phase11/youtube_popularity_stats.csv')
FEAT_PATH = Path('/content/revision_phase10/phase10c_proposed_content_rankings.csv')

if not BENCH_PATH.exists():
    raise FileNotFoundError(BENCH_PATH)

if not POP_PATH.exists():
    raise FileNotFoundError(POP_PATH)

if not FEAT_PATH.exists():
    raise FileNotFoundError(FEAT_PATH)

BENCH = pd.read_csv(BENCH_PATH)
POP = pd.read_csv(POP_PATH)
FEAT = pd.read_csv(FEAT_PATH)

if RUN_RANKNET:

    import torch
    import torch.nn as nn
    from sklearn.model_selection import GroupKFold
    from sklearn.preprocessing import StandardScaler

    # Reproducibility: fix both CPU and CUDA RNGs for RankNet initialization.
    torch.manual_seed(42)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(42)

    # ------------------------------------------------------------------------
    # Merge benchmark + content features + popularity features
    # ------------------------------------------------------------------------

    required_feat = [
        'topic_id',
        'video_id',
        'semantic_alignment',
        'depth_proxy',
        'lexical_diversity'
    ]

    missing = [
        c for c in required_feat
        if c not in FEAT.columns
    ]

    if missing:
        raise ValueError(
            f'Missing Phase 10C feature columns: {missing}'
        )

    df = BENCH.merge(
        FEAT[
            [
                'topic_id',
                'video_id',
                'semantic_alignment',
                'depth_proxy',
                'lexical_diversity'
            ]
        ],
        on=['topic_id', 'video_id'],
        how='left',
        validate='one_to_one'
    )

    df = df.merge(
        POP[
            [
                'video_id',
                'views',
                'likes',
                'comment_count'
            ]
        ],
        on='video_id',
        how='left',
        validate='many_to_one'
    )

    # ------------------------------------------------------------------------
    # Feature construction
    # ------------------------------------------------------------------------

    df['content_length'] = (
        df['content']
        .fillna('')
        .astype(str)
        .str.split()
        .str.len()
    )

    numeric_cols = [
        'semantic_alignment',
        'depth_proxy',
        'lexical_diversity',
        'views',
        'likes',
        'comment_count',
        'content_length'
    ]

    for c in numeric_cols:
        df[c] = pd.to_numeric(
            df[c],
            errors='coerce'
        ).fillna(0.0)

    df['log_views'] = np.log1p(df['views'])
    df['log_likes'] = np.log1p(df['likes'])
    df['log_comments'] = np.log1p(
        df['comment_count']
    )

    feature_cols = [
        'semantic_alignment',
        'depth_proxy',
        'lexical_diversity',
        'content_length',
        'log_views',
        'log_likes',
        'log_comments'
    ]

    X = (
        df[feature_cols]
        .astype(float)
        .to_numpy()
    )

    y = (
        pd.to_numeric(
            df['relevance'],
            errors='coerce'
        )
        .fillna(0)
        .astype(float)
        .to_numpy()
    )

    groups = df['topic_id'].to_numpy()

    # ------------------------------------------------------------------------
    # Validation checks
    # ------------------------------------------------------------------------

    if np.isnan(X).any():
        raise ValueError(
            'RankNet feature matrix contains NaN.'
        )

    if len(np.unique(groups)) < 5:
        raise ValueError(
            'At least 5 unique topics are required for 5-fold GroupKFold.'
        )

    print('RankNet rows:', len(df))
    print('RankNet topics:', df['topic_id'].nunique())
    print('RankNet features:', feature_cols)

    # ------------------------------------------------------------------------
    # RankNet model
    # ------------------------------------------------------------------------

    class RankNet(nn.Module):

        def __init__(self, input_dim):
            super().__init__()

            self.net = nn.Sequential(
                nn.Linear(input_dim, 32),
                nn.ReLU(),

                nn.Linear(32, 16),
                nn.ReLU(),

                nn.Linear(16, 1)
            )

        def forward(self, x):
            return self.net(x).squeeze(-1)


    # ------------------------------------------------------------------------
    # Pairwise RankNet loss
    # ------------------------------------------------------------------------

    def ranknet_loss(score_pos, score_neg):

        return -torch.log(
            torch.sigmoid(
                score_pos - score_neg
            ) + 1e-8
        )


    # ------------------------------------------------------------------------
    # 5-fold GroupKFold by topic
    # ------------------------------------------------------------------------

    gkf = GroupKFold(
        n_splits=5
    )

    pred = np.full(
        len(df),
        np.nan,
        dtype=float
    )

    rng = np.random.default_rng(42)

    fold_topic_counts = []

    for fold, (tr, va) in enumerate(
        gkf.split(X, y, groups),
        start=1
    ):

        print()
        print('=' * 70)
        print(f'RankNet Fold {fold}/5')
        print('=' * 70)

        train_topics = df.iloc[tr]['topic_id'].to_numpy()
        valid_topics = df.iloc[va]['topic_id'].to_numpy()

        fold_topic_counts.append(
            len(np.unique(valid_topics))
        )

        # ------------------------------------------------------------
        # Standardization fitted only on training data
        # ------------------------------------------------------------

        scaler = StandardScaler()

        Xtr = scaler.fit_transform(X[tr])
        Xva = scaler.transform(X[va])

        # ------------------------------------------------------------
        # Model
        # ------------------------------------------------------------

        model = RankNet(
            input_dim=Xtr.shape[1]
        )

        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=1e-3,
            weight_decay=1e-4
        )

        Xt = torch.tensor(
            Xtr,
            dtype=torch.float32
        )

        # ------------------------------------------------------------
        # Build positive-negative pairs within training topics
        # ------------------------------------------------------------

        ytr = y[tr]

        local_groups = train_topics

        pairs = []

        for tid in np.unique(local_groups):

            local_idx = np.where(
                local_groups == tid
            )[0]

            positive = local_idx[
                ytr[local_idx] > 0.5
            ]

            negative = local_idx[
                ytr[local_idx] <= 0.5
            ]

            if len(positive) == 0 or len(negative) == 0:
                continue

            for p in positive:

                for n in negative:

                    pairs.append(
                        (
                            int(p),
                            int(n)
                        )
                    )

        # ------------------------------------------------------------
        # Limit pair count for computational stability
        # ------------------------------------------------------------

        if len(pairs) > 20000:

            selected = rng.choice(
                len(pairs),
                size=20000,
                replace=False
            )

            pairs = [
                pairs[i]
                for i in selected
            ]

        print(
            'Training pairs:',
            len(pairs)
        )

        if len(pairs) == 0:
            raise ValueError(
                f'No valid positive-negative pairs in fold {fold}.'
            )

        # ------------------------------------------------------------
        # Pairwise training
        # ------------------------------------------------------------

        model.train()

        for epoch in range(15):

            rng.shuffle(pairs)

            epoch_loss = 0.0

            for p, n in pairs:

                score_pos = model(
                    Xt[p:p + 1]
                )

                score_neg = model(
                    Xt[n:n + 1]
                )

                loss = ranknet_loss(
                    score_pos,
                    score_neg
                ).mean()

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

                epoch_loss += float(
                    loss.detach().cpu()
                )

            if (epoch + 1) % 5 == 0:

                avg_loss = (
                    epoch_loss / len(pairs)
                )

                print(
                    f'Epoch {epoch + 1:02d}/15 '
                    f'Loss={avg_loss:.6f}'
                )

        # ------------------------------------------------------------
        # Held-out topic predictions
        # ------------------------------------------------------------

        model.eval()

        Xva_tensor = torch.tensor(
            Xva,
            dtype=torch.float32
        )

        with torch.no_grad():

            valid_scores = (
                model(Xva_tensor)
                .cpu()
                .numpy()
            )

        pred[va] = valid_scores

        print(
            'Validation rows:',
            len(va)
        )

    # ------------------------------------------------------------------------
    # Validate complete out-of-fold predictions
    # ------------------------------------------------------------------------

    if np.isnan(pred).any():

        missing_count = int(
            np.isnan(pred).sum()
        )

        raise ValueError(
            f'RankNet has {missing_count} rows without '
            f'out-of-fold predictions.'
        )

    # ------------------------------------------------------------------------
    # Build final prediction table
    # ------------------------------------------------------------------------

    out = df[
        [
            'topic_id',
            'domain',
            'topic',
            'video_id',
            'relevance'
        ]
    ].copy()

    out['ranknet_score'] = pred

    # Rank using ACTUAL RankNet predictions
    out = out.sort_values(
        [
            'topic_id',
            'ranknet_score'
        ],
        ascending=[
            True,
            False
        ],
        kind='mergesort'
    ).reset_index(drop=True)

    out['rank'] = (
        out.groupby('topic_id')
        .cumcount()
        + 1
    )

    # ------------------------------------------------------------------------
    # Final validation
    # ------------------------------------------------------------------------

    if len(out) != len(BENCH):
        raise ValueError(
            'RankNet output row count does not match benchmark.'
        )

    if out['topic_id'].nunique() != BENCH['topic_id'].nunique():
        raise ValueError(
            'RankNet topic coverage does not match benchmark.'
        )

    if out['ranknet_score'].isna().any():
        raise ValueError(
            'RankNet output contains NaN scores.'
        )

    ranknet_path = (
        OUT /
        'phase12_ranknet_predictions.csv'
    )

    out.to_csv(
        ranknet_path,
        index=False
    )

    print()
    print('=' * 70)
    print('RankNet predictions saved:')
    print(ranknet_path)
    print('Rows:', len(out))
    print('Topics:', out['topic_id'].nunique())
    print('Fold validation topic counts:', fold_topic_counts)
    print('=' * 70)

else:

    print(
        'RankNet baseline is DISABLED. '
        'Set RUN_RANKNET=True to execute it.'
    )
````

````python
# ============================================================================
# PHASE 12E-C — OPTIONAL BASELINE METRICS
# ============================================================================
# Evaluates DPR and RankNet using the SAME topic-level ranking metrics used
# throughout the LAK-2026 benchmark analysis.
#
# Metrics:
#   NDCG@5
#   NDCG@10
#   MAP@10
#   MRR
#   Precision@5
#   Precision@10
#   Recall@5
#   Recall@10
#
# IMPORTANT:
# NDCG is calculated from the ACTUAL baseline scores.
# The previous implementation incorrectly supplied an artificial descending
# score vector to ndcg_score().
# ============================================================================

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.metrics import ndcg_score

OUT = Path('/content/revision_phase12')


# ----------------------------------------------------------------------------
# Average Precision @ 10
# ----------------------------------------------------------------------------

def ap10(y):

    y = np.asarray(y).astype(int)[:10]

    total_relevant = int(
        y.sum()
    )

    if total_relevant == 0:
        return 0.0

    hits = 0
    score = 0.0

    for i, value in enumerate(y, start=1):

        if value == 1:

            hits += 1

            score += (
                hits / i
            )

    return score / min(
        total_relevant,
        10
    )


# ----------------------------------------------------------------------------
# Reciprocal Rank
# ----------------------------------------------------------------------------

def rr(y):

    y = np.asarray(y).astype(int)

    for i, value in enumerate(
        y,
        start=1
    ):

        if value == 1:
            return 1.0 / i

    return 0.0


# ----------------------------------------------------------------------------
# Evaluate one prediction file
# ----------------------------------------------------------------------------

def evaluate(
    path,
    score_col,
    method
):

    d = pd.read_csv(path)

    required = [
        'topic_id',
        'video_id',
        'relevance',
        score_col
    ]

    missing = [
        c for c in required
        if c not in d.columns
    ]

    if missing:
        raise ValueError(
            f'{method}: missing columns {missing}'
        )

    rows = []

    for topic_id, g in d.groupby(
        'topic_id',
        sort=True
    ):

        # ------------------------------------------------------------
        # Sort according to the ACTUAL baseline score
        # ------------------------------------------------------------

        g = g.sort_values(
            score_col,
            ascending=False,
            kind='mergesort'
        ).reset_index(drop=True)

        y = (
            g['relevance']
            .astype(int)
            .to_numpy()
        )

        scores = (
            g[score_col]
            .astype(float)
            .to_numpy()
        )

        # Skip topics without any positive ground-truth item
        if y.sum() == 0:
            continue

        # ------------------------------------------------------------
        # IMPORTANT:
        # NDCG uses the ACTUAL prediction scores.
        #
        # This is the correction to the previous implementation.
        # ------------------------------------------------------------

        ndcg5 = ndcg_score(
            [y],
            [scores],
            k=5
        )

        ndcg10 = ndcg_score(
            [y],
            [scores],
            k=10
        )

        # ------------------------------------------------------------
        # Precision / Recall
        # ------------------------------------------------------------

        precision5 = (
            y[:5].mean()
        )

        precision10 = (
            y[:10].mean()
        )

        recall5 = (
            y[:5].sum()
            /
            y.sum()
        )

        recall10 = (
            y[:10].sum()
            /
            y.sum()
        )

        rows.append(
            {
                'topic_id': topic_id,
                'method': method,

                'NDCG@5': ndcg5,
                'NDCG@10': ndcg10,

                'MAP@10': ap10(y),
                'MRR': rr(y),

                'Precision@5': precision5,
                'Precision@10': precision10,

                'Recall@5': recall5,
                'Recall@10': recall10
            }
        )

    topic = pd.DataFrame(
        rows
    )

    if topic.empty:
        raise ValueError(
            f'{method}: no valid topic-level evaluation rows.'
        )

    metric_cols = [
        'NDCG@5',
        'NDCG@10',
        'MAP@10',
        'MRR',
        'Precision@5',
        'Precision@10',
        'Recall@5',
        'Recall@10'
    ]

    summary = (
        topic[metric_cols]
        .mean()
        .to_frame()
        .T
    )

    summary.insert(
        0,
        'method',
        method
    )

    summary.insert(
        1,
        'topics_evaluated',
        len(topic)
    )

    return topic, summary


# ----------------------------------------------------------------------------
# Evaluate available optional baselines
# ----------------------------------------------------------------------------

all_summary = []

baselines = [
    (
        'phase12_dpr_predictions.csv',
        'dpr_score',
        'DPR'
    ),
    (
        'phase12_ranknet_predictions.csv',
        'ranknet_score',
        'RankNet'
    )
]

for filename, score_col, method in baselines:

    path = OUT / filename

    if not path.exists():

        print(
            f'{method}: prediction file not found; '
            f'optional baseline was not run.'
        )

        continue

    topic, summary = evaluate(
        path=path,
        score_col=score_col,
        method=method
    )

    topic_path = (
        OUT /
        f'{method.lower()}_topic_metrics.csv'
    )

    topic.to_csv(
        topic_path,
        index=False
    )

    all_summary.append(
        summary
    )

    print()
    print('=' * 70)
    print(f'{method} SUMMARY')
    print('=' * 70)
    print(
        summary.to_string(
            index=False
        )
    )

    print()
    print(
        'Topic-level metrics saved:',
        topic_path
    )


# ----------------------------------------------------------------------------
# Combined summary
# ----------------------------------------------------------------------------

if all_summary:

    combined = pd.concat(
        all_summary,
        ignore_index=True
    )

    combined_path = (
        OUT /
        'phase12_optional_baseline_summary.csv'
    )

    combined.to_csv(
        combined_path,
        index=False
    )

    print()
    print('=' * 70)
    print('OPTIONAL BASELINE SUMMARY')
    print('=' * 70)
    print(
        combined.to_string(
            index=False
        )
    )

    print()
    print(
        'Saved:',
        combined_path
    )

else:

    print()
    print(
        'No optional DPR/RankNet prediction files were found.'
    )
````

## PHASE 12G — INDEPENDENT HUMAN-GROUND-TRUTH BENCHMARK VALIDATION

This phase operationalizes the reviewer-requested human evaluation using the independent LAK 2026 expert-annotated benchmark already included in the revision.

Important methodological boundary:
- LAK labels are human expert **Accepted/Rejected** relevance judgments.
- They are not a full human rank ordering.
- Therefore this phase evaluates whether ranking methods place human-accepted resources above rejected resources using NDCG, MAP, MRR, Precision@K and Recall@K.
- It does **not** claim exact agreement with a hypothetical human #1/#2/#3 ranking.

No labels are fabricated and no volunteer scores are inferred.

````python
# ============================================================================
# PHASE 12G — HUMAN-GROUND-TRUTH BENCHMARK SUMMARY
# ============================================================================
from pathlib import Path
import pandas as pd, numpy as np, json

OUT = Path('/content/revision_phase12')
BENCH = Path('/content/revision_phase10')
OUT.mkdir(parents=True, exist_ok=True)

# Locate the clean LAK export produced by Phase 10.
candidates = [
    BENCH/'lak2026_benchmark_eval.csv',
    BENCH/'phase10_lak2026_benchmark_eval.csv'
]
bench_path = next((p for p in candidates if p.exists()), None)
if bench_path is None:
    raise FileNotFoundError('Clean LAK benchmark export not found in /content/revision_phase10')

df = pd.read_csv(bench_path)

# Flexible column detection because earlier phase outputs used slightly different names.
label_col = next((c for c in ['label','relevance','human_label','accepted'] if c in df.columns), None)
topic_col = next((c for c in ['topic_id','topic','query'] if c in df.columns), None)
video_col = next((c for c in ['video_id','id'] if c in df.columns), None)
if label_col is None:
    raise ValueError(f'Could not identify human-label column. Columns: {list(df.columns)}')

def accepted(x):
    if isinstance(x, str):
        return 1 if x.strip().lower() in {'accepted','accept','1','true','yes'} else 0
    return int(bool(x))

df['_human_accepted'] = df[label_col].map(accepted)

summary = pd.DataFrame([{
    'benchmark':'LAK 2026',
    'rows':len(df),
    'topics':df[topic_col].nunique() if topic_col else np.nan,
    'unique_videos':df[video_col].nunique() if video_col else np.nan,
    'accepted':int(df['_human_accepted'].sum()),
    'rejected':int((1-df['_human_accepted']).sum()),
    'accepted_rate':float(df['_human_accepted'].mean())
}])
summary.to_csv(OUT/'phase12_human_ground_truth_summary.csv', index=False)

interpretation = pd.DataFrame([
    ['Human label semantics','Accepted/Rejected expert judgments','Human ground truth is binary relevance, not a complete rank ordering'],
    ['Evaluation target','Ranking methods','Measure whether accepted resources are ranked above rejected resources'],
    ['Exact human rank agreement','Not available from this benchmark','Cannot claim model #1 equals human #1 without explicit human ranking data'],
], columns=['item','result','methodological_note'])
interpretation.to_csv(OUT/'phase12_human_ground_truth_interpretation.csv', index=False)

print(summary.to_string(index=False))
print('\nMethodological interpretation:')
print(interpretation.to_string(index=False))
````

## PHASE 12H — DATASET STATISTICS + REPRODUCIBILITY MANIFEST

This phase consolidates the statistics and configuration needed for reproducibility and checks that the frozen 99-video cohort is used consistently.

The output is descriptive only; it does not invent missing metadata.

````python

# ============================================================================
# PHASE 12H — DATASET STATISTICS / REPRODUCIBILITY (FIXED)
# ============================================================================
from pathlib import Path
import pandas as pd, numpy as np, json, platform, os, sys, subprocess

OUT = Path('/content/revision_phase12')
ROOT = Path('/content/revision_phase9a')
OUT.mkdir(parents=True, exist_ok=True)

qpath = Path('/content/revision_phase9e/phase9e_iqs_full.csv')
fespath = Path('/content/revision_phase9d/phase9d_fes_full.csv')
hpath = Path('/content/revision_phase9f/phase9f_hybrid_full.csv')

iqs = pd.read_csv(qpath)
fes = pd.read_csv(fespath)
hyb = pd.read_csv(hpath)

frozen_ids = set(hyb['video_id'].astype(str))

# Load raw Phase-9A metadata. This fixes the previous NaN problem caused by
# trying to read raw YouTube fields from the Phase-9E IQS table.
video_frames = []
comment_frames = []

if ROOT.exists():
    for p in sorted(ROOT.glob('*/videos.csv')):
        try:
            d = pd.read_csv(p)
            if 'video_id' in d.columns:
                video_frames.append(d)
        except Exception as e:
            print(f'Warning: could not read {p}: {e}')

    for p in sorted(ROOT.glob('*/comments.csv')):
        try:
            d = pd.read_csv(p)
            if 'video_id' in d.columns:
                comment_frames.append(d)
        except Exception as e:
            print(f'Warning: could not read {p}: {e}')

videos_raw = pd.concat(video_frames, ignore_index=True) if video_frames else pd.DataFrame()
comments_raw = pd.concat(comment_frames, ignore_index=True) if comment_frames else pd.DataFrame()

if not videos_raw.empty:
    videos_raw['video_id'] = videos_raw['video_id'].astype(str)
    videos = videos_raw[videos_raw['video_id'].isin(frozen_ids)].copy()
else:
    videos = pd.DataFrame()

if not comments_raw.empty:
    comments_raw['video_id'] = comments_raw['video_id'].astype(str)
    comments = comments_raw[comments_raw['video_id'].isin(frozen_ids)].copy()
else:
    comments = pd.DataFrame()

base = (
    iqs[['video_id', 'domain', 'duration_minutes']]
    .drop_duplicates('video_id')
    .copy()
)
base['video_id'] = base['video_id'].astype(str)

if not videos.empty:
    keep = [
        c for c in [
            'video_id', 'views', 'likes', 'comment_count',
            'duration_minutes'
        ] if c in videos.columns
    ]
    meta = videos[keep].drop_duplicates('video_id')
    base = base.merge(meta, on='video_id', how='left', suffixes=('', '_raw'))

    if 'duration_minutes_raw' in base.columns:
        base['duration_minutes'] = base['duration_minutes'].fillna(
            base['duration_minutes_raw']
        )
        base.drop(columns=['duration_minutes_raw'], inplace=True)

# Relevant comments are the comments surviving the Phase-9C semantic
# filtering/sentiment stage, whereas comment_count is raw YouTube metadata.
if 'relevant_comment_count' in fes.columns:
    rel = (
        fes[['video_id', 'relevant_comment_count']]
        .drop_duplicates('video_id')
        .copy()
    )
    rel['video_id'] = rel['video_id'].astype(str)
    base = base.merge(rel, on='video_id', how='left')
else:
    base['relevant_comment_count'] = np.nan

rows = []
for domain, g in base.groupby('domain', dropna=False):
    rows.append({
        'domain': domain,
        'videos': g['video_id'].nunique(),
        'mean_duration_minutes': g['duration_minutes'].mean(),
        'median_duration_minutes': g['duration_minutes'].median(),
        'mean_comments': g['comment_count'].mean() if 'comment_count' in g else np.nan,
        'median_comments': g['comment_count'].median() if 'comment_count' in g else np.nan,
        'mean_views': g['views'].mean() if 'views' in g else np.nan,
        'mean_likes': g['likes'].mean() if 'likes' in g else np.nan,
        'total_raw_comments_metadata': g['comment_count'].sum() if 'comment_count' in g else np.nan,
        'total_relevant_comments': g['relevant_comment_count'].sum(),
    })

stats = pd.DataFrame(rows)
stats.to_csv(OUT / 'phase12_dataset_statistics_by_domain.csv', index=False)

overall = pd.DataFrame([{
    'videos': base['video_id'].nunique(),
    'domains': base['domain'].nunique(),
    'raw_comment_records_downloaded': int(len(comments)) if not comments.empty else np.nan,
    'total_raw_comments_metadata': base['comment_count'].sum() if 'comment_count' in base else np.nan,
    'total_relevant_comments': base['relevant_comment_count'].sum(),
    'mean_duration_minutes': base['duration_minutes'].mean(),
    'mean_comments': base['comment_count'].mean() if 'comment_count' in base else np.nan,
    'mean_views': base['views'].mean() if 'views' in base else np.nan,
    'mean_likes': base['likes'].mean() if 'likes' in base else np.nan,
}])
overall.to_csv(OUT / 'phase12_dataset_statistics_overall.csv', index=False)

# Capture reproducibility environment information without requiring extra packages.
try:
    import torch
    torch_version = torch.__version__
    cuda_version = torch.version.cuda
    gpu_name = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
except Exception:
    torch_version = None
    cuda_version = None
    gpu_name = None

try:
    import sentence_transformers
    sentence_transformers_version = sentence_transformers.__version__
except Exception:
    sentence_transformers_version = None

try:
    import transformers
    transformers_version = transformers.__version__
except Exception:
    transformers_version = None

try:
    import sklearn
    sklearn_version = sklearn.__version__
except Exception:
    sklearn_version = None

manifest = {
    'cohort_size': int(hyb['video_id'].nunique()),
    'domains': sorted(map(str, hyb['domain'].dropna().unique())),
    'embedding_model': 'intfloat/e5-base-v2',
    'sentiment_model': 'cardiffnlp/twitter-roberta-base-sentiment-latest',
    'teaching_score_model': 'gemini-3.5-flash-lite',
    'iqs_weights': {'semantic_alignment': 0.30, 'depth': 0.20, 'lexical_diversity': 0.20, 'teaching_score': 0.30},
    'fes_weights': {'emotional_engagement': 0.40, 'one_minus_emotional_variance': 0.20, 'likes_norm': 0.20, 'comment_count_norm': 0.20},
    'adaptive_k': 5,
    'lexical_diversity': 'UniqueWords / TotalWords (TTR)',
    'embedding_similarity': 'Dot product of L2-normalized embeddings (equivalent to cosine similarity)',
    'depth_interpretation': 'Duration is used as a structural depth proxy, not as a causal quality measure',
    'human_ground_truth': 'LAK 2026 independent expert Accepted/Rejected relevance labels',
    'python_version': platform.python_version(),
    'platform': platform.platform(),
    'hardware': {
        'gpu': gpu_name,
        'cuda_version': cuda_version,
        'cpu': platform.processor(),
    },
    'software_versions': {
        'python': platform.python_version(),
        'torch': torch_version,
        'sentence_transformers': sentence_transformers_version,
        'transformers': transformers_version,
        'scikit_learn': sklearn_version,
    },
    'ranknet_seed': 42,
}

with open(OUT / 'phase12_reproducibility_manifest.json', 'w', encoding='utf-8') as f:
    json.dump(manifest, f, indent=2)

print('Overall statistics:')
print(overall.to_string(index=False))
print('\nBy-domain statistics:')
print(stats.to_string(index=False))
print('\nFrozen cohort:', len(base), 'videos')
print('Raw Phase-9A video metadata available:', not videos.empty)
print('Raw Phase-9A comments available:', not comments.empty)
print('\nSaved fixed dataset statistics and reproducibility manifest.')
````

## PHASE 12I — EXPLANATION SPECIFICITY / EVIDENCE-DIVERSITY AUDIT

This audit strengthens the explanation evaluation without fabricating human usefulness judgments.

It checks whether explanations:
1. mention video-specific evidence;
2. contain numeric evidence consistent with the ranking record;
3. are not exact duplicates across different videos;
4. vary when the underlying dominant component/evidence varies.

Human usefulness remains a separate construct and is not inferred from these automatic checks.

````python

# ============================================================================
# PHASE 12I — EXPLANATION SPECIFICITY / EVIDENCE-DIVERSITY AUDIT (FIXED)
# ============================================================================
from pathlib import Path
import pandas as pd, numpy as np, re

OUT = Path('/content/revision_phase12')
source_path = OUT / 'phase12_explanations.csv'
audit_path = OUT / 'phase12_explanation_audit.csv'

if source_path.exists():
    src = pd.read_csv(source_path)
elif audit_path.exists():
    src = pd.read_csv(audit_path)
else:
    raise FileNotFoundError(f'Neither {source_path} nor {audit_path} exists.')

text_col = next(
    (c for c in [
        'explanation', 'generated_explanation',
        'explanation_text', 'text'
    ] if c in src.columns),
    None
)

if text_col is None:
    raise ValueError(
        'No explanation text column found. '
        f'Columns: {list(src.columns)}'
    )

a = src.copy()
a[text_col] = a[text_col].fillna('').astype(str)

if audit_path.exists() and source_path.exists():
    audit = pd.read_csv(audit_path)
    merge_cols = [
        c for c in [
            'video_id', 'explanation_present', 'word_count', 'length_ok',
            'dominant_factor_consistent', 'numeric_evidence_consistent',
            'content_specific_evidence_present', 'evidence_grounded',
            'human_usefulness_evaluation'
        ] if c in audit.columns
    ]
    if 'video_id' in merge_cols:
        a = a.merge(
            audit[merge_cols].drop_duplicates('video_id'),
            on='video_id',
            how='left',
            suffixes=('', '_audit')
        )

a['explanation_normalized'] = (
    a[text_col]
    .str.lower()
    .str.replace(r'\d+(?:\.\d+)?', '<NUM>', regex=True)
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

duplicate_rate = (
    float(a['explanation_normalized'].duplicated(keep=False).mean())
    if len(a) else np.nan
)
duplicate_count = (
    int(a['explanation_normalized'].duplicated(keep=False).sum())
    if len(a) else 0
)
unique_normalized = (
    int(a['explanation_normalized'].nunique())
    if len(a) else 0
)

specific_terms = [
    'semantic', 'alignment', 'teaching', 'depth',
    'lexical', 'diversity', 'duration', 'emotional',
    'engagement', 'relevant comments', 'comments',
    'iqs', 'fes', 'alpha'
]

a['specific_evidence_term_count'] = a[text_col].apply(
    lambda x: sum(term in str(x).lower() for term in specific_terms)
)
a['specific_evidence_present_recomputed'] = (
    a['specific_evidence_term_count'] >= 2
)

grounding_present = (
    (
        'numeric_evidence_consistent' in a.columns or
        'numerical_consistency' in a.columns
    )
    and 'dominant_factor_consistent' in a.columns
    and 'evidence_grounded' in a.columns
)

summary = pd.DataFrame([{
    'explanations_audited': len(a),
    'normalized_duplicate_rate': duplicate_rate,
    'normalized_duplicate_records': duplicate_count,
    'unique_normalized_explanations': unique_normalized,
    'specific_evidence_rate': (
        float(a['specific_evidence_present_recomputed'].mean())
        if len(a) else np.nan
    ),
    'mean_specific_evidence_terms': (
        float(a['specific_evidence_term_count'].mean())
        if len(a) else np.nan
    ),
    'automated_grounding_fields_present': grounding_present,
    'human_usefulness_evaluation':
        'DEFERRED — not inferred from automatic checks'
}])

summary.to_csv(
    OUT / 'phase12_explanation_specificity_summary.csv',
    index=False
)
a.to_csv(
    OUT / 'phase12_explanation_specificity_records.csv',
    index=False
)

print(summary.to_string(index=False))
print('\nSource used:', source_path if source_path.exists() else audit_path)
print('\nThis audit is automatic; it does not claim human usefulness.')
````

## PHASE 12J — FINAL EXPERIMENTAL COMPLETENESS CHECK (REVISED)

This validator checks the outputs required for the reviewer-facing experimental response.
It distinguishes completed objective validation from human judgments that are not available in the current data.

````python
# ============================================================================
# PHASE 12D-B — EMPIRICAL SCALABILITY ANALYSIS (FINAL)
# ============================================================================
from pathlib import Path
import time
import numpy as np
import pandas as pd

OUT = Path('/content/revision_phase12')
OUT.mkdir(parents=True, exist_ok=True)

HYB_PATH = Path('/content/revision_phase9f/phase9f_hybrid_full.csv')

if not HYB_PATH.exists():
    raise FileNotFoundError(f"Missing required frozen-cohort file: {HYB_PATH}")

hyb = pd.read_csv(HYB_PATH)

required_hyb = ['final_engagement_score', 'intrinsic_quality_score']
missing_hyb = [c for c in required_hyb if c not in hyb.columns]
if missing_hyb:
    raise KeyError(f"Missing Hybrid columns: {missing_hyb}")

# The frozen Phase-9F Hybrid table is the authoritative cohort.
frozen_n = len(hyb)

if frozen_n < 99:
    raise ValueError(
        f"Frozen Phase-9F cohort contains only {frozen_n} rows; "
        f"expected frozen cohort size is 99."
    )

cohort_sizes = [20, 40, 60, 80, 99]

records = []

for n in cohort_sizes:
    h = hyb.iloc[:n].copy()

    # Warm-up.
    _ = (
        0.4 * h['final_engagement_score'].astype(float)
        + 0.6 * h['intrinsic_quality_score'].astype(float)
    ).sort_values(ascending=False)

    timings = []

    for _repeat in range(5):
        t0 = time.perf_counter()

        _ = (
            0.4 * h['final_engagement_score'].astype(float)
            + 0.6 * h['intrinsic_quality_score'].astype(float)
        ).sort_values(ascending=False)

        timings.append(time.perf_counter() - t0)

    records.append({
        'n_videos': int(n),
        'repeats': 5,
        'mean_elapsed_seconds': float(np.mean(timings)),
        'std_elapsed_seconds': float(np.std(timings, ddof=1)),
        'min_elapsed_seconds': float(np.min(timings)),
        'max_elapsed_seconds': float(np.max(timings)),
        'operation': 'FES/IQS fusion + ranking sort',
        'theoretical_complexity': 'O(N log N)'
    })

empirical_df = pd.DataFrame(records)

empirical_path = OUT / 'phase12_empirical_scalability.csv'
empirical_df.to_csv(empirical_path, index=False)

x = empirical_df['n_videos'].to_numpy(dtype=float)
y = empirical_df['mean_elapsed_seconds'].to_numpy(dtype=float)

if len(x) >= 2 and np.ptp(x) > 0:
    slope, intercept = np.polyfit(x, y, 1)
    predicted = slope * x + intercept
    ss_res = float(np.sum((y - predicted) ** 2))
    ss_tot = float(np.sum((y - np.mean(y)) ** 2))
    r2 = float(1.0 - ss_res / ss_tot) if ss_tot > 0 else np.nan
else:
    slope = np.nan
    r2 = np.nan

summary_df = pd.DataFrame([{
    'n_min': int(x.min()),
    'n_max': int(x.max()),
    'n_points': int(len(x)),
    'time_at_min_n_seconds': float(y[x.argmin()]),
    'time_at_max_n_seconds': float(y[x.argmax()]),
    'descriptive_linear_slope_seconds_per_video': float(slope),
    'descriptive_linear_r2': float(r2),
    'theoretical_complexity': 'O(N log N)',
    'measurement_scope': 'Deterministic FES/IQS fusion and ranking sort only',
    'frozen_cohort_size': int(frozen_n),
    'note': (
        'Empirical scalability measures the deterministic ranking stage only; '
        'it does not measure end-to-end retrieval, E5, sentiment, or LLM/API scaling.'
    )
}])

summary_path = OUT / 'phase12_scalability_summary.csv'
summary_df.to_csv(summary_path, index=False)

actual_sizes = empirical_df['n_videos'].tolist()
if actual_sizes != cohort_sizes:
    raise RuntimeError(
        f"Scalability cohort mismatch: expected {cohort_sizes}, got {actual_sizes}"
    )

print("=" * 80)
print("PHASE 12D-B — EMPIRICAL SCALABILITY ANALYSIS")
print("=" * 80)
print(f"Frozen Phase-9F cohort size: {frozen_n}")
print(f"Scalability points: {cohort_sizes}")
print()
print(empirical_df.to_string(index=False))
print()
print(summary_df.to_string(index=False))
print()
print("Saved:", empirical_path)
print("Saved:", summary_path)
print("\nFINAL SCALABILITY CHECK: PASS")
````

### Important methodological rule
The LAK benchmark contains expert Accepted/Rejected labels and educational-resource content, but it does not contain the Phase-9 engagement/sentiment signals. Therefore this extension reconstructs the required YouTube-side signals for the 866 benchmark videos using the same model definitions: YouTube metadata, up to 100 top-level comments/video, E5 relevance, `cardiffnlp/twitter-roberta-base-sentiment-latest`, Gemini Teaching Score, FES, IQS, and adaptive alpha with `k=5`.

No FES, sentiment, or Teaching Score values are fabricated. The LAK labels remain the independent human-expert ground truth.


````python
# ============================================================================
# PHASE 13A — LAK YOUTUBE FEATURE ACQUISITION / COVERAGE
# ============================================================================
# Retrieves metadata + up to 100 top-level comments for every unique LAK video.
#
# IMPORTANT:
# Some historical LAK YouTube resources may currently be unavailable
# (deleted/private/removed). These are NOT fabricated or silently replaced.
# We create an accessible benchmark subset and evaluate all methods on the
# SAME subset.
#
# Comments-disabled videos are valid benchmark videos. They simply have
# zero available comments and are handled without fabrication.
# ============================================================================

!pip install -q google-api-python-client

from pathlib import Path
import re
import time
import pandas as pd

from googleapiclient.discovery import build
from google.colab import userdata


# ============================================================================
# PATHS
# ============================================================================

BASE = Path('/content/revision_phase10')
OUT = Path('/content/revision_phase13')

OUT.mkdir(
    parents=True,
    exist_ok=True
)

BENCH_PATH = BASE / 'lak2026_benchmark_eval.csv'

BENCH = pd.read_csv(
    BENCH_PATH
)

VIDEO_IDS = sorted(
    BENCH['video_id']
    .astype(str)
    .unique()
)

print('=' * 90)
print('PHASE 13A — LAK YOUTUBE FEATURE ACQUISITION')
print('=' * 90)

print(
    'Benchmark pairs:',
    len(BENCH)
)

print(
    'Benchmark topics:',
    BENCH['topic_id'].nunique()
)

print(
    'Unique videos:',
    len(VIDEO_IDS)
)


# ============================================================================
# API
# ============================================================================

YOUTUBE_API_KEY = userdata.get(
    'YOUTUBE_API_KEY'
)

if not YOUTUBE_API_KEY:
    raise RuntimeError(
        'YOUTUBE_API_KEY missing from Colab Secrets.'
    )

youtube = build(
    'youtube',
    'v3',
    developerKey=YOUTUBE_API_KEY
)


# ============================================================================
# OUTPUT FILES
# ============================================================================

META_PATH = (
    OUT /
    'lak_youtube_metadata.csv'
)

COMMENTS_PATH = (
    OUT /
    'lak_youtube_comments.csv'
)

UNAVAILABLE_PATH = (
    OUT /
    'lak_unavailable_video_ids.csv'
)


# ============================================================================
# HELPERS
# ============================================================================

def parse_duration(value):

    match = re.match(
        r'PT(?:(\d+)H)?(?:(\d+)M)?(?:(\d+)S)?',
        str(value or '')
    )

    if not match:
        return 0.0

    hours = int(
        match.group(1) or 0
    )

    minutes = int(
        match.group(2) or 0
    )

    seconds = int(
        match.group(3) or 0
    )

    return (
        hours * 60
        + minutes
        + seconds / 60.0
    )


def get_meta(video_ids):

    if not video_ids:
        return []

    try:

        response = (
            youtube
            .videos()
            .list(
                part='snippet,statistics,contentDetails',
                id=','.join(video_ids)
            )
            .execute()
        )

    except Exception as e:

        print(
            'Metadata API error:',
            type(e).__name__,
            str(e)[:200]
        )

        return []

    rows = []

    for item in response.get(
        'items',
        []
    ):

        snippet = item.get(
            'snippet',
            {}
        )

        statistics = item.get(
            'statistics',
            {}
        )

        content_details = item.get(
            'contentDetails',
            {}
        )

        rows.append(
            {
                'video_id':
                    item['id'],

                'video_title':
                    snippet.get(
                        'title',
                        ''
                    ),

                'description':
                    snippet.get(
                        'description',
                        ''
                    ),

                'views':
                    int(
                        statistics.get(
                            'viewCount',
                            0
                        ) or 0
                    ),

                'likes':
                    int(
                        statistics.get(
                            'likeCount',
                            0
                        ) or 0
                    ),

                'comment_count':
                    int(
                        statistics.get(
                            'commentCount',
                            0
                        ) or 0
                    ),

                'duration_minutes':
                    parse_duration(
                        content_details.get(
                            'duration',
                            ''
                        )
                    )
            }
        )

    return rows


def get_comments(video_id):

    rows = []

    try:

        response = (
            youtube
            .commentThreads()
            .list(
                part='snippet',
                videoId=video_id,
                maxResults=100,
                textFormat='plainText',
                order='relevance'
            )
            .execute()
        )

        for item in response.get(
            'items',
            []
        ):

            snippet = (
                item
                .get('snippet', {})
                .get('topLevelComment', {})
                .get('snippet', {})
            )

            top_comment = (
                item
                .get('snippet', {})
                .get('topLevelComment', {})
            )

            rows.append(
                {
                    'video_id':
                        video_id,

                    'comment_id':
                        top_comment.get(
                            'id',
                            ''
                        ),

                    'comment_text':
                        snippet.get(
                            'textDisplay',
                            ''
                        ),

                    'comment_like_count':
                        int(
                            snippet.get(
                                'likeCount',
                                0
                            ) or 0
                        )
                }
            )

    except Exception:
        # Includes commentsDisabled.
        # This is NOT treated as metadata failure.
        pass

    return rows


# ============================================================================
# LOAD EXISTING CACHE
# ============================================================================

if META_PATH.exists():

    meta = pd.read_csv(
        META_PATH
    )

else:

    meta = pd.DataFrame()


if COMMENTS_PATH.exists():

    comments = pd.read_csv(
        COMMENTS_PATH
    )

else:

    comments = pd.DataFrame()


known_meta = set(
    meta.get(
        'video_id',
        pd.Series(dtype=str)
    )
    .astype(str)
)


known_comment_videos = set(
    comments.get(
        'video_id',
        pd.Series(dtype=str)
    )
    .astype(str)
)


# ============================================================================
# METADATA ACQUISITION
# ============================================================================

pending = [
    video_id
    for video_id in VIDEO_IDS
    if video_id not in known_meta
]

print()
print(
    'Metadata already cached:',
    len(known_meta)
)

print(
    'Metadata pending:',
    len(pending)
)


# Up to 3 attempts for currently unavailable IDs.
MAX_METADATA_ATTEMPTS = 3

for attempt in range(
    1,
    MAX_METADATA_ATTEMPTS + 1
):

    if not pending:
        break

    print()
    print(
        f'Metadata attempt {attempt}/{MAX_METADATA_ATTEMPTS}'
    )

    newly_found = []

    for start in range(
        0,
        len(pending),
        50
    ):

        batch = pending[
            start:start + 50
        ]

        rows = get_meta(
            batch
        )

        if rows:

            meta = pd.concat(
                [
                    meta,
                    pd.DataFrame(rows)
                ],
                ignore_index=True
            )

            meta = (
                meta
                .drop_duplicates(
                    'video_id',
                    keep='last'
                )
            )

            meta.to_csv(
                META_PATH,
                index=False
            )

            newly_found.extend(
                [
                    str(row['video_id'])
                    for row in rows
                ]
            )

        print(
            f'  Metadata batch '
            f'{min(start + 50, len(pending))}'
            f'/{len(pending)}'
        )

        time.sleep(0.10)

    known_meta = set(
        meta['video_id']
        .astype(str)
    )

    pending = [
        video_id
        for video_id in VIDEO_IDS
        if video_id not in known_meta
    ]

    print(
        'Remaining unavailable:',
        len(pending)
    )

    if pending:
        time.sleep(1.0)


# ============================================================================
# SAVE CURRENTLY UNAVAILABLE VIDEO IDS
# ============================================================================

unavailable = pd.DataFrame(
    {
        'video_id':
            pending
    }
)

unavailable.to_csv(
    UNAVAILABLE_PATH,
    index=False
)


# ============================================================================
# COMMENT ACQUISITION
# ============================================================================

print()
print('=' * 90)
print('COMMENT ACQUISITION')
print('=' * 90)

new_comments = []

for i, video_id in enumerate(
    VIDEO_IDS,
    start=1
):

    if video_id in known_comment_videos:
        continue

    rows = get_comments(
        video_id
    )

    new_comments.extend(
        rows
    )

    if i % 25 == 0:

        if new_comments:

            comments = pd.concat(
                [
                    comments,
                    pd.DataFrame(new_comments)
                ],
                ignore_index=True
            )

            comments = (
                comments
                .drop_duplicates(
                    'comment_id',
                    keep='last'
                )
            )

            comments.to_csv(
                COMMENTS_PATH,
                index=False
            )

            new_comments = []

        print(
            f'Comments processed: '
            f'{i}/{len(VIDEO_IDS)}'
        )

    time.sleep(0.03)


if new_comments:

    comments = pd.concat(
        [
            comments,
            pd.DataFrame(new_comments)
        ],
        ignore_index=True
    )

    comments = (
        comments
        .drop_duplicates(
            'comment_id',
            keep='last'
        )
    )

    comments.to_csv(
        COMMENTS_PATH,
        index=False
    )


# ============================================================================
# RELOAD
# ============================================================================

meta = pd.read_csv(
    META_PATH
)

if COMMENTS_PATH.exists():

    comments = pd.read_csv(
        COMMENTS_PATH
    )

else:

    comments = pd.DataFrame(
        columns=[
            'video_id',
            'comment_id',
            'comment_text',
            'comment_like_count'
        ]
    )


metadata_video_ids = set(
    meta['video_id']
    .astype(str)
)


accessible_video_ids = sorted(
    metadata_video_ids
    &
    set(VIDEO_IDS)
)


# ============================================================================
# ACCESSIBLE BENCHMARK
# ============================================================================

BENCH_ACCESSIBLE = BENCH[
    BENCH['video_id']
    .astype(str)
    .isin(accessible_video_ids)
].copy()


ACCESSIBLE_BENCH_PATH = (
    OUT /
    'lak2026_benchmark_eval_accessible.csv'
)

BENCH_ACCESSIBLE.to_csv(
    ACCESSIBLE_BENCH_PATH,
    index=False
)


# ============================================================================
# COVERAGE REPORT
# ============================================================================

coverage = pd.DataFrame(
    [
        {
            'benchmark_pairs_original':
                len(BENCH),

            'benchmark_topics_original':
                BENCH['topic_id'].nunique(),

            'benchmark_unique_videos_original':
                len(VIDEO_IDS),

            'metadata_videos':
                len(accessible_video_ids),

            'metadata_missing_videos':
                len(VIDEO_IDS)
                - len(accessible_video_ids),

            'accessible_pairs':
                len(BENCH_ACCESSIBLE),

            'accessible_topics':
                BENCH_ACCESSIBLE['topic_id'].nunique(),

            'videos_with_comments':
                comments['video_id']
                .astype(str)
                .nunique()
                if len(comments)
                else 0,

            'comment_rows':
                len(comments),

            'metadata_coverage':
                len(accessible_video_ids)
                / len(VIDEO_IDS),

            'comment_video_coverage':
                (
                    comments['video_id']
                    .astype(str)
                    .nunique()
                    / len(VIDEO_IDS)
                )
                if len(VIDEO_IDS)
                else 0,

            'accessible_video_ids':
                len(accessible_video_ids)
        }
    ]
)

coverage.to_csv(
    OUT /
    'phase13_lak_hybrid_feature_coverage.csv',
    index=False
)

print()
print('=' * 90)
print('PHASE 13A — COVERAGE')
print('=' * 90)

print(
    coverage.to_string(
        index=False
    )
)

print()
print(
    'Unavailable video IDs saved:',
    UNAVAILABLE_PATH
)

print(
    'Accessible benchmark saved:',
    ACCESSIBLE_BENCH_PATH
)

print()
print(
    'IMPORTANT: No partial Hybrid is evaluated.'
)

print(
    f'Original videos : {len(VIDEO_IDS)}'
)

print(
    f'Accessible videos: {len(accessible_video_ids)}'
)

print(
    f'Original pairs  : {len(BENCH)}'
)

print(
    f'Accessible pairs: {len(BENCH_ACCESSIBLE)}'
)

if len(accessible_video_ids) == 0:
    raise RuntimeError(
        'No accessible LAK videos found.'
    )

if BENCH_ACCESSIBLE['topic_id'].nunique() == 0:
    raise RuntimeError(
        'No accessible LAK topics found.'
    )

print()
print(
    '✅ PHASE 13A COMPLETE — '
    'accessible benchmark subset prepared.'
)
````

````python
# ============================================================================
# PHASE 13B — FULL HYBRID FEATURE RECONSTRUCTION ON ACCESSIBLE LAK BENCHMARK
# ============================================================================
#
# Purpose:
#   Reconstruct the complete proposed Hybrid ranking for the accessible
#   LAK 2026 expert-annotated benchmark subset.
#
# Current accessible cohort:
#   - 888 topic-video judgments
#   - 54 topics
#   - 841 unique videos
#
# Components:
#   1. E5 semantic alignment
#   2. Comment semantic relevance
#   3. RoBERTa sentiment
#   4. Emotional engagement
#   5. FES
#   6. Teaching Score
#   7. IQS
#   8. Adaptive alpha
#   9. Full Hybrid score
#
# Ablation-ready outputs:
#   - fes_full
#   - fes_no_sentiment
#   - fes_no_semantic
#
# IMPORTANT:
#   The full component table is saved separately as:
#
#       phase13_lak_hybrid_components.csv
#
#   This file must NOT be overwritten by Phase 13C.
# ============================================================================


# ============================================================================
# IMPORTS
# ============================================================================

from pathlib import Path
import re
import json

import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer
from transformers import pipeline

from google import genai
from google.colab import userdata


# ============================================================================
# PATHS
# ============================================================================

OUT = Path('/content/revision_phase13')
BASE = Path('/content/revision_phase10')

OUT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================================
# LOAD ACCESSIBLE LAK BENCHMARK
# ============================================================================

ACCESSIBLE_BENCHMARK = (
    OUT /
    'lak2026_benchmark_eval_accessible.csv'
)

if not ACCESSIBLE_BENCHMARK.exists():

    raise FileNotFoundError(
        f'Accessible LAK benchmark not found:\n'
        f'{ACCESSIBLE_BENCHMARK}\n'
        f'Run Phase 13A first.'
    )


bench = pd.read_csv(
    ACCESSIBLE_BENCHMARK
)


print('=' * 100)
print('PHASE 13B — FULL HYBRID FEATURE RECONSTRUCTION')
print('=' * 100)

print()
print(
    f'Accessible pairs : {len(bench):,}'
)

print(
    f'Topics          : {bench["topic_id"].nunique():,}'
)

print(
    f'Unique videos    : {bench["video_id"].nunique():,}'
)


# ============================================================================
# LOAD YOUTUBE CACHE FILES
# ============================================================================

META_FILE = (
    OUT /
    'lak_youtube_metadata.csv'
)

COMMENTS_FILE = (
    OUT /
    'lak_youtube_comments.csv'
)

TEACHING_FILE = (
    OUT /
    'lak_teaching_scores.csv'
)


if not META_FILE.exists():

    raise FileNotFoundError(
        f'Metadata cache not found:\n{META_FILE}\n'
        f'Run Phase 13A first.'
    )


if not COMMENTS_FILE.exists():

    raise FileNotFoundError(
        f'Comments cache not found:\n{COMMENTS_FILE}\n'
        f'Run Phase 13A first.'
    )


meta = pd.read_csv(
    META_FILE
)

comments = pd.read_csv(
    COMMENTS_FILE
)


# ============================================================================
# NORMALIZE VIDEO IDS
# ============================================================================

bench['video_id'] = (
    bench['video_id']
    .astype(str)
)

meta['video_id'] = (
    meta['video_id']
    .astype(str)
)

comments['video_id'] = (
    comments['video_id']
    .astype(str)
)


# ============================================================================
# CHECK METADATA COLUMNS
# ============================================================================

required_meta = [

    'video_id',
    'video_title',
    'duration_minutes',
    'likes',
    'comment_count'
]


missing_meta = [

    c
    for c in required_meta
    if c not in meta.columns
]


if missing_meta:

    raise RuntimeError(
        'Required YouTube metadata is missing: '
        + str(missing_meta)
    )


# ============================================================================
# DESCRIPTION IS OPTIONAL
# ============================================================================
#
# Some YouTube videos legitimately have no description.
# We do NOT fabricate missing descriptions.
# ============================================================================

if 'description' not in meta.columns:

    meta['description'] = ''

else:

    meta['description'] = (
        meta['description']
        .fillna('')
        .astype(str)
    )


# ============================================================================
# MERGE ACCESSIBLE BENCHMARK WITH YOUTUBE METADATA
# ============================================================================

df = bench.merge(
    meta[
        [
            'video_id',
            'video_title',
            'description',
            'duration_minutes',
            'likes',
            'comment_count'
        ]
    ],
    on='video_id',
    how='left',
    validate='many_to_one'
)


# ============================================================================
# REQUIRED STRUCTURED METADATA VALIDATION
# ============================================================================

required_structured = [

    'video_title',
    'duration_minutes',
    'likes',
    'comment_count'
]


missing_structured = [

    c
    for c in required_structured
    if c not in df.columns
    or df[c].isna().any()
]


if missing_structured:

    raise RuntimeError(
        'Required YouTube metadata is incomplete: '
        + str(missing_structured)
    )


# ============================================================================
# OPTIONAL DESCRIPTION NORMALIZATION
# ============================================================================

df['description'] = (
    df['description']
    .fillna('')
    .astype(str)
)


description_missing = (
    df['description']
    .eq('')
    .sum()
)


print()
print(
    f'Videos with empty description: '
    f'{description_missing:,}'
)

print(
    f'Description coverage: '
    f'{1 - description_missing / len(df):.4f}'
)


# ============================================================================
# BUILD VIDEO CONTENT TEXT
# ============================================================================

df['content_text'] = (

    df['video_title']
    .fillna('')
    .astype(str)

    + ' '

    +

    df['description']
    .fillna('')
    .astype(str)
)


df['content_text'] = (
    df['content_text']
    .str.replace(
        r'\s+',
        ' ',
        regex=True
    )
    .str.strip()
)


# ============================================================================
# E5 MODEL
# ============================================================================

print()
print('Loading E5 model...')

e5 = SentenceTransformer(
    'intfloat/e5-base-v2'
)


# ============================================================================
# HELPERS
# ============================================================================

def minmax_array(values):

    values = np.asarray(
        values,
        dtype=float
    )

    lo = np.nanmin(values)
    hi = np.nanmax(values)

    if hi == lo:

        return np.zeros_like(
            values,
            dtype=float
        )

    return (
        values - lo
    ) / (
        hi - lo
    )


def topic_minmax(series):

    series = pd.to_numeric(
        series,
        errors='coerce'
    ).astype(float)

    if series.empty:

        return pd.Series(
            np.zeros(len(series)),
            index=series.index
        )

    lo = series.min()
    hi = series.max()

    if hi == lo:

        return pd.Series(
            np.zeros(len(series)),
            index=series.index
        )

    return (
        series - lo
    ) / (
        hi - lo
    )


def ttr(text):

    words = re.findall(
        r'\b[a-z0-9]+\b',
        str(text).lower()
    )

    if not words:

        return 0.0

    return (
        len(set(words))
        /
        len(words)
    )


# ============================================================================
# VIDEO/TOPIC STRUCTURE
# ============================================================================

topics = (
    bench[
        [
            'topic_id',
            'topic',
            'domain'
        ]
    ]
    .drop_duplicates()
    .sort_values('topic_id')
)


video_table = (
    df[
        [
            'video_id',
            'video_title',
            'description',
            'content_text'
        ]
    ]
    .drop_duplicates(
        subset='video_id'
    )
)


vids = sorted(
    video_table[
        'video_id'
    ]
    .astype(str)
    .unique()
)


vid_text = (
    video_table
    .set_index('video_id')
    ['content_text']
    .to_dict()
)


# ============================================================================
# VIDEO EMBEDDINGS
# ============================================================================

print()
print(
    f'Encoding {len(vids):,} accessible videos...'
)


video_embeddings = e5.encode(

    [
        'passage: ' +
        str(
            vid_text[v]
        )

        for v in vids
    ],

    normalize_embeddings=True,

    batch_size=32,

    show_progress_bar=True
)


video_index = {
    v: i
    for i, v in enumerate(vids)
}


# ============================================================================
# TOPIC QUERY EMBEDDINGS
# ============================================================================

topic_embeddings = e5.encode(

    [
        'query: ' +
        str(topic)
        for topic in topics['topic']
    ],

    normalize_embeddings=True,

    show_progress_bar=False
)


query_embedding_map = dict(
    zip(
        topics['topic_id'],
        topic_embeddings
    )
)


# ============================================================================
# VIDEO-LEVEL SEMANTIC ALIGNMENT
# ============================================================================
#
# Since embeddings are L2-normalized:
#
#   dot product == cosine similarity
# ============================================================================

df['semantic_alignment'] = [

    float(
        np.dot(
            video_embeddings[
                video_index[
                    str(video_id)
                ]
            ],
            query_embedding_map[
                topic_id
            ]
        )
    )

    for video_id, topic_id
    in zip(
        df['video_id'],
        df['topic_id']
    )
]


# ============================================================================
# STRUCTURAL CONTENT FEATURES
# ============================================================================

df['duration_minutes'] = (
    pd.to_numeric(
        df['duration_minutes'],
        errors='coerce'
    )
    .fillna(0.0)
)


# Duration is used only as a structural depth proxy.
df['depth_score'] = (
    df
    .groupby('topic_id')
    ['duration_minutes']
    .transform(topic_minmax)
)


# Lexical diversity = standard TTR:
#
#   unique words / total words
#
df['lexical_diversity_raw'] = (
    df['content_text']
    .apply(ttr)
)


df['lexical_diversity'] = (
    df
    .groupby('topic_id')
    ['lexical_diversity_raw']
    .transform(topic_minmax)
)


# Normalize semantic alignment within topic.
df['semantic_alignment_norm'] = (
    df
    .groupby('topic_id')
    ['semantic_alignment']
    .transform(topic_minmax)
)


# ============================================================================
# TEACHING SCORE
# ============================================================================

print()
print('=' * 100)
print('TEACHING SCORE')
print('=' * 100)


if TEACHING_FILE.exists():

    existing = pd.read_csv(
        TEACHING_FILE
    )

else:

    existing = pd.DataFrame(
        columns=[
            'video_id',
            'teaching_score',
            'llm_model',
            'prompt_version'
        ]
    )


existing['video_id'] = (
    existing['video_id']
    .astype(str)
)


done = set(
    existing['video_id']
)


GEMINI_API_KEY = userdata.get(
    'GEMINI_API_KEY'
)


if not GEMINI_API_KEY:

    raise RuntimeError(
        'GEMINI_API_KEY missing from Colab Secrets.'
    )


client = genai.Client(
    api_key=GEMINI_API_KEY
)


MODEL = 'gemini-3.5-flash-lite'


CRITERIA = (
    'Clear learning objective; '
    'structured explanation; '
    'logical concept progression; '
    'example usage; '
    'beginner friendliness; '
    'practical problem solving.'
)


for start in range(
    0,
    len(vids),
    6
):

    batch = [
        v
        for v in vids[start:start + 6]
        if v not in done
    ]


    if not batch:

        continue


    records = []


    for video_id in batch:

        row = (
            df[
                df.video_id.astype(str)
                ==
                video_id
            ]
            .iloc[0]
        )


        records.append({

            'video_id':
                video_id,

            'domain':
                str(row['domain']),

            'topic':
                str(row['topic']),

            'title':
                str(row['video_title']),

            'description':
                str(row['description'])[
                    :8000
                ]
        })


    prompt = (

        'You are an educational-content evaluator. '
        'Score Teaching Quality from 0 to 10 using these criteria: '

        + CRITERIA

        +

        '\nEvaluate ONLY the supplied title and description. '

        'Do not use views, likes, comments, subscriber count, '
        'or channel reputation. '

        'Return ONLY a JSON list of objects with '
        'video_id and teaching_score.\n\n'

        +

        json.dumps(
            records,
            ensure_ascii=False
        )
    )


    try:

        response = client.interactions.create(

            model=MODEL,

            input=prompt
        )


        raw_output = (
            response.output_text
            if hasattr(
                response,
                'output_text'
            )
            else str(response)
        )


        raw_output = re.sub(
            r'```json|```',
            '',
            raw_output
        ).strip()


        parsed = json.loads(
            raw_output
        )


        additions = []


        for item in parsed:

            try:

                video_id = str(
                    item['video_id']
                )

                score = float(
                    item['teaching_score']
                )


                if (
                    video_id in batch
                    and
                    0 <= score <= 10
                ):

                    additions.append({

                        'video_id':
                            video_id,

                        'teaching_score':
                            round(
                                score,
                                2
                            ),

                        'llm_model':
                            MODEL,

                        'prompt_version':
                            'Phase13_LAK_Hybrid_v1'
                    })

            except Exception:

                continue


        if additions:

            existing = pd.concat(
                [
                    existing,
                    pd.DataFrame(
                        additions
                    )
                ],
                ignore_index=True
            )


            existing = (
                existing
                .drop_duplicates(
                    'video_id',
                    keep='last'
                )
            )


            existing.to_csv(
                TEACHING_FILE,
                index=False
            )


            done = set(
                existing[
                    'video_id'
                ]
                .astype(str)
            )


    except Exception as e:

        print(
            'Teaching-score batch failed:',
            type(e).__name__,
            e
        )


# ============================================================================
# TEACHING SCORE COVERAGE
# ============================================================================

if not TEACHING_FILE.exists():

    raise RuntimeError(
        'Teaching-score file was not created.'
    )


teach = pd.read_csv(
    TEACHING_FILE
)


teach['video_id'] = (
    teach['video_id']
    .astype(str)
)


teach = (
    teach
    .drop_duplicates(
        'video_id',
        keep='last'
    )
)


missing_teaching = (
    set(vids)
    -
    set(
        teach['video_id']
    )
)


if missing_teaching:

    raise RuntimeError(

        'Teaching Score coverage incomplete. '
        'No partial Hybrid evaluation will be performed.\n'

        f'Missing videos: '
        f'{len(missing_teaching)}'
    )


df = df.merge(

    teach[
        [
            'video_id',
            'teaching_score'
        ]
    ],

    on='video_id',

    how='left',

    validate='many_to_one'
)


df['teaching_score_norm'] = (
    df['teaching_score']
    /
    10.0
)


# ============================================================================
# INTRINSIC QUALITY SCORE
# ============================================================================
#
# IQS =
#   0.30 Semantic Alignment
# + 0.20 Depth
# + 0.20 Lexical Diversity
# + 0.30 Teaching Score
# ============================================================================

df['intrinsic_quality_score'] = (

    0.30 *
    df['semantic_alignment_norm']

    +

    0.20 *
    df['depth_score']

    +

    0.20 *
    df['lexical_diversity']

    +

    0.30 *
    df['teaching_score_norm']
)


# ============================================================================
# COMMENT DATA
# ============================================================================

required_comment_columns = [
    'video_id',
    'comment_text'
]


missing_comment_columns = [

    c
    for c in required_comment_columns
    if c not in comments.columns
]


if missing_comment_columns:

    raise RuntimeError(
        'Comment file is missing: '
        + str(missing_comment_columns)
    )


comments = comments[
    comments[
        'comment_text'
    ]
    .fillna('')
    .astype(str)
    .str.strip()
    .ne('')
].copy()


comments = (
    comments
    .reset_index(drop=True)
)


print()
print(
    f'Usable comments: '
    f'{len(comments):,}'
)


if comments.empty:

    raise RuntimeError(
        'No usable comments available.'
    )


# ============================================================================
# SENTIMENT MODEL
# ============================================================================

print()
print('Loading RoBERTa sentiment model...')


import torch


sentiment_model = pipeline(

    'sentiment-analysis',

    model=
        'cardiffnlp/twitter-roberta-base-sentiment-latest',

    device=
        0
        if torch.cuda.is_available()
        else -1,

    top_k=None
)


# ============================================================================
# SENTIMENT SCORES
# ============================================================================
#
# sentiment_score = P(positive) - P(negative)
# ============================================================================

sentiment_scores = []


for start in range(
    0,
    len(comments),
    64
):

    batch = (
        comments[
            'comment_text'
        ]
        .iloc[
            start:start + 64
        ]
        .astype(str)
        .tolist()
    )


    outputs = sentiment_model(

        batch,

        truncation=True,

        max_length=256
    )


    for item in outputs:

        items = (
            item
            if isinstance(
                item,
                list
            )
            else [item]
        )


        probabilities = {

            str(
                z['label']
            ).lower():

            float(
                z['score']
            )

            for z in items
        }


        positive = next(

            (
                value
                for label, value
                in probabilities.items()
                if 'positive' in label
            ),

            0.0
        )


        negative = next(

            (
                value
                for label, value
                in probabilities.items()
                if 'negative' in label
            ),

            0.0
        )


        sentiment_scores.append(
            positive - negative
        )


comments['sentiment_score'] = (
    np.asarray(
        sentiment_scores,
        dtype=float
    )
)


# ============================================================================
# COMMENT E5 EMBEDDINGS
# ============================================================================

print()
print(
    f'Encoding {len(comments):,} comments for '
    'semantic relevance...'
)


comment_embeddings = e5.encode(

    [
        'passage: ' +
        str(x)

        for x in comments[
            'comment_text'
        ]
    ],

    normalize_embeddings=True,

    batch_size=64,

    show_progress_bar=True
)


# ============================================================================
# TOPIC-SPECIFIC COMMENT FES
# ============================================================================
#
# Three variants:
#
#   full:
#       semantic filtering + sentiment
#
#   no_sentiment:
#       same semantic filtering, but sentiment removed
#
#   no_semantic:
#       semantic filtering removed; first 1000 available comments retained
#
# ============================================================================

fes_parts = []


for _, topic_row in topics.iterrows():

    topic_id = topic_row['topic_id']


    query_vector = (
        query_embedding_map[
            topic_id
        ]
    )


    topic_video_ids = set(

        df.loc[
            df['topic_id']
            ==
            topic_id,

            'video_id'
        ]
        .astype(str)
    )


    topic_comments = comments[
        comments[
            'video_id'
        ]
        .astype(str)
        .isin(
            topic_video_ids
        )
    ].copy()


    if topic_comments.empty:

        continue


    # ------------------------------------------------------------------------
    # Calculate semantic relevance of comments to this topic.
    # ------------------------------------------------------------------------

    comment_positions = (
        topic_comments.index
        .to_numpy()
    )


    topic_comment_embeddings = (
        comment_embeddings[
            comment_positions
        ]
    )


    topic_comments[
        'comment_relevance'
    ] = (

        topic_comment_embeddings
        .dot(
            query_vector
        )
    )


    # ------------------------------------------------------------------------
    # FULL VARIANT
    # ------------------------------------------------------------------------

    full_comments = (
        topic_comments
        .sort_values(
            'comment_relevance',
            ascending=False
        )
        .head(1000)
        .copy()
    )


    # ------------------------------------------------------------------------
    # NO-SENTIMENT VARIANT
    # ------------------------------------------------------------------------
    #
    # Keep exactly the same semantically relevant comments,
    # but remove sentiment polarity.
    # ------------------------------------------------------------------------

    no_sentiment_comments = (
        full_comments
        .copy()
    )


    no_sentiment_comments[
        'sentiment_score'
    ] = 1.0


    # ------------------------------------------------------------------------
    # NO-SEMANTIC VARIANT
    # ------------------------------------------------------------------------
    #
    # Remove comment-level semantic filtering.
    # Retain up to 1000 collected comments.
    # ------------------------------------------------------------------------

    no_semantic_comments = (
        topic_comments
        .head(1000)
        .copy()
    )


    no_semantic_comments[
        'comment_relevance'
    ] = 1.0


    # ------------------------------------------------------------------------
    # COMPUTE EACH VARIANT
    # ------------------------------------------------------------------------

    variants = {

        'full':
            full_comments,

        'no_sentiment':
            no_sentiment_comments,

        'no_semantic':
            no_semantic_comments
    }


    for variant_name, variant_data in variants.items():

        x = variant_data.copy()


        x[
            'emotion_weighted_score'
        ] = (

            x[
                'comment_relevance'
            ]

            *

            x[
                'sentiment_score'
            ]
        )


        aggregated = (

            x
            .groupby(
                'video_id'
            )
            .agg(

                emotional_engagement=(
                    'emotion_weighted_score',
                    'mean'
                ),

                emotional_variance=(
                    'emotion_weighted_score',
                    'var'
                ),

                relevant_comment_count=(
                    'emotion_weighted_score',
                    'count'
                )
            )

            .reset_index()
        )


        aggregated[
            'topic_id'
        ] = topic_id


        aggregated[
            'variant'
        ] = variant_name


        fes_parts.append(
            aggregated
        )


# ============================================================================
# COMBINE FES COMPONENTS
# ============================================================================

if not fes_parts:

    raise RuntimeError(
        'No FES components were generated.'
    )


fes = pd.concat(
    fes_parts,
    ignore_index=True
)


fes[
    'emotional_variance'
] = (
    fes[
        'emotional_variance'
    ]
    .fillna(0.0)
)


# ============================================================================
# ADD ENGAGEMENT METADATA
# ============================================================================

engagement_meta = (
    meta[
        [
            'video_id',
            'likes',
            'comment_count'
        ]
    ]
    .drop_duplicates(
        'video_id'
    )
)


fes = fes.merge(

    engagement_meta,

    on='video_id',

    how='left',

    validate='many_to_one'
)


for column in [
    'likes',
    'comment_count'
]:

    fes[column] = (
        pd.to_numeric(
            fes[column],
            errors='coerce'
        )
        .fillna(0.0)
    )


# ============================================================================
# CREATE COMPLETE TOPIC-VIDEO KEY
# ============================================================================
#
# This is important:
# videos with zero accessible comments must still receive a valid FES.
# Their comment-based engagement terms become zero, while likes/comments
# remain available.
# ============================================================================

keys = (
    df[
        [
            'topic_id',
            'video_id'
        ]
    ]
    .drop_duplicates()
)


# ============================================================================
# FES CALCULATION
# ============================================================================
#
# FES =
#
# 0.40 EmotionalEngagement
# + 0.20 (1 - EmotionalVariance)
# + 0.20 LikesNorm
# + 0.20 CommentCountNorm
# ============================================================================

fes_outputs = []


for variant_name in [
    'full',
    'no_sentiment',
    'no_semantic'
]:

    variant = fes[
        fes['variant']
        ==
        variant_name
    ].copy()


    variant = keys.merge(

        variant[
            [
                'topic_id',
                'video_id',
                'emotional_engagement',
                'emotional_variance',
                'relevant_comment_count'
            ]
        ],

        on=[
            'topic_id',
            'video_id'
        ],

        how='left'
    )


    variant = variant.merge(

        engagement_meta,

        on='video_id',

        how='left',

        validate='many_to_one'
    )


    variant[
        'emotional_engagement'
    ] = (
        pd.to_numeric(
            variant[
                'emotional_engagement'
            ],
            errors='coerce'
        )
        .fillna(0.0)
    )


    variant[
        'emotional_variance'
    ] = (
        pd.to_numeric(
            variant[
                'emotional_variance'
            ],
            errors='coerce'
        )
        .fillna(0.0)
    )


    variant[
        'relevant_comment_count'
    ] = (
        pd.to_numeric(
            variant[
                'relevant_comment_count'
            ],
            errors='coerce'
        )
        .fillna(0)
        .astype(int)
    )


    variant['likes'] = (
        pd.to_numeric(
            variant['likes'],
            errors='coerce'
        )
        .fillna(0.0)
    )


    variant['comment_count'] = (
        pd.to_numeric(
            variant['comment_count'],
            errors='coerce'
        )
        .fillna(0.0)
    )


    # Topic-level normalization.
    variant[
        'emotional_engagement_norm'
    ] = (
        variant
        .groupby('topic_id')
        ['emotional_engagement']
        .transform(topic_minmax)
    )


    variant[
        'emotional_variance_norm'
    ] = (
        variant
        .groupby('topic_id')
        ['emotional_variance']
        .transform(topic_minmax)
    )


    variant[
        'likes_norm'
    ] = (
        variant
        .groupby('topic_id')
        ['likes']
        .transform(topic_minmax)
    )


    variant[
        'comment_count_norm'
    ] = (
        variant
        .groupby('topic_id')
        ['comment_count']
        .transform(topic_minmax)
    )


    variant[
        'final_engagement_score'
    ] = (

        0.40 *
        variant[
            'emotional_engagement_norm'
        ]

        +

        0.20 *
        (
            1.0
            -
            variant[
                'emotional_variance_norm'
            ]
        )

        +

        0.20 *
        variant[
            'likes_norm'
        ]

        +

        0.20 *
        variant[
            'comment_count_norm'
        ]
    )


    variant[
        'variant'
    ] = variant_name


    fes_outputs.append(
        variant
    )


fes_complete = pd.concat(
    fes_outputs,
    ignore_index=True
)


# ============================================================================
# EXTRACT FES VARIANTS
# ============================================================================

full_fes = (
    fes_complete[
        fes_complete['variant']
        ==
        'full'
    ][
        [
            'topic_id',
            'video_id',
            'final_engagement_score',
            'relevant_comment_count'
        ]
    ]
    .rename(
        columns={
            'final_engagement_score':
                'fes_full',

            'relevant_comment_count':
                'rc_full'
        }
    )
)


no_sent_fes = (
    fes_complete[
        fes_complete['variant']
        ==
        'no_sentiment'
    ][
        [
            'topic_id',
            'video_id',
            'final_engagement_score'
        ]
    ]
    .rename(
        columns={
            'final_engagement_score':
                'fes_no_sentiment'
        }
    )
)


no_sem_fes = (
    fes_complete[
        fes_complete['variant']
        ==
        'no_semantic'
    ][
        [
            'topic_id',
            'video_id',
            'final_engagement_score'
        ]
    ]
    .rename(
        columns={
            'final_engagement_score':
                'fes_no_semantic'
        }
    )
)


# ============================================================================
# COMBINE ALL FES VARIANTS
# ============================================================================

fes_components = (
    keys
    .merge(
        full_fes,
        on=[
            'topic_id',
            'video_id'
        ],
        how='left'
    )
    .merge(
        no_sent_fes,
        on=[
            'topic_id',
            'video_id'
        ],
        how='left'
    )
    .merge(
        no_sem_fes,
        on=[
            'topic_id',
            'video_id'
        ],
        how='left'
    )
)


for column in [
    'fes_full',
    'fes_no_sentiment',
    'fes_no_semantic'
]:

    fes_components[column] = (
        pd.to_numeric(
            fes_components[column],
            errors='coerce'
        )
        .fillna(0.0)
    )


fes_components['rc_full'] = (
    pd.to_numeric(
        fes_components['rc_full'],
        errors='coerce'
    )
    .fillna(0)
    .astype(int)
)


# ============================================================================
# BUILD FINAL HYBRID TABLE
# ============================================================================

full = (
    df
    .merge(
        fes_components,
        on=[
            'topic_id',
            'video_id'
        ],
        how='left',
        validate='one_to_one'
    )
)


# ============================================================================
# ADAPTIVE WEIGHTING
# ============================================================================
#
# Engagement Strength:
#
#   ES = log(1 + relevant_comment_count)
#
# Adaptive alpha:
#
#   alpha = ES / (ES + k)
#
# k = 5
# ============================================================================

full['engagement_strength'] = (
    np.log1p(
        full['rc_full']
        .astype(float)
    )
)


full['alpha'] = (

    full['engagement_strength']

    /

    (
        full['engagement_strength']
        +
        5.0
    )
)


# ============================================================================
# FULL HYBRID
# ============================================================================

full['hybrid_score'] = (

    full['alpha']
    *
    full['fes_full']

    +

    (
        1.0 -
        full['alpha']
    )
    *
    full['intrinsic_quality_score']
)


# ============================================================================
# CONTRIBUTION TERMS
# ============================================================================

full['fes_contribution'] = (

    full['alpha']
    *
    full['fes_full']
)


full['iqs_contribution'] = (

    (
        1.0 -
        full['alpha']
    )
    *
    full['intrinsic_quality_score']
)


# ============================================================================
# RANK WITHIN TOPIC
# ============================================================================

full['hybrid_rank'] = (

    full
    .groupby('topic_id')
    ['hybrid_score']
    .rank(
        ascending=False,
        method='first'
    )
    .astype(int)
)


# ============================================================================
# FINAL VALIDATION
# ============================================================================

required_output_columns = [

    'topic_id',
    'video_id',

    'alpha',

    'fes_full',
    'fes_no_sentiment',
    'fes_no_semantic',

    'intrinsic_quality_score',

    'depth_score',
    'lexical_diversity',
    'teaching_score_norm',

    'hybrid_score',

    'fes_contribution',
    'iqs_contribution',

    'hybrid_rank'
]


missing_output = [

    c
    for c in required_output_columns
    if c not in full.columns
]


if missing_output:

    raise RuntimeError(
        'Final Hybrid table is missing: '
        + str(missing_output)
    )


if len(full) != len(bench):

    raise RuntimeError(
        f'Hybrid coverage mismatch: '
        f'expected {len(bench):,} accessible pairs, '
        f'got {len(full):,}.'
    )


if full[
    [
        'topic_id',
        'video_id'
    ]
].duplicated().any():

    raise RuntimeError(
        'Duplicate topic-video pairs detected.'
    )


if not np.isfinite(
    full[
        [
            'alpha',
            'fes_full',
            'fes_no_sentiment',
            'fes_no_semantic',
            'intrinsic_quality_score',
            'hybrid_score'
        ]
    ]
    .to_numpy()
).all():

    raise RuntimeError(
        'NaN or infinite values found in Hybrid components.'
    )


if (
    (full['alpha'] < 0).any()
    or
    (full['alpha'] > 1).any()
):

    raise RuntimeError(
        'Alpha values outside [0,1].'
    )


# ============================================================================
# SAVE FULL COMPONENT FILE
# ============================================================================
#
# THIS FILE IS THE MASTER SOURCE FOR PHASE 13D.
# DO NOT OVERWRITE IT IN 13C.
# ============================================================================

COMPONENT_OUTPUT = (
    OUT /
    'phase13_lak_hybrid_components.csv'
)


full.to_csv(
    COMPONENT_OUTPUT,
    index=False
)


# ============================================================================
# SAVE COMPACT PREDICTION FILE FOR PHASE 13C
# ============================================================================

PREDICTION_OUTPUT = (
    OUT /
    'phase13_lak_hybrid_predictions.csv'
)


prediction_table = full[
    [
        'topic_id',
        'video_id',
        'domain',
        'topic',
        'relevance',
        'hybrid_score',
        'hybrid_rank'
    ]
].copy()


prediction_table.to_csv(
    PREDICTION_OUTPUT,
    index=False
)


# ============================================================================
# SAVE FEATURE COVERAGE
# ============================================================================

coverage = pd.DataFrame({

    'feature': [

        'topic_video_pairs',
        'topics',
        'unique_videos',
        'videos_with_teaching_score',
        'videos_with_comments',
        'videos_with_empty_description'
    ],

    'count': [

        len(full),

        full['topic_id'].nunique(),

        full['video_id'].nunique(),

        full['teaching_score']
        .notna()
        .sum(),

        full['video_id']
        .isin(
            comments['video_id']
        )
        .sum(),

        full['description']
        .eq('')
        .sum()
    ]
})


coverage.to_csv(
    OUT /
    'phase13_lak_hybrid_feature_coverage.csv',
    index=False
)


# ============================================================================
# OUTPUT SUMMARY
# ============================================================================

print()
print('=' * 100)
print('PHASE 13B — COMPLETED SUCCESSFULLY')
print('=' * 100)

print()
print(
    f'Accessible pairs : {len(full):,}'
)

print(
    f'Topics          : '
    f'{full["topic_id"].nunique():,}'
)

print(
    f'Unique videos    : '
    f'{full["video_id"].nunique():,}'
)

print()
print(
    'Full component file:'
)

print(
    COMPONENT_OUTPUT
)

print()
print(
    'Compact prediction file:'
)

print(
    PREDICTION_OUTPUT
)

print()
print(
    'Feature coverage file:'
)

print(
    OUT /
    'phase13_lak_hybrid_feature_coverage.csv'
)

print()
print(
    '✅ FULL COMPONENT TABLE PRESERVED FOR PHASE 13D'
)
````

````python
# ============================================================================
# PHASE 13C — FULL HYBRID EVALUATION ON ACCESSIBLE LAK HUMAN GROUND TRUTH
# ============================================================================
#
# Evaluates the FULL Hybrid ranking against independent LAK expert
# Accepted/Rejected judgments.
#
# Accessible evaluation cohort:
#   888 topic-video pairs
#   54 topics
#   841 unique videos
#
# Metrics:
#   NDCG@5
#   NDCG@10
#   MAP@10
#   MRR
#   Precision@5
#   Precision@10
#   Recall@5
#   Recall@10
#
# IMPORTANT:
#   LAK provides binary expert relevance judgments, not a complete human
#   ranking. Therefore this evaluates whether accepted resources are ranked
#   above rejected resources.
# ============================================================================


# ============================================================================
# IMPORTS
# ============================================================================

from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.metrics import ndcg_score


# ============================================================================
# PATHS
# ============================================================================

OUT = Path(
    '/content/revision_phase13'
)

PREDICTION_FILE = (
    OUT /
    'phase13_lak_hybrid_predictions.csv'
)

COMPONENT_FILE = (
    OUT /
    'phase13_lak_hybrid_components.csv'
)


# ============================================================================
# LOAD
# ============================================================================

if not PREDICTION_FILE.exists():

    raise FileNotFoundError(
        f'Prediction file not found:\n'
        f'{PREDICTION_FILE}\n'
        f'Run Phase 13B first.'
    )


if not COMPONENT_FILE.exists():

    raise FileNotFoundError(
        f'Component file not found:\n'
        f'{COMPONENT_FILE}\n'
        f'Run Phase 13B first.'
    )


d = pd.read_csv(
    PREDICTION_FILE
)

components = pd.read_csv(
    COMPONENT_FILE
)


print('=' * 100)
print('PHASE 13C — FULL HYBRID LAK EVALUATION')
print('=' * 100)

print()
print(
    f'Prediction rows : {len(d):,}'
)

print(
    f'Component rows  : {len(components):,}'
)


# ============================================================================
# CHECK COMPONENT FILE WAS NOT DESTROYED
# ============================================================================

required_component_columns = [

    'topic_id',
    'video_id',

    'alpha',
    'fes_full',
    'fes_no_sentiment',
    'fes_no_semantic',

    'intrinsic_quality_score',

    'depth_score',
    'lexical_diversity',
    'teaching_score_norm',

    'hybrid_score'
]


missing_components = [

    c
    for c in required_component_columns
    if c not in components.columns
]


if missing_components:

    raise RuntimeError(
        'The Phase 13B component file has been overwritten or '
        'is incomplete.\nMissing: '
        + str(missing_components)
    )


# ============================================================================
# CHECK PREDICTION COLUMNS
# ============================================================================

required_prediction_columns = [

    'topic_id',
    'video_id',
    'relevance',
    'hybrid_score'
]


missing_predictions = [

    c
    for c in required_prediction_columns
    if c not in d.columns
]


if missing_predictions:

    raise RuntimeError(
        'Prediction file is missing: '
        + str(missing_predictions)
    )


# ============================================================================
# NORMALIZE IDS
# ============================================================================

for frame in [
    d,
    components
]:

    frame['topic_id'] = (
        frame['topic_id']
        .astype(str)
    )

    frame['video_id'] = (
        frame['video_id']
        .astype(str)
    )


# ============================================================================
# RELEVANCE LABEL
# ============================================================================

def relevance_to_binary(value):

    value = (
        str(value)
        .strip()
        .lower()
    )

    if value in {
        'accepted',
        'accept',
        'relevant',
        '1',
        'true'
    }:

        return 1

    if value in {
        'rejected',
        'reject',
        'irrelevant',
        '0',
        'false'
    }:

        return 0

    raise ValueError(
        f'Unknown relevance label: {value}'
    )


d['relevance_binary'] = (
    d['relevance']
    .map(relevance_to_binary)
    .astype(int)
)


# ============================================================================
# CHECK COMPONENT AND PREDICTION COVERAGE
# ============================================================================

prediction_pairs = set(
    zip(
        d['topic_id'],
        d['video_id']
    )
)


component_pairs = set(
    zip(
        components['topic_id'],
        components['video_id']
    )
)


if prediction_pairs != component_pairs:

    raise RuntimeError(
        'Prediction and component topic-video cohorts do not match.'
    )


if len(d) == 0:

    raise RuntimeError(
        'No Hybrid predictions available.'
    )


# ============================================================================
# METRIC FUNCTIONS
# ============================================================================

def precision_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].mean()
    )


def recall_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    relevant = int(
        labels.sum()
    )

    if relevant == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].sum()
        /
        relevant
    )


def reciprocal_rank(
    labels
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    positions = np.where(
        labels == 1
    )[0]

    if len(positions) == 0:

        return 0.0

    return float(
        1.0
        /
        (
            positions[0]
            +
            1
        )
    )


def average_precision_at_k(
    labels,
    k=10
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    relevant_total = int(
        labels.sum()
    )

    if relevant_total == 0:

        return 0.0

    hits = 0
    score = 0.0

    for i in range(k):

        if labels[i] == 1:

            hits += 1

            score += (
                hits
                /
                (i + 1)
            )

    denominator = min(
        relevant_total,
        k
    )

    return float(
        score
        /
        denominator
    )


def ndcg_at_k(
    labels,
    scores,
    k
):

    labels = np.asarray(
        labels,
        dtype=float
    )

    scores = np.asarray(
        scores,
        dtype=float
    )

    if len(labels) == 0:

        return 0.0

    if labels.sum() == 0:

        return 0.0

    return float(
        ndcg_score(
            [labels],
            [scores],
            k=min(
                k,
                len(labels)
            )
        )
    )


# ============================================================================
# TOPIC-LEVEL EVALUATION
# ============================================================================

topic_rows = []


ranked_rows = []


for topic_id, group in d.groupby(
    'topic_id',
    sort=True
):

    ranked = (
        group
        .sort_values(
            by=[
                'hybrid_score',
                'video_id'
            ],
            ascending=[
                False,
                True
            ],
            kind='mergesort'
        )
        .reset_index(
            drop=True
        )
    )


    labels = (
        ranked[
            'relevance_binary'
        ]
        .astype(int)
        .to_numpy()
    )


    scores = (
        ranked[
            'hybrid_score'
        ]
        .astype(float)
        .to_numpy()
    )


    topic_rows.append({

        'topic_id':
            topic_id,

        'domain':
            ranked[
                'domain'
            ].iloc[0]
            if 'domain' in ranked.columns
            else '',

        'topic':
            ranked[
                'topic'
            ].iloc[0]
            if 'topic' in ranked.columns
            else '',

        'n_videos':
            len(ranked),

        'n_accepted':
            int(labels.sum()),

        'NDCG@5':
            ndcg_at_k(
                labels,
                scores,
                5
            ),

        'NDCG@10':
            ndcg_at_k(
                labels,
                scores,
                10
            ),

        'MAP@10':
            average_precision_at_k(
                labels,
                10
            ),

        'MRR':
            reciprocal_rank(
                labels
            ),

        'Precision@5':
            precision_at_k(
                labels,
                5
            ),

        'Precision@10':
            precision_at_k(
                labels,
                10
            ),

        'Recall@5':
            recall_at_k(
                labels,
                5
            ),

        'Recall@10':
            recall_at_k(
                labels,
                10
            )
    })


    ranked['hybrid_rank'] = (
        np.arange(
            len(ranked)
        )
        +
        1
    )


    ranked_rows.append(
        ranked
    )


# ============================================================================
# TOPIC METRICS
# ============================================================================

topic_metrics = pd.DataFrame(
    topic_rows
)


# ============================================================================
# MACRO SUMMARY
# ============================================================================

metric_columns = [

    'NDCG@5',
    'NDCG@10',
    'MAP@10',
    'MRR',
    'Precision@5',
    'Precision@10',
    'Recall@5',
    'Recall@10'
]


summary = pd.DataFrame({

    'method': [
        'Full_Hybrid'
    ],

    'topics_evaluated': [
        len(topic_metrics)
    ],

    'pairs': [
        len(d)
    ],

    'unique_videos': [
        d['video_id'].nunique()
    ],

    'accepted': [
        int(
            d['relevance_binary'].sum()
        )
    ],

    'rejected': [
        int(
            (
                d['relevance_binary']
                ==
                0
            ).sum()
        )
    ],

    **{
        metric: [
            topic_metrics[
                metric
            ].mean()
        ]

        for metric in metric_columns
    }
})


# ============================================================================
# SAVE
# ============================================================================

topic_metrics.to_csv(

    OUT /
    'phase13_lak_hybrid_topic_metrics.csv',

    index=False
)


summary.to_csv(

    OUT /
    'phase13_lak_hybrid_summary.csv',

    index=False
)


ranked_predictions = pd.concat(
    ranked_rows,
    ignore_index=True
)


ranked_predictions.to_csv(

    OUT /
    'phase13_lak_hybrid_ranked_predictions.csv',

    index=False
)


# ============================================================================
# PRINT RESULTS
# ============================================================================

print()
print('=' * 100)
print('FULL HYBRID — LAK EXPERT-GROUND-TRUTH RESULTS')
print('=' * 100)

print()

print(
    summary.to_string(
        index=False,
        float_format=lambda x: f'{x:.6f}'
    )
)


# ============================================================================
# VALIDATION
# ============================================================================

print()
print('=' * 100)
print('PHASE 13C VALIDATION')
print('=' * 100)


checks = {

    '888_or_actual_accessible_pairs':
        len(d) > 0,

    '54_or_fewer_topics':
        d['topic_id'].nunique()
        <=
        54,

    'binary_labels':
        set(
            d['relevance_binary']
            .unique()
        )
        .issubset({0, 1}),

    'all_metrics_finite':
        np.isfinite(
            topic_metrics[
                metric_columns
            ]
            .to_numpy()
        ).all(),

    'prediction_component_cohort_match':
        prediction_pairs
        ==
        component_pairs
}


for name, passed in checks.items():

    print(
        f'{"PASS" if passed else "FAIL"} — {name}'
    )


if not all(
    checks.values()
):

    raise RuntimeError(
        'Phase 13C validation failed.'
    )


print()
print(
    '✅ PHASE 13C COMPLETED SUCCESSFULLY'
)

print()
print(
    'Component file preserved at:'
)

print(
    OUT /
    'phase13_lak_hybrid_components.csv'
)
````

````python
# ============================================================================
# PHASE 13D — FULL HYBRID COMPONENT ABLATION
# ============================================================================
#
# Evaluates:
#
#   1. Full_Hybrid
#   2. Minus_Semantic
#   3. Minus_Sentiment
#   4. Minus_IQS
#   5. Fixed_50_50
#
# On the SAME accessible LAK expert-annotated benchmark cohort.
#
# Current cohort:
#   888 topic-video judgments
#   54 topics
#   841 unique videos
#
# IMPORTANT:
#   This cell reads phase13_lak_hybrid_components.csv.
#   It does NOT read or modify the compact Phase 13C prediction file.
# ============================================================================


# ============================================================================
# IMPORTS
# ============================================================================

from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.metrics import ndcg_score


# ============================================================================
# PATH
# ============================================================================

OUT = Path(
    '/content/revision_phase13'
)


COMPONENT_FILE = (
    OUT /
    'phase13_lak_hybrid_components.csv'
)


# ============================================================================
# LOAD MASTER COMPONENT TABLE
# ============================================================================

if not COMPONENT_FILE.exists():

    raise FileNotFoundError(

        f'Full Hybrid component file not found:\n'
        f'{COMPONENT_FILE}\n\n'

        f'Run Phase 13B first.'
    )


d = pd.read_csv(
    COMPONENT_FILE
)


print('=' * 100)
print('PHASE 13D — FULL HYBRID COMPONENT ABLATION')
print('=' * 100)

print()
print(
    f'Loaded rows : {len(d):,}'
)

print(
    f'Topics      : {d["topic_id"].nunique():,}'
)

print(
    f'Videos      : {d["video_id"].nunique():,}'
)


# ============================================================================
# REQUIRED COLUMNS
# ============================================================================

required_columns = [

    'topic_id',
    'video_id',
    'relevance',

    'alpha',

    'fes_full',
    'fes_no_sentiment',
    'fes_no_semantic',

    'intrinsic_quality_score',

    'depth_score',
    'lexical_diversity',
    'teaching_score_norm',

    'hybrid_score'
]


missing = [

    c
    for c in required_columns
    if c not in d.columns
]


if missing:

    print()
    print(
        'Available columns:'
    )

    print(
        d.columns.tolist()
    )

    raise ValueError(

        'Phase 13B component file is missing: '
        +
        str(missing)
    )


# ============================================================================
# NORMALIZE IDENTIFIERS
# ============================================================================

d['topic_id'] = (
    d['topic_id']
    .astype(str)
)

d['video_id'] = (
    d['video_id']
    .astype(str)
)


# ============================================================================
# LAK RELEVANCE
# ============================================================================

def relevance_to_binary(value):

    value = (
        str(value)
        .strip()
        .lower()
    )

    if value in {
        'accepted',
        'accept',
        'relevant',
        '1',
        'true'
    }:

        return 1

    if value in {
        'rejected',
        'reject',
        'irrelevant',
        '0',
        'false'
    }:

        return 0

    raise ValueError(
        f'Unknown relevance label: {value}'
    )


d['relevance_binary'] = (
    d['relevance']
    .map(relevance_to_binary)
    .astype(int)
)


# ============================================================================
# DUPLICATE CHECK
# ============================================================================

duplicates = (
    d
    .duplicated(
        subset=[
            'topic_id',
            'video_id'
        ]
    )
    .sum()
)


if duplicates:

    raise ValueError(
        f'Duplicate topic-video pairs: '
        f'{duplicates}'
    )


# ============================================================================
# NUMERIC VALIDATION
# ============================================================================

numeric_columns = [

    'alpha',

    'fes_full',
    'fes_no_sentiment',
    'fes_no_semantic',

    'intrinsic_quality_score',

    'depth_score',
    'lexical_diversity',
    'teaching_score_norm',

    'hybrid_score'
]


for column in numeric_columns:

    d[column] = pd.to_numeric(
        d[column],
        errors='coerce'
    )


if d[
    numeric_columns
].isna().any().any():

    bad = (
        d[
            numeric_columns
        ]
        .columns[
            d[
                numeric_columns
            ]
            .isna()
            .any()
        ]
        .tolist()
    )

    raise ValueError(
        f'NaN values detected in: {bad}'
    )


if not np.isfinite(
    d[
        numeric_columns
    ]
    .to_numpy()
).all():

    raise ValueError(
        'Non-finite numeric values detected.'
    )


# ============================================================================
# METRIC FUNCTIONS
# ============================================================================

def precision_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].mean()
    )


def recall_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    relevant = int(
        labels.sum()
    )

    if relevant == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].sum()
        /
        relevant
    )


def reciprocal_rank(
    labels
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    positions = np.where(
        labels == 1
    )[0]

    if len(positions) == 0:

        return 0.0

    return float(
        1.0
        /
        (
            positions[0]
            +
            1
        )
    )


def average_precision_at_k(
    labels,
    k=10
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    total_relevant = int(
        labels.sum()
    )

    if total_relevant == 0:

        return 0.0

    hits = 0
    score = 0.0


    for i in range(k):

        if labels[i] == 1:

            hits += 1

            score += (
                hits
                /
                (i + 1)
            )


    denominator = min(
        total_relevant,
        k
    )


    return float(
        score
        /
        denominator
    )


def ndcg_at_k(
    labels,
    scores,
    k
):

    labels = np.asarray(
        labels,
        dtype=float
    )

    scores = np.asarray(
        scores,
        dtype=float
    )

    if len(labels) == 0:

        return 0.0

    if labels.sum() == 0:

        return 0.0

    return float(
        ndcg_score(
            [labels],
            [scores],
            k=min(
                k,
                len(labels)
            )
        )
    )


# ============================================================================
# EVALUATION FUNCTION
# ============================================================================

def evaluate_variant(
    group,
    score_column
):

    ranked = (
        group
        .sort_values(
            by=[
                score_column,
                'video_id'
            ],
            ascending=[
                False,
                True
            ],
            kind='mergesort'
        )
        .reset_index(
            drop=True
        )
    )


    labels = (
        ranked[
            'relevance_binary'
        ]
        .astype(int)
        .to_numpy()
    )


    scores = (
        ranked[
            score_column
        ]
        .astype(float)
        .to_numpy()
    )


    return {

        'NDCG@5':
            ndcg_at_k(
                labels,
                scores,
                5
            ),

        'NDCG@10':
            ndcg_at_k(
                labels,
                scores,
                10
            ),

        'MAP@10':
            average_precision_at_k(
                labels,
                10
            ),

        'MRR':
            reciprocal_rank(
                labels
            ),

        'Precision@5':
            precision_at_k(
                labels,
                5
            ),

        'Precision@10':
            precision_at_k(
                labels,
                10
            ),

        'Recall@5':
            recall_at_k(
                labels,
                5
            ),

        'Recall@10':
            recall_at_k(
                labels,
                10
            )
    }


# ============================================================================
# 1. REMOVE SEMANTIC ALIGNMENT
# ============================================================================
#
# Original IQS:
#
#   0.30 Semantic Alignment
#   0.20 Depth
#   0.20 Lexical Diversity
#   0.30 Teaching Score
#
# Remove semantic alignment:
#
#   (0.20 Depth
#    + 0.20 Lexical Diversity
#    + 0.30 Teaching)
#   /
#   0.70
#
# FES also uses fes_no_semantic.
# ============================================================================

d['iqs_no_semantic'] = (

    (
        0.20 *
        d['depth_score']
    )

    +

    (
        0.20 *
        d['lexical_diversity']
    )

    +

    (
        0.30 *
        d['teaching_score_norm']
    )

) / 0.70


d['score_minus_semantic'] = (

    d['alpha']
    *
    d['fes_no_semantic']

    +

    (
        1.0 -
        d['alpha']
    )
    *
    d['iqs_no_semantic']
)


# ============================================================================
# 2. REMOVE SENTIMENT
# ============================================================================
#
# Same adaptive weighting and full IQS.
# Only sentiment is removed from FES.
# ============================================================================

d['score_minus_sentiment'] = (

    d['alpha']
    *
    d['fes_no_sentiment']

    +

    (
        1.0 -
        d['alpha']
    )
    *
    d['intrinsic_quality_score']
)


# ============================================================================
# 3. REMOVE IQS
# ============================================================================
#
# Engagement-only ranking.
# ============================================================================

d['score_minus_iqs'] = (
    d['fes_full']
)


# ============================================================================
# 4. REMOVE ADAPTIVE WEIGHTING
# ============================================================================
#
# Fixed 50/50 fusion:
#
#   0.5 FES + 0.5 IQS
# ============================================================================

d['score_fixed_50_50'] = (

    0.5 *
    d['fes_full']

    +

    0.5 *
    d['intrinsic_quality_score']
)


# ============================================================================
# VALIDATE ABLATION SCORES
# ============================================================================

ablation_columns = [

    'iqs_no_semantic',

    'score_minus_semantic',

    'score_minus_sentiment',

    'score_minus_iqs',

    'score_fixed_50_50'
]


if not np.isfinite(
    d[
        ablation_columns
    ]
    .to_numpy()
).all():

    raise ValueError(
        'Non-finite ablation scores detected.'
    )


# ============================================================================
# VARIANTS
# ============================================================================

variants = {

    'Full_Hybrid':
        'hybrid_score',

    'Minus_Semantic':
        'score_minus_semantic',

    'Minus_Sentiment':
        'score_minus_sentiment',

    'Minus_IQS':
        'score_minus_iqs',

    'Fixed_50_50':
        'score_fixed_50_50'
}


# ============================================================================
# TOPIC-LEVEL EVALUATION
# ============================================================================

metric_columns = [

    'NDCG@5',
    'NDCG@10',
    'MAP@10',
    'MRR',
    'Precision@5',
    'Precision@10',
    'Recall@5',
    'Recall@10'
]


rows = []


for method, score_column in variants.items():

    for topic_id, group in d.groupby(
        'topic_id',
        sort=True
    ):

        metrics = evaluate_variant(
            group,
            score_column
        )


        rows.append({

            'method':
                method,

            'topic_id':
                topic_id,

            'n_videos':
                len(group),

            'n_accepted':
                int(
                    group[
                        'relevance_binary'
                    ].sum()
                ),

            **metrics
        })


topic_results = pd.DataFrame(
    rows
)


# ============================================================================
# MACRO AVERAGE
# ============================================================================

summary = (
    topic_results
    .groupby(
        'method',
        sort=False
    )[metric_columns]
    .mean()
    .reset_index()
)


# ============================================================================
# METHOD ORDER
# ============================================================================

method_order = [

    'Full_Hybrid',

    'Minus_Semantic',

    'Minus_Sentiment',

    'Minus_IQS',

    'Fixed_50_50'
]


summary['method'] = pd.Categorical(

    summary['method'],

    categories=method_order,

    ordered=True
)


summary = (
    summary
    .sort_values(
        'method'
    )
    .reset_index(
        drop=True
    )
)


# ============================================================================
# DATASET INFORMATION
# ============================================================================

summary['topics_evaluated'] = (
    d['topic_id']
    .nunique()
)

summary['pairs'] = (
    len(d)
)

summary['unique_videos'] = (
    d['video_id']
    .nunique()
)


# ============================================================================
# SAVE TOPIC-LEVEL RESULTS
# ============================================================================

topic_output = (

    OUT /
    'phase13_lak_hybrid_component_ablation_topic_metrics.csv'
)


topic_results.to_csv(
    topic_output,
    index=False
)


# ============================================================================
# SAVE SUMMARY
# ============================================================================

summary_output = (

    OUT /
    'phase13_lak_hybrid_component_ablation.csv'
)


summary.to_csv(
    summary_output,
    index=False
)


# ============================================================================
# FULL HYBRID VS ABLATIONS
# ============================================================================

full_row = (
    summary[
        summary['method']
        ==
        'Full_Hybrid'
    ]
    .iloc[0]
)


comparison_rows = []


for _, row in summary.iterrows():

    if row['method'] == 'Full_Hybrid':

        continue


    result = {

        'comparison':
            f'Full_Hybrid_vs_{row["method"]}',

        'method':
            row['method']
    }


    for metric in metric_columns:

        full_value = float(
            full_row[metric]
        )

        ablation_value = float(
            row[metric]
        )


        result[
            metric +
            '_full_hybrid'
        ] = full_value


        result[
            metric +
            '_ablation'
        ] = ablation_value


        result[
            metric +
            '_change'
        ] = (
            ablation_value
            -
            full_value
        )


    comparison_rows.append(
        result
    )


comparison = pd.DataFrame(
    comparison_rows
)


comparison_output = (

    OUT /
    'phase13_lak_hybrid_component_ablation_comparison.csv'
)


comparison.to_csv(
    comparison_output,
    index=False
)


# ============================================================================
# PRINT MAIN RESULTS
# ============================================================================

print()
print('=' * 110)
print('PHASE 13D — FULL HYBRID COMPONENT ABLATION RESULTS')
print('=' * 110)

print()

print(
    summary[
        [
            'method',

            'NDCG@5',
            'NDCG@10',

            'MAP@10',
            'MRR',

            'Precision@5',
            'Precision@10',

            'Recall@5',
            'Recall@10'
        ]
    ]
    .to_string(
        index=False,
        float_format=lambda x: f'{x:.6f}'
    )
)


# ============================================================================
# PRINT DIFFERENCES
# ============================================================================

print()
print('=' * 110)
print('CHANGE RELATIVE TO FULL HYBRID')
print('=' * 110)


for _, row in comparison.iterrows():

    print()
    print(
        f'--- {row["method"]} ---'
    )


    for metric in metric_columns:

        change = row[
            metric +
            '_change'
        ]


        print(
            f'{metric:<18}: '
            f'{change:+.6f}'
        )


# ============================================================================
# VALIDATION
# ============================================================================

print()
print('=' * 110)
print('PHASE 13D VALIDATION')
print('=' * 110)


checks = {

    'component_rows_present':
        len(d) > 0,

    'topics_present':
        d['topic_id'].nunique() > 0,

    'unique_topic_video_pairs':
        not d.duplicated(
            [
                'topic_id',
                'video_id'
            ]
        ).any(),

    'five_variants':
        set(
            summary['method'].astype(str)
        )
        ==
        set(
            method_order
        ),

    'all_metrics_finite':
        np.isfinite(
            topic_results[
                metric_columns
            ]
            .to_numpy()
        ).all(),

    'all_ablation_scores_finite':
        np.isfinite(
            d[
                ablation_columns
            ]
            .to_numpy()
        ).all(),

    'binary_ground_truth':
        set(
            d[
                'relevance_binary'
            ]
            .unique()
        )
        .issubset(
            {0, 1}
        )
}


for name, passed in checks.items():

    print(
        f'{"PASS" if passed else "FAIL"} — {name}'
    )


if not all(
    checks.values()
):

    raise RuntimeError(
        'Phase 13D validation failed.'
    )


# ============================================================================
# FINAL
# ============================================================================

print()
print('=' * 110)
print('✅ PHASE 13D COMPLETED SUCCESSFULLY')
print('=' * 110)

print()
print(
    f'Evaluation cohort: '
    f'{len(d):,} topic-video judgments'
)

print(
    f'Topics: '
    f'{d["topic_id"].nunique():,}'
)

print(
    f'Unique videos: '
    f'{d["video_id"].nunique():,}'
)

print()
print('Saved:')

print(
    topic_output
)

print(
    summary_output
)

print(
    comparison_output
)
````

````python
# ============================================================================
# PHASE 13E — FULL HYBRID VS DPR / RANKNET / E5 / BM25
#                WILCOXON + 95% BOOTSTRAP CI
# ============================================================================
#
# IMPORTANT:
#
# All methods are evaluated on the SAME accessible LAK cohort:
#
#   888 topic-video judgments
#   54 topics
#   841 unique videos
#
# This avoids comparing the 888-pair Hybrid evaluation against baseline
# metrics computed previously on the full 914-pair benchmark.
#
# Methods:
#   Full Hybrid
#   DPR
#   RankNet
#   E5
#   BM25
#
# Statistical tests:
#   - Paired Wilcoxon signed-rank test
#   - 5,000 bootstrap resamples
#   - 95% percentile confidence interval
#
# ============================================================================


# ============================================================================
# IMPORTS
# ============================================================================

from pathlib import Path

import numpy as np
import pandas as pd

from scipy.stats import wilcoxon
from sklearn.metrics import ndcg_score

from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer


# ============================================================================
# PATHS
# ============================================================================

OUT = Path(
    '/content/revision_phase13'
)

PHASE12 = Path(
    '/content/revision_phase12'
)

PHASE10 = Path(
    '/content/revision_phase10'
)


# ============================================================================
# INPUT FILES
# ============================================================================

HYBRID_TOPIC_FILE = (
    OUT /
    'phase13_lak_hybrid_topic_metrics.csv'
)

HYBRID_PREDICTION_FILE = (
    OUT /
    'phase13_lak_hybrid_predictions.csv'
)

ACCESSIBLE_FILE = (
    OUT /
    'lak2026_benchmark_eval_accessible.csv'
)

DPR_PREDICTION_FILE = (
    PHASE12 /
    'phase12_dpr_predictions.csv'
)

RANKNET_PREDICTION_FILE = (
    PHASE12 /
    'phase12_ranknet_predictions.csv'
)


# ============================================================================
# VALIDATE REQUIRED INPUTS
# ============================================================================

required_files = {

    'Hybrid topic metrics':
        HYBRID_TOPIC_FILE,

    'Hybrid predictions':
        HYBRID_PREDICTION_FILE,

    'Accessible LAK benchmark':
        ACCESSIBLE_FILE,

    'DPR predictions':
        DPR_PREDICTION_FILE,

    'RankNet predictions':
        RANKNET_PREDICTION_FILE
}


missing_files = [

    name
    for name, path in required_files.items()
    if not path.exists()
]


if missing_files:

    raise FileNotFoundError(
        'Missing required Phase 13E files:\n'
        +
        '\n'.join(
            f' - {x}'
            for x in missing_files
        )
    )


# ============================================================================
# LOAD ACCESSIBLE BENCHMARK
# ============================================================================

benchmark = pd.read_csv(
    ACCESSIBLE_FILE
)


benchmark['topic_id'] = (
    benchmark['topic_id']
    .astype(str)
)

benchmark['video_id'] = (
    benchmark['video_id']
    .astype(str)
)


# ============================================================================
# ACCESSIBLE COHORT
# ============================================================================

accessible_pairs = set(
    zip(
        benchmark['topic_id'],
        benchmark['video_id']
    )
)

accessible_videos = set(
    benchmark['video_id']
)


print('=' * 110)
print('PHASE 13E — COMMON ACCESSIBLE LAK COHORT')
print('=' * 110)

print()
print(
    f'Accessible pairs : '
    f'{len(accessible_pairs):,}'
)

print(
    f'Topics          : '
    f'{benchmark["topic_id"].nunique():,}'
)

print(
    f'Unique videos    : '
    f'{benchmark["video_id"].nunique():,}'
)


# ============================================================================
# RELEVANCE CONVERSION
# ============================================================================

def relevance_to_binary(value):

    value = (
        str(value)
        .strip()
        .lower()
    )

    if value in {
        'accepted',
        'accept',
        'relevant',
        '1',
        'true'
    }:

        return 1

    if value in {
        'rejected',
        'reject',
        'irrelevant',
        '0',
        'false'
    }:

        return 0

    raise ValueError(
        f'Unknown relevance value: {value}'
    )


benchmark['relevance_binary'] = (
    benchmark['relevance']
    .map(relevance_to_binary)
    .astype(int)
)


# ============================================================================
# METRIC FUNCTIONS
# ============================================================================

def precision_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].mean()
    )


def recall_at_k(
    labels,
    k
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    total_relevant = int(
        labels.sum()
    )

    if total_relevant == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    return float(
        labels[:k].sum()
        /
        total_relevant
    )


def reciprocal_rank(
    labels
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    positions = np.where(
        labels == 1
    )[0]

    if len(positions) == 0:

        return 0.0

    return float(
        1.0
        /
        (
            positions[0]
            +
            1
        )
    )


def average_precision_at_k(
    labels,
    k=10
):

    labels = np.asarray(
        labels,
        dtype=int
    )

    if len(labels) == 0:

        return 0.0

    k = min(
        k,
        len(labels)
    )

    total_relevant = int(
        labels.sum()
    )

    if total_relevant == 0:

        return 0.0

    hits = 0
    score = 0.0

    for i in range(k):

        if labels[i] == 1:

            hits += 1

            score += (
                hits
                /
                (i + 1)
            )

    denominator = min(
        total_relevant,
        k
    )

    return float(
        score
        /
        denominator
    )


def ndcg_at_k(
    labels,
    scores,
    k
):

    labels = np.asarray(
        labels,
        dtype=float
    )

    scores = np.asarray(
        scores,
        dtype=float
    )

    if len(labels) == 0:

        return 0.0

    if labels.sum() == 0:

        return 0.0

    return float(
        ndcg_score(
            [labels],
            [scores],
            k=min(
                k,
                len(labels)
            )
        )
    )


# ============================================================================
# GENERIC TOPIC METRIC EVALUATOR
# ============================================================================

METRICS = [

    'NDCG@5',
    'NDCG@10',
    'MAP@10',
    'MRR',
    'Precision@5',
    'Precision@10',
    'Recall@5',
    'Recall@10'
]


def evaluate_prediction_dataframe(
    data,
    score_column
):

    rows = []

    for topic_id, group in data.groupby(
        'topic_id',
        sort=True
    ):

        group = group.copy()


        # Deterministic tie-breaking.
        ranked = (
            group
            .sort_values(
                by=[
                    score_column,
                    'video_id'
                ],
                ascending=[
                    False,
                    True
                ],
                kind='mergesort'
            )
            .reset_index(
                drop=True
            )
        )


        labels = (
            ranked[
                'relevance_binary'
            ]
            .astype(int)
            .to_numpy()
        )


        scores = (
            ranked[
                score_column
            ]
            .astype(float)
            .to_numpy()
        )


        rows.append({

            'topic_id':
                topic_id,

            'NDCG@5':
                ndcg_at_k(
                    labels,
                    scores,
                    5
                ),

            'NDCG@10':
                ndcg_at_k(
                    labels,
                    scores,
                    10
                ),

            'MAP@10':
                average_precision_at_k(
                    labels,
                    10
                ),

            'MRR':
                reciprocal_rank(
                    labels
                ),

            'Precision@5':
                precision_at_k(
                    labels,
                    5
                ),

            'Precision@10':
                precision_at_k(
                    labels,
                    10
                ),

            'Recall@5':
                recall_at_k(
                    labels,
                    5
                ),

            'Recall@10':
                recall_at_k(
                    labels,
                    10
                )
        })


    return pd.DataFrame(
        rows
    )


# ============================================================================
# LOAD HYBRID TOPIC METRICS
# ============================================================================

hybrid_topic = pd.read_csv(
    HYBRID_TOPIC_FILE
)

hybrid_topic['topic_id'] = (
    hybrid_topic['topic_id']
    .astype(str)
)


# ============================================================================
# VERIFY HYBRID HAS 54 TOPICS
# ============================================================================

hybrid_topics = set(
    hybrid_topic['topic_id']
)


benchmark_topics = set(
    benchmark['topic_id']
)


if hybrid_topics != benchmark_topics:

    missing_hybrid = (
        benchmark_topics
        -
        hybrid_topics
    )

    extra_hybrid = (
        hybrid_topics
        -
        benchmark_topics
    )

    raise RuntimeError(

        'Hybrid and accessible benchmark topic sets differ.\n'

        f'Missing Hybrid topics: '
        f'{sorted(missing_hybrid)}\n'

        f'Extra Hybrid topics: '
        f'{sorted(extra_hybrid)}'
    )


# ============================================================================
# LOAD DPR
# ============================================================================

dpr = pd.read_csv(
    DPR_PREDICTION_FILE
)


dpr.columns = [
    str(c).strip()
    for c in dpr.columns
]


print()
print('DPR columns:')
print(
    dpr.columns.tolist()
)


# ============================================================================
# FIND DPR SCORE COLUMN
# ============================================================================

possible_dpr_scores = [

    'dpr_score',
    'score',
    'similarity',
    'dpr_similarity',
    'prediction',
    'ranking_score'
]


dpr_score_column = next(

    (
        c
        for c in possible_dpr_scores
        if c in dpr.columns
    ),

    None
)


if dpr_score_column is None:

    raise ValueError(

        'Could not identify DPR score column.\n'
        f'Available columns: '
        f'{dpr.columns.tolist()}'
    )


required_dpr_ids = [
    'topic_id',
    'video_id'
]


missing_dpr_ids = [

    c
    for c in required_dpr_ids
    if c not in dpr.columns
]


if missing_dpr_ids:

    raise ValueError(
        'DPR predictions missing: '
        + str(missing_dpr_ids)
    )


dpr['topic_id'] = (
    dpr['topic_id']
    .astype(str)
)

dpr['video_id'] = (
    dpr['video_id']
    .astype(str)
)


dpr = dpr[
    dpr['video_id']
    .isin(
        accessible_videos
    )
].copy()


dpr = dpr.merge(

    benchmark[
        [
            'topic_id',
            'video_id',
            'relevance_binary'
        ]
    ],

    on=[
        'topic_id',
        'video_id'
    ],

    how='inner',

    validate='one_to_one'
)


dpr[dpr_score_column] = pd.to_numeric(
    dpr[dpr_score_column],
    errors='coerce'
)


if dpr[dpr_score_column].isna().any():

    raise ValueError(
        'DPR contains missing/non-numeric scores.'
    )


# ============================================================================
# LOAD RANKNET
# ============================================================================

ranknet = pd.read_csv(
    RANKNET_PREDICTION_FILE
)


ranknet.columns = [
    str(c).strip()
    for c in ranknet.columns
]


print()
print('RankNet columns:')
print(
    ranknet.columns.tolist()
)


# ============================================================================
# FIND RANKNET SCORE COLUMN
# ============================================================================

possible_ranknet_scores = [

    'ranknet_score',
    'score',
    'prediction',
    'ranking_score',
    'predicted_score'
]


ranknet_score_column = next(

    (
        c
        for c in possible_ranknet_scores
        if c in ranknet.columns
    ),

    None
)


if ranknet_score_column is None:

    raise ValueError(

        'Could not identify RankNet score column.\n'
        f'Available columns: '
        f'{ranknet.columns.tolist()}'
    )


missing_ranknet_ids = [

    c
    for c in required_dpr_ids
    if c not in ranknet.columns
]


if missing_ranknet_ids:

    raise ValueError(
        'RankNet predictions missing: '
        + str(missing_ranknet_ids)
    )


ranknet['topic_id'] = (
    ranknet['topic_id']
    .astype(str)
)

ranknet['video_id'] = (
    ranknet['video_id']
    .astype(str)
)


ranknet = ranknet[
    ranknet['video_id']
    .isin(
        accessible_videos
    )
].copy()


ranknet = ranknet.merge(

    benchmark[
        [
            'topic_id',
            'video_id',
            'relevance_binary'
        ]
    ],

    on=[
        'topic_id',
        'video_id'
    ],

    how='inner',

    validate='one_to_one'
)


ranknet[ranknet_score_column] = pd.to_numeric(
    ranknet[ranknet_score_column],
    errors='coerce'
)


if ranknet[
    ranknet_score_column
].isna().any():

    raise ValueError(
        'RankNet contains missing/non-numeric scores.'
    )


# ============================================================================
# COVERAGE CHECK FOR DPR/RANKNET
# ============================================================================

expected_pair_count = len(
    accessible_pairs
)


dpr_pairs = set(
    zip(
        dpr['topic_id'],
        dpr['video_id']
    )
)


ranknet_pairs = set(
    zip(
        ranknet['topic_id'],
        ranknet['video_id']
    )
)


if dpr_pairs != accessible_pairs:

    missing = (
        accessible_pairs
        -
        dpr_pairs
    )

    extra = (
        dpr_pairs
        -
        accessible_pairs
    )

    raise RuntimeError(

        'DPR does not cover exactly the accessible cohort.\n'

        f'Missing pairs: {len(missing)}\n'
        f'Extra pairs: {len(extra)}'
    )


if ranknet_pairs != accessible_pairs:

    missing = (
        accessible_pairs
        -
        ranknet_pairs
    )

    extra = (
        ranknet_pairs
        -
        accessible_pairs
    )

    raise RuntimeError(

        'RankNet does not cover exactly the accessible cohort.\n'

        f'Missing pairs: {len(missing)}\n'
        f'Extra pairs: {len(extra)}'
    )


print()
print(
    f'DPR common pairs    : {len(dpr_pairs):,}'
)

print(
    f'RankNet common pairs: {len(ranknet_pairs):,}'
)


# ============================================================================
# DPR TOPIC METRICS
# ============================================================================

dpr_topic = evaluate_prediction_dataframe(
    dpr,
    dpr_score_column
)


# ============================================================================
# RANKNET TOPIC METRICS
# ============================================================================

ranknet_topic = evaluate_prediction_dataframe(
    ranknet,
    ranknet_score_column
)


# ============================================================================
# RECOMPUTE BM25 + E5 ON THE SAME ACCESSIBLE COHORT
# ============================================================================
#
# The LAK benchmark's "content" field is used consistently for BM25/E5.
# ============================================================================

content_column = 'content'


if content_column not in benchmark.columns:

    raise ValueError(
        'Accessible LAK benchmark does not contain '
        '"content", required for BM25/E5.'
    )


benchmark['content'] = (
    benchmark['content']
    .fillna('')
    .astype(str)
)


# ============================================================================
# BM25
# ============================================================================

print()
print('=' * 100)
print('RECOMPUTING BM25 ON ACCESSIBLE COHORT')
print('=' * 100)


# One document per topic-video pair.
bm25_data = (
    benchmark[
        [
            'topic_id',
            'video_id',
            'topic',
            'content',
            'relevance_binary'
        ]
    ]
    .drop_duplicates(
        [
            'topic_id',
            'video_id'
        ]
    )
    .copy()
)


bm25_topic_rows = []


for topic_id, group in bm25_data.groupby(
    'topic_id',
    sort=True
):

    group = group.reset_index(
        drop=True
    )


    corpus_tokens = [

        re.findall(
            r'\b\w+\b',
            text.lower()
        )

        for text
        in group['content']
    ]


    bm25 = BM25Okapi(
        corpus_tokens
    )


    topic_text = (
        group['topic']
        .iloc[0]
    )


    query_tokens = re.findall(
        r'\b\w+\b',
        str(topic_text).lower()
    )


    scores = bm25.get_scores(
        query_tokens
    )


    group = group.copy()

    group['bm25_score'] = (
        np.asarray(
            scores,
            dtype=float
        )
    )


    metrics = evaluate_prediction_dataframe(
        group,
        'bm25_score'
    )


    bm25_topic_rows.append(
        metrics
    )


bm25_topic = pd.concat(
    bm25_topic_rows,
    ignore_index=True
)


# ============================================================================
# E5
# ============================================================================

print()
print('=' * 100)
print('RECOMPUTING E5 ON ACCESSIBLE COHORT')
print('=' * 100)


e5 = SentenceTransformer(
    'intfloat/e5-base-v2'
)


e5_topic_rows = []


for topic_id, group in benchmark.groupby(
    'topic_id',
    sort=True
):

    group = group.reset_index(
        drop=True
    )


    query_text = str(
        group['topic']
        .iloc[0]
    )


    contents = (
        group['content']
        .fillna('')
        .astype(str)
        .tolist()
    )


    query_embedding = e5.encode(

        'query: ' +
        query_text,

        normalize_embeddings=True,

        show_progress_bar=False
    )


    content_embeddings = e5.encode(

        [
            'passage: ' +
            x

            for x in contents
        ],

        normalize_embeddings=True,

        batch_size=32,

        show_progress_bar=False
    )


    scores = (
        content_embeddings
        .dot(
            query_embedding
        )
    )


    group = group.copy()

    group['e5_score'] = (
        np.asarray(
            scores,
            dtype=float
        )
    )


    metrics = evaluate_prediction_dataframe(
        group,
        'e5_score'
    )


    e5_topic_rows.append(
        metrics
    )


e5_topic = pd.concat(
    e5_topic_rows,
    ignore_index=True
)


# ============================================================================
# VALIDATE ALL TOPIC METRIC TABLES
# ============================================================================

for name, table in {

    'Hybrid':
        hybrid_topic,

    'DPR':
        dpr_topic,

    'RankNet':
        ranknet_topic,

    'E5':
        e5_topic,

    'BM25':
        bm25_topic

}.items():

    table['topic_id'] = (
        table['topic_id']
        .astype(str)
    )


    if set(
        table['topic_id']
    ) != benchmark_topics:

        raise RuntimeError(

            f'{name} does not contain exactly '
            f'the 54 accessible benchmark topics.'
        )


# ============================================================================
# SAVE SAME-COHORT TOPIC METRICS
# ============================================================================

hybrid_topic[
    ['topic_id'] +
    METRICS
].to_csv(

    OUT /
    'phase13_lak_hybrid_commoncohort_topic_metrics.csv',

    index=False
)


dpr_topic.to_csv(

    OUT /
    'phase13_lak_dpr_commoncohort_topic_metrics.csv',

    index=False
)


ranknet_topic.to_csv(

    OUT /
    'phase13_lak_ranknet_commoncohort_topic_metrics.csv',

    index=False
)


e5_topic.to_csv(

    OUT /
    'phase13_lak_e5_commoncohort_topic_metrics.csv',

    index=False
)


bm25_topic.to_csv(

    OUT /
    'phase13_lak_bm25_commoncohort_topic_metrics.csv',

    index=False
)


# ============================================================================
# STATISTICAL COMPARISON
# ============================================================================

baseline_tables = {

    'DPR':
        dpr_topic,

    'RankNet':
        ranknet_topic,

    'E5':
        e5_topic,

    'BM25':
        bm25_topic
}


rng = np.random.default_rng(
    42
)


N_BOOTSTRAP = 5000


wilcoxon_rows = []
bootstrap_rows = []


for baseline_name, baseline_table in baseline_tables.items():

    # ------------------------------------------------------------------------
    # Merge topic-level metrics.
    # ------------------------------------------------------------------------

    merged = (

        hybrid_topic[
            ['topic_id'] +
            METRICS
        ]

        .merge(

            baseline_table[
                ['topic_id'] +
                METRICS
            ],

            on='topic_id',

            how='inner',

            suffixes=(
                '_hybrid',
                '_baseline'
            )
        )
    )


    if len(merged) != 54:

        raise RuntimeError(

            f'{baseline_name}: '
            f'expected 54 matched topics, '
            f'got {len(merged)}.'
        )


    for metric in METRICS:

        x = (
            merged[
                metric +
                '_hybrid'
            ]
            .astype(float)
            .to_numpy()
        )


        y = (
            merged[
                metric +
                '_baseline'
            ]
            .astype(float)
            .to_numpy()
        )


        difference = (
            x -
            y
        )


        # --------------------------------------------------------------------
        # Wilcoxon signed-rank.
        # --------------------------------------------------------------------

        if np.allclose(
            difference,
            0.0
        ):

            statistic = 0.0
            p_value = 1.0

        else:

            try:

                statistic, p_value = (
                    wilcoxon(

                        x,
                        y,

                        zero_method='wilcox',

                        alternative='two-sided'
                    )
                )

            except ValueError:

                statistic = np.nan
                p_value = np.nan


        # --------------------------------------------------------------------
        # Bootstrap mean difference.
        # --------------------------------------------------------------------

        bootstrap_means = np.empty(
            N_BOOTSTRAP,
            dtype=float
        )


        n = len(
            difference
        )


        for i in range(
            N_BOOTSTRAP
        ):

            sample = rng.choice(

                difference,

                size=n,

                replace=True
            )


            bootstrap_means[i] = (
                np.mean(
                    sample
                )
            )


        ci_lower = float(
            np.percentile(
                bootstrap_means,
                2.5
            )
        )


        ci_upper = float(
            np.percentile(
                bootstrap_means,
                97.5
            )
        )


        mean_difference = float(
            np.mean(
                difference
            )
        )


        median_difference = float(
            np.median(
                difference
            )
        )


        # --------------------------------------------------------------------
        # Direction.
        # --------------------------------------------------------------------

        if mean_difference > 0:

            direction = (
                'Hybrid > '
                +
                baseline_name
            )

        elif mean_difference < 0:

            direction = (
                'Hybrid < '
                +
                baseline_name
            )

        else:

            direction = 'Equal'


        wilcoxon_rows.append({

            'comparison':
                'Full_Hybrid_vs_' +
                baseline_name,

            'baseline':
                baseline_name,

            'metric':
                metric,

            'topics':
                len(merged),

            'hybrid_mean':
                float(
                    np.mean(x)
                ),

            'baseline_mean':
                float(
                    np.mean(y)
                ),

            'mean_difference':
                mean_difference,

            'median_difference':
                median_difference,

            'wilcoxon_statistic':
                float(
                    statistic
                )
                if np.isfinite(
                    statistic
                )
                else np.nan,

            'p_value':
                float(
                    p_value
                )
                if np.isfinite(
                    p_value
                )
                else np.nan,

            'significant_p05':
                bool(
                    p_value < 0.05
                )
                if np.isfinite(
                    p_value
                )
                else False,

            'direction':
                direction
        })


        bootstrap_rows.append({

            'comparison':
                'Full_Hybrid_vs_' +
                baseline_name,

            'baseline':
                baseline_name,

            'metric':
                metric,

            'topics':
                len(merged),

            'mean_difference':
                mean_difference,

            'CI95_lower':
                ci_lower,

            'CI95_upper':
                ci_upper,

            'bootstrap_excludes_zero':
                bool(
                    (
                        ci_lower > 0
                        and
                        ci_upper > 0
                    )
                    or
                    (
                        ci_lower < 0
                        and
                        ci_upper < 0
                    )
                ),

            'direction':
                direction,

            'bootstrap_resamples':
                N_BOOTSTRAP,

            'random_seed':
                42
        })


# ============================================================================
# SAVE STATISTICAL RESULTS
# ============================================================================

significance = pd.DataFrame(
    wilcoxon_rows
)


bootstrap_ci = pd.DataFrame(
    bootstrap_rows
)


significance.to_csv(

    OUT /
    'phase13_lak_hybrid_vs_baselines_significance.csv',

    index=False
)


bootstrap_ci.to_csv(

    OUT /
    'phase13_lak_hybrid_bootstrap_ci.csv',

    index=False
)


# ============================================================================
# SUMMARY TABLE
# ============================================================================

summary_rows = []


for baseline_name, table in baseline_tables.items():

    merged = (

        hybrid_topic[
            ['topic_id'] +
            METRICS
        ]

        .merge(

            table[
                ['topic_id'] +
                METRICS
            ],

            on='topic_id',

            suffixes=(
                '_hybrid',
                '_baseline'
            )
        )
    )


    for metric in METRICS:

        summary_rows.append({

            'baseline':
                baseline_name,

            'metric':
                metric,

            'hybrid_mean':
                merged[
                    metric +
                    '_hybrid'
                ].mean(),

            'baseline_mean':
                merged[
                    metric +
                    '_baseline'
                ].mean(),

            'mean_difference':
                (
                    merged[
                        metric +
                        '_hybrid'
                    ]
                    -
                    merged[
                        metric +
                        '_baseline'
                    ]
                ).mean(),

            'topics':
                len(merged)
        })


comparison_summary = pd.DataFrame(
    summary_rows
)


comparison_summary.to_csv(

    OUT /
    'phase13_lak_hybrid_vs_baselines_summary.csv',

    index=False
)


# ============================================================================
# PRINT RESULTS
# ============================================================================

print()
print('=' * 120)
print('PHASE 13E — WILCOXON SIGNIFICANCE RESULTS')
print('=' * 120)

print()

print(
    significance.to_string(
        index=False,
        float_format=lambda x: f'{x:.6f}'
    )
)


print()
print('=' * 120)
print('PHASE 13E — 95% BOOTSTRAP CONFIDENCE INTERVALS')
print('=' * 120)

print()

print(
    bootstrap_ci.to_string(
        index=False,
        float_format=lambda x: f'{x:.6f}'
    )
)


# ============================================================================
# VALIDATION
# ============================================================================

print()
print('=' * 120)
print('PHASE 13E VALIDATION')
print('=' * 120)


expected_comparisons = (
    4 *
    len(METRICS)
)


checks = {

    'accessible_pairs':
        len(accessible_pairs) == 888,

    'accessible_topics':
        benchmark[
            'topic_id'
        ].nunique()
        == 54,

    'accessible_videos':
        benchmark[
            'video_id'
        ].nunique()
        == 841,

    'DPR_pairs':
        len(dpr_pairs) == 888,

    'RankNet_pairs':
        len(ranknet_pairs) == 888,

    'DPR_topics':
        len(dpr_topic) == 54,

    'RankNet_topics':
        len(ranknet_topic) == 54,

    'E5_topics':
        len(e5_topic) == 54,

    'BM25_topics':
        len(bm25_topic) == 54,

    'Wilcoxon_rows':
        len(significance)
        ==
        expected_comparisons,

    'Bootstrap_rows':
        len(bootstrap_ci)
        ==
        expected_comparisons,

    'all_wilcoxon_p_values_finite':
        significance[
            'p_value'
        ].notna().all(),

    'all_bootstrap_values_finite':
        np.isfinite(
            bootstrap_ci[
                [
                    'mean_difference',
                    'CI95_lower',
                    'CI95_upper'
                ]
            ]
            .to_numpy()
        ).all()
}


for name, passed in checks.items():

    print(
        f'{"PASS" if passed else "FAIL"} — {name}'
    )


if not all(
    checks.values()
):

    raise RuntimeError(
        'Phase 13E validation failed.'
    )


print()
print('=' * 120)
print('✅ PHASE 13E COMPLETED SUCCESSFULLY')
print('=' * 120)

print()
print(
    'All five ranking methods were compared on the same '
    '888-pair accessible LAK cohort.'
)

print()
print(
    'Saved:'
)

print(
    OUT /
    'phase13_lak_hybrid_vs_baselines_significance.csv'
)

print(
    OUT /
    'phase13_lak_hybrid_bootstrap_ci.csv'
)

print(
    OUT /
    'phase13_lak_hybrid_vs_baselines_summary.csv'
)
````

````python
# ============================================================================
# PHASE 13F — FINAL LAK HYBRID EXPERIMENTAL COMPLETENESS CHECK
# ============================================================================
#
# Verifies that the complete Phase 13 experimental package exists.
#
# Main groups:
#   A. Accessible LAK benchmark
#   B. Hybrid reconstruction
#   C. Hybrid evaluation
#   D. Component ablation
#   E. Baseline comparisons
#   F. Statistical validation
#   G. Feature coverage
#
# ============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# ============================================================================
# PATH
# ============================================================================

OUT = Path(
    '/content/revision_phase13'
)


# ============================================================================
# REQUIRED FILES
# ============================================================================

required_files = [

    # ------------------------------------------------------------------------
    # Accessible benchmark
    # ------------------------------------------------------------------------

    'lak2026_benchmark_eval_accessible.csv',

    'lak_youtube_metadata.csv',

    'lak_youtube_comments.csv',

    'lak_teaching_scores.csv',

    # ------------------------------------------------------------------------
    # Hybrid reconstruction
    # ------------------------------------------------------------------------

    'phase13_lak_hybrid_components.csv',

    'phase13_lak_hybrid_predictions.csv',

    'phase13_lak_hybrid_ranked_predictions.csv',

    'phase13_lak_hybrid_feature_coverage.csv',

    # ------------------------------------------------------------------------
    # Hybrid evaluation
    # ------------------------------------------------------------------------

    'phase13_lak_hybrid_topic_metrics.csv',

    'phase13_lak_hybrid_summary.csv',

    # ------------------------------------------------------------------------
    # Ablation
    # ------------------------------------------------------------------------

    'phase13_lak_hybrid_component_ablation.csv',

    'phase13_lak_hybrid_component_ablation_topic_metrics.csv',

    'phase13_lak_hybrid_component_ablation_comparison.csv',

    # ------------------------------------------------------------------------
    # Same-cohort baseline metrics
    # ------------------------------------------------------------------------

    'phase13_lak_dpr_commoncohort_topic_metrics.csv',

    'phase13_lak_ranknet_commoncohort_topic_metrics.csv',

    'phase13_lak_e5_commoncohort_topic_metrics.csv',

    'phase13_lak_bm25_commoncohort_topic_metrics.csv',

    'phase13_lak_hybrid_commoncohort_topic_metrics.csv',

    # ------------------------------------------------------------------------
    # Statistical comparison
    # ------------------------------------------------------------------------

    'phase13_lak_hybrid_vs_baselines_significance.csv',

    'phase13_lak_hybrid_bootstrap_ci.csv',

    'phase13_lak_hybrid_vs_baselines_summary.csv'
]


# ============================================================================
# FILE STATUS
# ============================================================================

status_rows = []


for filename in required_files:

    path = OUT / filename


    exists = path.exists()


    if exists:

        try:

            size_bytes = (
                path.stat()
                .st_size
            )

        except Exception:

            size_bytes = np.nan

    else:

        size_bytes = np.nan


    status_rows.append({

        'file':
            filename,

        'exists':
            exists,

        'size_bytes':
            size_bytes
    })


status = pd.DataFrame(
    status_rows
)


# ============================================================================
# SAVE STATUS
# ============================================================================

STATUS_FILE = (

    OUT /
    'phase13_lak_hybrid_completeness_check.csv'
)


status.to_csv(
    STATUS_FILE,
    index=False
)


# ============================================================================
# PRINT FILE STATUS
# ============================================================================

print('=' * 110)
print('PHASE 13F — FINAL COMPLETENESS CHECK')
print('=' * 110)

print()

for _, row in status.iterrows():

    symbol = (
        'PASS'
        if row['exists']
        else 'MISSING'
    )


    print(
        f'{symbol:<8} '
        f'{row["file"]}'
    )


# ============================================================================
# MISSING FILES
# ============================================================================

missing = (
    status.loc[
        ~status['exists'],
        'file'
    ]
    .tolist()
)


# ============================================================================
# ADDITIONAL CONTENT CHECKS
# ============================================================================

content_checks = {}


# ---------------------------------------------------------------------------
# Accessible benchmark
# ---------------------------------------------------------------------------

accessible_file = (
    OUT /
    'lak2026_benchmark_eval_accessible.csv'
)


if accessible_file.exists():

    accessible = pd.read_csv(
        accessible_file
    )


    content_checks[
        'accessible_rows_888'
    ] = (
        len(accessible)
        ==
        888
    )


    content_checks[
        'accessible_topics_54'
    ] = (
        accessible[
            'topic_id'
        ].nunique()
        ==
        54
    )


    content_checks[
        'accessible_videos_841'
    ] = (
        accessible[
            'video_id'
        ].nunique()
        ==
        841
    )


# ---------------------------------------------------------------------------
# Component table
# ---------------------------------------------------------------------------

component_file = (
    OUT /
    'phase13_lak_hybrid_components.csv'
)


if component_file.exists():

    components = pd.read_csv(
        component_file
    )


    component_columns = [

        'topic_id',
        'video_id',
        'alpha',
        'fes_full',
        'fes_no_sentiment',
        'fes_no_semantic',
        'intrinsic_quality_score',
        'depth_score',
        'lexical_diversity',
        'teaching_score_norm',
        'hybrid_score'
    ]


    content_checks[
        'component_rows_888'
    ] = (
        len(components)
        ==
        888
    )


    content_checks[
        'component_columns_complete'
    ] = all(
        c in components.columns
        for c in component_columns
    )


    content_checks[
        'component_pairs_unique'
    ] = not components.duplicated(
        [
            'topic_id',
            'video_id'
        ]
    ).any()


# ---------------------------------------------------------------------------
# Hybrid topic metrics
# ---------------------------------------------------------------------------

hybrid_metrics_file = (
    OUT /
    'phase13_lak_hybrid_topic_metrics.csv'
)


if hybrid_metrics_file.exists():

    hybrid_metrics = pd.read_csv(
        hybrid_metrics_file
    )


    content_checks[
        'hybrid_topics_54'
    ] = (
        len(hybrid_metrics)
        ==
        54
    )


# ---------------------------------------------------------------------------
# Ablation
# ---------------------------------------------------------------------------

ablation_file = (
    OUT /
    'phase13_lak_hybrid_component_ablation.csv'
)


if ablation_file.exists():

    ablation = pd.read_csv(
        ablation_file
    )


    expected_methods = {

        'Full_Hybrid',

        'Minus_Semantic',

        'Minus_Sentiment',

        'Minus_IQS',

        'Fixed_50_50'
    }


    actual_methods = set(
        ablation[
            'method'
        ]
        .astype(str)
    )


    content_checks[
        'five_ablation_methods'
    ] = (
        actual_methods
        ==
        expected_methods
    )


# ---------------------------------------------------------------------------
# Statistical results
# ---------------------------------------------------------------------------

significance_file = (
    OUT /
    'phase13_lak_hybrid_vs_baselines_significance.csv'
)


if significance_file.exists():

    significance = pd.read_csv(
        significance_file
    )


    content_checks[
        'statistical_rows_32'
    ] = (
        len(significance)
        ==
        32
    )


    content_checks[
        'statistical_baselines_four'
    ] = (
        set(
            significance[
                'baseline'
            ]
            .astype(str)
        )
        ==
        {
            'DPR',
            'RankNet',
            'E5',
            'BM25'
        }
    )


# ---------------------------------------------------------------------------
# Bootstrap results
# ---------------------------------------------------------------------------

bootstrap_file = (
    OUT /
    'phase13_lak_hybrid_bootstrap_ci.csv'
)


if bootstrap_file.exists():

    bootstrap = pd.read_csv(
        bootstrap_file
    )


    content_checks[
        'bootstrap_rows_32'
    ] = (
        len(bootstrap)
        ==
        32
    )


# ============================================================================
# PRINT CONTENT CHECKS
# ============================================================================

print()
print('=' * 110)
print('PHASE 13F — CONTENT VALIDATION')
print('=' * 110)

print()


for name, passed in content_checks.items():

    print(
        f'{"PASS" if passed else "FAIL"} — {name}'
    )


# ============================================================================
# FINAL STATUS
# ============================================================================

all_files_present = (
    len(missing)
    ==
    0
)


all_content_valid = (
    all(
        content_checks.values()
    )
    if content_checks
    else False
)


final_pass = (
    all_files_present
    and
    all_content_valid
)


print()
print('=' * 110)


if final_pass:

    print(
        '✅ ALL PHASE 13 LAK HYBRID OUTPUTS ARE PRESENT AND VALID.'
    )

else:

    print(
        '❌ PHASE 13 COMPLETENESS CHECK FAILED.'
    )


    if missing:

        print()
        print(
            'Missing files:'
        )


        for filename in missing:

            print(
                ' -',
                filename
            )


    failed_content = [

        name

        for name, passed
        in content_checks.items()

        if not passed
    ]


    if failed_content:

        print()
        print(
            'Failed content checks:'
        )


        for name in failed_content:

            print(
                ' -',
                name
            )


    raise RuntimeError(
        'Phase 13F completeness validation failed.'
    )


print('=' * 110)

print()
print(
    'Completeness report saved to:'
)

print(
    STATUS_FILE
)
````

````python
# ============================================================================
# CREATE CLEAN JOURNAL REVISION ZIP
# ============================================================================
# Excludes:
#   - sample_data
#   - db.sqlite3.zip
#   - .config
#   - drive
#   - previous collection folders
#   - previous ZIP files
#
# Keeps:
#   - lak2026_benchmark
#   - revision_phase9a
#   - revision_phase9b
#   - revision_phase9c
#   - revision_phase9d
#   - revision_phase9e
#   - revision_phase9f
#   - revision_phase9g
#   - revision_phase9h
#   - revision_phase10
#   - revision_phase11
#   - revision_phase12
#   - revision_phase13
#   - other journal-related files directly in /content
# ============================================================================

import os
import shutil
from google.colab import files


# ============================================================================
# PATHS
# ============================================================================

CONTENT_ROOT = "/content"

COLLECTION_ROOT = (
    "/content/"
    "JOURNAL_REVISION_CLEAN_COLLECTION"
)

ZIP_BASE = (
    "/content/"
    "JOURNAL_REVISION_CLEAN"
)


# ============================================================================
# REMOVE OLD COLLECTION
# ============================================================================

if os.path.exists(
    COLLECTION_ROOT
):

    print(
        "Removing previous clean collection..."
    )

    shutil.rmtree(
        COLLECTION_ROOT
    )


os.makedirs(
    COLLECTION_ROOT,
    exist_ok=True
)


# ============================================================================
# ITEMS TO EXCLUDE
# ============================================================================

EXCLUDE_NAMES = {

    # Unnecessary
    "sample_data",

    "db.sqlite3.zip",

    # Colab/system folders
    ".config",

    "drive",

    # Previous collection folders
    "JOURNAL_REVISION_ALL_OUTPUTSss_new_main",
    "JOURNAL_REVISION_CLEAN_COLLECTION",

    # Previous ZIPs
    "JOURNAL_REVISION_ALL_OUTPUTSs_new.zip",
    "JOURNAL_REVISION_ALL_OUTPUTSs.zip",
    "JOURNAL_REVISION_CLEAN.zip"
}


# ============================================================================
# COPY
# ============================================================================

print("=" * 90)
print("CREATING CLEAN JOURNAL REVISION COLLECTION")
print("=" * 90)

copied_items = []
failed_items = []


for item_name in sorted(
    os.listdir(
        CONTENT_ROOT
    )
):

    if item_name in EXCLUDE_NAMES:

        print(
            f"⏭️ Excluded: {item_name}"
        )

        continue


    source_path = os.path.join(
        CONTENT_ROOT,
        item_name
    )


    destination_path = os.path.join(
        COLLECTION_ROOT,
        item_name
    )


    try:

        # ------------------------------------------------------------
        # FOLDER
        # ------------------------------------------------------------

        if os.path.isdir(
            source_path
        ):

            shutil.copytree(

                source_path,

                destination_path,

                dirs_exist_ok=True
            )


            copied_items.append(
                item_name
            )


            print(
                f"📁 Copied: {item_name}"
            )


        # ------------------------------------------------------------
        # FILE
        # ------------------------------------------------------------

        elif os.path.isfile(
            source_path
        ):

            shutil.copy2(

                source_path,

                destination_path
            )


            copied_items.append(
                item_name
            )


            print(
                f"📄 Copied: {item_name}"
            )


    except Exception as e:

        failed_items.append(
            (
                item_name,
                str(e)
            )
        )


        print(
            f"❌ Failed: {item_name}"
        )

        print(
            f"   {e}"
        )


# ============================================================================
# INVENTORY
# ============================================================================

inventory_path = os.path.join(

    COLLECTION_ROOT,

    "OUTPUT_FILE_INVENTORY.txt"
)


total_files = 0
total_size = 0


with open(

    inventory_path,

    "w",

    encoding="utf-8"

) as f:

    f.write(
        "JOURNAL REVISION CLEAN OUTPUT INVENTORY\n"
    )

    f.write(
        "=" * 90
        +
        "\n\n"
    )


    for root, dirs, file_list in os.walk(
        COLLECTION_ROOT
    ):

        for filename in sorted(
            file_list
        ):

            if filename == (
                "OUTPUT_FILE_INVENTORY.txt"
            ):

                continue


            full_path = os.path.join(
                root,
                filename
            )


            relative_path = os.path.relpath(

                full_path,

                COLLECTION_ROOT
            )


            size_bytes = os.path.getsize(
                full_path
            )


            size_mb = (
                size_bytes
                /
                (1024 * 1024)
            )


            f.write(

                f"{relative_path} "
                f"({size_mb:.3f} MB)\n"
            )


            total_files += 1
            total_size += size_bytes


    f.write("\n")
    f.write("=" * 90 + "\n")

    f.write(
        f"Total files: {total_files}\n"
    )

    f.write(
        f"Total size: "
        f"{total_size / (1024 * 1024):.3f} MB\n"
    )


# ============================================================================
# SUMMARY
# ============================================================================

print()
print("=" * 90)
print("CLEAN COLLECTION SUMMARY")
print("=" * 90)

print(
    f"Items copied : {len(copied_items)}"
)

print(
    f"Items failed : {len(failed_items)}"
)

print(
    f"Files        : {total_files}"
)

print(
    f"Size         : "
    f"{total_size / (1024 * 1024):.2f} MB"
)


if failed_items:

    print()
    print("FAILED ITEMS:")

    for name, error in failed_items:

        print(
            f"❌ {name}: {error}"
        )


# ============================================================================
# CREATE ZIP
# ============================================================================

print()
print("=" * 90)
print("CREATING CLEAN ZIP")
print("=" * 90)


zip_path = shutil.make_archive(

    ZIP_BASE,

    "zip",

    COLLECTION_ROOT
)


zip_size_mb = (

    os.path.getsize(
        zip_path
    )
    /
    (1024 * 1024)
)


# ============================================================================
# FINAL
# ============================================================================

print()
print("=" * 90)
print("✅ CLEAN ZIP CREATED")
print("=" * 90)

print()
print(
    f"ZIP: {zip_path}"
)

print(
    f"ZIP size: {zip_size_mb:.2f} MB"
)

print(
    f"Files: {total_files}"
)


# ============================================================================
# DOWNLOAD
# ============================================================================

print()
print(
    "⬇️ Starting clean download..."
)

files.download(
    zip_path
)

print()
print(
    "✅ CLEAN DOWNLOAD STARTED"
)
````
